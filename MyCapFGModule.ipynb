```python
# -----------------------------------------------------------------------------
# CapFG‑enhanced Capsule Network with PSTanh activation for Fashion-MNIST
# -----------------------------------------------------------------------------

import os
# Enable expandable segments to reduce memory fragmentation
os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision.datasets import FashionMNIST
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import random
import copy

# -----------------------------------------------------------------------------
# Hyperparameters (adjusted for Fashion-MNIST)
# -----------------------------------------------------------------------------
BATCH_SIZE = 50                      # Reduced to fit GPU memory
NUM_CLASSES = 10
NUM_EPOCHS = 50                       # Maximum epochs
NUM_ROUTING_ITERATIONS = 3
IMAGE_SIZE = 32                        # Resized from 28 to 32
IMAGE_CHANNELS = 1                      # Grayscale
LEARNING_RATE = 0.0005
PATIENCE = 5                          # Early stopping patience (epochs)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

# -----------------------------------------------------------------------------
# PSTanh Activation Function (Parametric Scaled Hyperbolic Tangent)
# -----------------------------------------------------------------------------
class PSTanh(nn.Module):
    """
    Parametric Scaled Hyperbolic Tangent activation.
    f(x) = λ * x * (1 + tanh(β * x))
    λ and β are learnable parameters.
    """
    def __init__(self, lambda_init=0.5, beta_init=1.5):
        super().__init__()
        self.lambda_ = nn.Parameter(torch.tensor(lambda_init, dtype=torch.float32))
        self.beta = nn.Parameter(torch.tensor(beta_init, dtype=torch.float32))

    def forward(self, x):
        return self.lambda_ * x * (1 + torch.tanh(self.beta * x))

# -----------------------------------------------------------------------------
# Helper functions
# -----------------------------------------------------------------------------
def softmax(input, dim=1):
    """Custom softmax used in routing (from original paper)."""
    transposed = input.transpose(dim, len(input.size()) - 1)
    softmaxed = F.softmax(transposed.contiguous().view(-1, transposed.size(-1)), dim=-1)
    return softmaxed.view(*transposed.size()).transpose(dim, len(input.size()) - 1)

def augmentation(x, max_shift=2):
    """Random shift augmentation for Fashion-MNIST (works on grayscale)."""
    _, _, height, width = x.size()
    h_shift, w_shift = np.random.randint(-max_shift, max_shift + 1, size=2)
    src_h = slice(max(0, h_shift), h_shift + height)
    src_w = slice(max(0, w_shift), w_shift + width)
    tgt_h = slice(max(0, -h_shift), -h_shift + height)
    tgt_w = slice(max(0, -w_shift), -w_shift + width)
    shifted = torch.zeros_like(x)
    shifted[:, :, src_h, src_w] = x[:, :, tgt_h, tgt_w]
    return shifted.float()

def generate_adaptive_foreground_mask(images, device):
    """Generate a binary mask with a centered rectangle of random size per image."""
    batch_size = images.size(0)
    _, _, H, W = images.shape
    mask = torch.zeros(batch_size, 1, H, W, device=device)
    for i in range(batch_size):
        fg_h = random.randint(int(0.4 * H), int(0.9 * H))
        fg_w = random.randint(int(0.4 * W), int(0.9 * W))
        start_h = (H - fg_h) // 2
        start_w = (W - fg_w) // 2
        mask[i, :, start_h:start_h+fg_h, start_w:start_w+fg_w] = 1.0
    return mask

# -----------------------------------------------------------------------------
# Capsule Layer (standard)
# -----------------------------------------------------------------------------
class CapsuleLayer(nn.Module):
    def __init__(self, num_capsules, num_route_nodes, in_channels, out_channels,
                 kernel_size=None, stride=None, num_iterations=NUM_ROUTING_ITERATIONS):
        super(CapsuleLayer, self).__init__()
        self.num_route_nodes = num_route_nodes
        self.num_iterations = num_iterations
        self.num_capsules = num_capsules

        if num_route_nodes != -1:
            self.route_weights = nn.Parameter(
                torch.randn(num_capsules, num_route_nodes, in_channels, out_channels)
            )
        else:
            self.capsules = nn.ModuleList([
                nn.Conv2d(in_channels, out_channels, kernel_size=kernel_size, stride=stride, padding=0)
                for _ in range(num_capsules)
            ])

    def squash(self, tensor, dim=-1):
        squared_norm = (tensor ** 2).sum(dim=dim, keepdim=True)
        scale = squared_norm / (1 + squared_norm)
        return scale * tensor / torch.sqrt(squared_norm + 1e-8)

    def forward(self, x):
        if self.num_route_nodes != -1:
            # x shape: [batch, num_route_nodes, in_channels]
            priors = x[None, :, :, None, :] @ self.route_weights[:, None, :, :, :]
            logits = torch.zeros(*priors.size()).to(x.device)
            for i in range(self.num_iterations):
                probs = softmax(logits, dim=2)
                outputs = self.squash((probs * priors).sum(dim=2, keepdim=True))
                if i != self.num_iterations - 1:
                    delta_logits = (priors * outputs).sum(dim=-1, keepdim=True)
                    logits = logits + delta_logits
            return outputs.squeeze(2)   # [num_caps, batch, out_channels]
        else:
            outputs = [capsule(x).view(x.size(0), -1, 1) for capsule in self.capsules]
            outputs = torch.cat(outputs, dim=-1)
            outputs = self.squash(outputs)
            return outputs

# -----------------------------------------------------------------------------
# CapFG Module (Improved)
# -----------------------------------------------------------------------------
class CapFGModule(nn.Module):
    """
    Generates a soft spatial mask from primary capsules using:
    - Activation (norm across channels)
    - Agreement (inverse variance across channels)
    Mask values are clamped to [0.1, 1.0] to avoid complete suppression.
    """
    def __init__(self, channels_per_type=32, H=8, W=8, num_types=16):
        super().__init__()
        self.channels_per_type = channels_per_type
        self.H = H
        self.W = W
        self.num_types = num_types

        # Learnable mask generator: input has 2*num_types channels
        self.mask_conv = nn.Sequential(
            nn.Conv2d(2 * num_types, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 1, kernel_size=3, padding=1)
        )
        # Initialize bias of last layer to 2.0 so initial masks are near 0.88
        nn.init.constant_(self.mask_conv[-1].bias, 2.0)
        nn.init.kaiming_normal_(self.mask_conv[-1].weight)

    def forward(self, primary_caps, debug=False):
        B = primary_caps.size(0)
        # Reshape to [B, channels_per_type, H, W, num_types]
        caps = primary_caps.view(B, self.channels_per_type, self.H, self.W, self.num_types)

        # Activation (norm across channels) -> [B, H, W, num_types]
        activations = torch.norm(caps, dim=1)

        # Agreement: low variance across channels
        variance = caps.var(dim=1)                     # [B, H, W, num_types]
        agreement = 1.0 / (variance + 1e-8)            # [B, H, W, num_types]

        # Combine and prepare for CNN
        combined = torch.cat([activations, agreement], dim=-1)          # [B, H, W, 2*num_types]
        combined = combined.permute(0, 3, 1, 2)                          # [B, 2*num_types, H, W]

        raw_mask = self.mask_conv(combined)                              # [B, 1, H, W]
        mask = 0.1 + 0.9 * torch.sigmoid(raw_mask)                       # range [0.1, 1.0]

        if debug:
            print(f"[CapFG] raw_mask mean={raw_mask.mean().item():.3f}, "
                  f"mask min={mask.min().item():.3f}, max={mask.max().item():.3f}, "
                  f"mean={mask.mean().item():.3f}")

        # Apply mask (broadcast over capsule channels)
        mask = mask[:, :, :, :, None]                                    # [B, 1, H, W, 1]
        gated = caps * mask
        return gated.view(B, -1, self.num_types)

# -----------------------------------------------------------------------------
# Larger CapFG Capsule Network with PSTanh activation (adapted for grayscale)
# -----------------------------------------------------------------------------
class CapFGCapsuleNet(nn.Module):
    def __init__(self, use_checkpointing=False):
        super(CapFGCapsuleNet, self).__init__()
        self.use_checkpointing = use_checkpointing

        # Convolutional layers with PSTanh activation
        # First layer now accepts 1 channel (grayscale)
        self.conv1 = nn.Conv2d(1, 128, kernel_size=5, stride=1, padding=2)
        self.act1 = PSTanh(lambda_init=0.5, beta_init=1.5)

        self.conv2 = nn.Conv2d(128, 256, kernel_size=5, stride=1, padding=2)
        self.act2 = PSTanh(lambda_init=0.5, beta_init=1.5)

        self.conv3 = nn.Conv2d(256, 256, kernel_size=3, stride=2, padding=1)  # to 16x16
        self.act3 = PSTanh(lambda_init=0.5, beta_init=1.5)

        # Primary capsules: 16 types, each producing 32 channels
        self.primary_caps = CapsuleLayer(
            num_capsules=16, num_route_nodes=-1, in_channels=256, out_channels=32,
            kernel_size=3, stride=1
        )
        # After primary_caps, spatial size becomes (16-3+1)=14 -> H=W=14
        # Total capsules = 32 * 14 * 14 = 6272, each 16D

        self.capfg = CapFGModule(channels_per_type=32, H=14, W=14, num_types=16)

        # Digit capsules: 10 types, each 32D
        self.digit_caps = CapsuleLayer(
            num_capsules=NUM_CLASSES, num_route_nodes=6272, in_channels=16, out_channels=32
        )

        # Decoder with PSTanh in hidden layers
        # Output dimension now 32*32*1 = 1024 (grayscale)
        self.decoder = nn.Sequential(
            nn.Linear(32 * NUM_CLASSES, 1024),    # 32 * 10 = 320
            PSTanh(lambda_init=0.5, beta_init=1.5),
            nn.Linear(1024, 2048),
            PSTanh(lambda_init=0.5, beta_init=1.5),
            nn.Linear(2048, IMAGE_SIZE * IMAGE_SIZE * IMAGE_CHANNELS),   # 1024
            nn.Sigmoid()
        )

    def forward(self, x, y=None, mask=None):
        if mask is not None:
            x = x * mask

        x = self.act1(self.conv1(x))
        x = self.act2(self.conv2(x))
        x = self.act3(self.conv3(x))                 # [B, 256, 16, 16]

        # Optional gradient checkpointing to save memory
        if self.use_checkpointing and self.training:
            x = torch.utils.checkpoint.checkpoint(self.primary_caps, x)
        else:
            x = self.primary_caps(x)                  # [B, 6272, 16]

        x = self.capfg(x, debug=False)                # [B, 6272, 16]

        if self.use_checkpointing and self.training:
            x = torch.utils.checkpoint.checkpoint(self.digit_caps, x)
        else:
            x = self.digit_caps(x)                     # [10, B, 32]

        x = x.squeeze().transpose(0, 1)                 # [B, 10, 32]

        classes = (x ** 2).sum(dim=-1) ** 0.5           # [B, 10]
        classes = F.softmax(classes, dim=-1)

        if y is None:
            _, max_idx = classes.max(dim=1)
            y = torch.eye(NUM_CLASSES, device=x.device).index_select(dim=0, index=max_idx)

        reconstructions = self.decoder((x * y[:, :, None]).reshape(x.size(0), -1))
        return classes, reconstructions

# -----------------------------------------------------------------------------
# Loss function
# -----------------------------------------------------------------------------
class CapsuleLoss(nn.Module):
    def __init__(self):
        super(CapsuleLoss, self).__init__()
        self.reconstruction_loss = nn.MSELoss(reduction='sum')

    def forward(self, images, labels, classes, reconstructions):
        left = F.relu(0.9 - classes, inplace=True) ** 2
        right = F.relu(classes - 0.1, inplace=True) ** 2
        margin_loss = labels * left + 0.5 * (1. - labels) * right
        margin_loss = margin_loss.sum()

        images_flat = images.view(reconstructions.size(0), -1)
        recon_loss = self.reconstruction_loss(reconstructions, images_flat)

        return (margin_loss + 0.0005 * recon_loss) / images.size(0)

# -----------------------------------------------------------------------------
# Data loading for Fashion-MNIST
# -----------------------------------------------------------------------------
def get_fashion_mnist_loaders(batch_size=BATCH_SIZE):
    transform_train = transforms.Compose([
        transforms.Resize(IMAGE_SIZE),               # Resize 28 -> 32
        transforms.RandomHorizontalFlip(),            # Optional for fashion items
        transforms.ToTensor(),
    ])
    transform_test = transforms.Compose([
        transforms.Resize(IMAGE_SIZE),
        transforms.ToTensor(),
    ])

    trainset = FashionMNIST(root='./data', train=True, download=True, transform=transform_train)
    testset = FashionMNIST(root='./data', train=False, download=True, transform=transform_test)

    train_loader = DataLoader(trainset, batch_size=batch_size, shuffle=True, num_workers=2)
    test_loader = DataLoader(testset, batch_size=batch_size, shuffle=False, num_workers=2)
    return train_loader, test_loader

# -----------------------------------------------------------------------------
# Training function with early stopping (unchanged)
# -----------------------------------------------------------------------------
def train_model(model, train_loader, test_loader, num_epochs, patience):
    model = model.to(DEVICE)
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)
    criterion = CapsuleLoss()

    train_losses, test_losses = [], []
    train_accs, test_accs = [], []
    best_test_loss = float('inf')
    best_model_state = None
    epochs_no_improve = 0

    for epoch in range(1, num_epochs + 1):
        # -------------------- Training --------------------
        model.train()
        train_loss = 0.0
        correct = 0
        total = 0
        pbar = tqdm(train_loader, desc=f'Epoch {epoch}/{num_epochs} [Train]')
        for batch_idx, (data, target) in enumerate(pbar):
            data, target = data.to(DEVICE), target.to(DEVICE)
            target_onehot = torch.eye(NUM_CLASSES, device=DEVICE)[target]

            data = augmentation(data)
            mask = generate_adaptive_foreground_mask(data, DEVICE)

            classes, reconstructions = model(data, target_onehot, mask)
            loss = criterion(data, target_onehot, classes, reconstructions)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            pred = classes.argmax(dim=1)
            correct += pred.eq(target).sum().item()
            total += target.size(0)

            pbar.set_postfix({'loss': train_loss/(batch_idx+1), 'acc': 100.*correct/total})

        train_losses.append(train_loss / len(train_loader))
        train_accs.append(100. * correct / total)

        # -------------------- Validation --------------------
        model.eval()
        test_loss = 0.0
        correct = 0
        total = 0
        with torch.no_grad():
            for data, target in tqdm(test_loader, desc=f'Epoch {epoch}/{num_epochs} [Test]'):
                data, target = data.to(DEVICE), target.to(DEVICE)
                target_onehot = torch.eye(NUM_CLASSES, device=DEVICE)[target]

                mask = generate_adaptive_foreground_mask(data, DEVICE)
                classes, reconstructions = model(data, target_onehot, mask)
                loss = criterion(data, target_onehot, classes, reconstructions)

                test_loss += loss.item()
                pred = classes.argmax(dim=1)
                correct += pred.eq(target).sum().item()
                total += target.size(0)

        test_losses.append(test_loss / len(test_loader))
        test_acc = 100. * correct / total
        test_accs.append(test_acc)

        scheduler.step()

        print(f'Epoch {epoch}: Train Acc: {train_accs[-1]:.2f}%, Test Acc: {test_acc:.2f}%, Test Loss: {test_losses[-1]:.4f}')

        # Early stopping check
        if test_losses[-1] < best_test_loss:
            best_test_loss = test_losses[-1]
            best_model_state = copy.deepcopy(model.state_dict())
            epochs_no_improve = 0
            torch.save(model.state_dict(), 'best_capfg_pstanh_fashion.pth')
            print(f'  -> Best model saved (test loss: {best_test_loss:.4f})')
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f'Early stopping triggered after {epoch} epochs.')
                break

    # Load best model
    model.load_state_dict(best_model_state)
    return train_losses, test_losses, train_accs, test_accs

# -----------------------------------------------------------------------------
# Main execution
# -----------------------------------------------------------------------------
if __name__ == "__main__":
    print("Loading Fashion-MNIST data...")
    train_loader, test_loader = get_fashion_mnist_loaders(BATCH_SIZE)

    print("Initializing CapFG Capsule Network with PSTanh activation for Fashion-MNIST...")
    model = CapFGCapsuleNet(use_checkpointing=False)
    print(f"Total parameters: {sum(p.numel() for p in model.parameters())}")

    print("Starting training with early stopping...")
    train_losses, test_losses, train_accs, test_accs = train_model(
        model, train_loader, test_loader, NUM_EPOCHS, PATIENCE
    )

    # -------------------------------------------------------------------------
    # Plot training curves
    # -------------------------------------------------------------------------
    epochs = range(1, len(train_accs) + 1)
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_accs, 'b-', label='Train Accuracy')
    plt.plot(epochs, test_accs, 'r-', label='Test Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.title('Accuracy over Epochs')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_losses, 'b-', label='Train Loss')
    plt.plot(epochs, test_losses, 'r-', label='Test Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Loss over Epochs')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig('training_curves_pstanh_fashion.png')
    plt.show()
    print("Training complete. Best test accuracy: {:.2f}%".format(max(test_accs)))
    print("Best model saved as 'best_capfg_pstanh_fashion.pth'")
    print("Training curves saved as 'training_curves_pstanh_fashion.png'")
```

```python
# -----------------------------------------------------------------------------
# CapFG‑enhanced Capsule Network with PSTanh activation for Fashion-MNIST
# (Binary foreground masking integrated into the model)
# -----------------------------------------------------------------------------

import os
# Enable expandable segments to reduce memory fragmentation
os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision.datasets import FashionMNIST
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import random
import copy

# -----------------------------------------------------------------------------
# Hyperparameters (adjusted for Fashion-MNIST)
# -----------------------------------------------------------------------------
BATCH_SIZE = 50                      # Reduced to fit GPU memory
NUM_CLASSES = 10
NUM_EPOCHS = 50                       # Maximum epochs
NUM_ROUTING_ITERATIONS = 3
IMAGE_SIZE = 32                        # Resized from 28 to 32
IMAGE_CHANNELS = 1                      # Grayscale
LEARNING_RATE = 0.0005
PATIENCE = 5                          # Early stopping patience (epochs)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

# -----------------------------------------------------------------------------
# PSTanh Activation Function (Parametric Scaled Hyperbolic Tangent)
# -----------------------------------------------------------------------------
class PSTanh(nn.Module):
    """
    Parametric Scaled Hyperbolic Tangent activation.
    f(x) = λ * x * (1 + tanh(β * x))
    λ and β are learnable parameters.
    """
    def __init__(self, lambda_init=0.5, beta_init=1.5):
        super().__init__()
        self.lambda_ = nn.Parameter(torch.tensor(lambda_init, dtype=torch.float32))
        self.beta = nn.Parameter(torch.tensor(beta_init, dtype=torch.float32))

    def forward(self, x):
        return self.lambda_ * x * (1 + torch.tanh(self.beta * x))

# -----------------------------------------------------------------------------
# Helper functions
# -----------------------------------------------------------------------------
def softmax(input, dim=1):
    """Custom softmax used in routing (from original paper)."""
    transposed = input.transpose(dim, len(input.size()) - 1)
    softmaxed = F.softmax(transposed.contiguous().view(-1, transposed.size(-1)), dim=-1)
    return softmaxed.view(*transposed.size()).transpose(dim, len(input.size()) - 1)

def augmentation(x, max_shift=2):
    """Random shift augmentation for Fashion-MNIST (works on grayscale)."""
    _, _, height, width = x.size()
    h_shift, w_shift = np.random.randint(-max_shift, max_shift + 1, size=2)
    src_h = slice(max(0, h_shift), h_shift + height)
    src_w = slice(max(0, w_shift), w_shift + width)
    tgt_h = slice(max(0, -h_shift), -h_shift + height)
    tgt_w = slice(max(0, -w_shift), -w_shift + width)
    shifted = torch.zeros_like(x)
    shifted[:, :, src_h, src_w] = x[:, :, tgt_h, tgt_w]
    return shifted.float()

# -----------------------------------------------------------------------------
# Capsule Layer (standard)
# -----------------------------------------------------------------------------
class CapsuleLayer(nn.Module):
    def __init__(self, num_capsules, num_route_nodes, in_channels, out_channels,
                 kernel_size=None, stride=None, num_iterations=NUM_ROUTING_ITERATIONS):
        super(CapsuleLayer, self).__init__()
        self.num_route_nodes = num_route_nodes
        self.num_iterations = num_iterations
        self.num_capsules = num_capsules

        if num_route_nodes != -1:
            self.route_weights = nn.Parameter(
                torch.randn(num_capsules, num_route_nodes, in_channels, out_channels)
            )
        else:
            self.capsules = nn.ModuleList([
                nn.Conv2d(in_channels, out_channels, kernel_size=kernel_size, stride=stride, padding=0)
                for _ in range(num_capsules)
            ])

    def squash(self, tensor, dim=-1):
        squared_norm = (tensor ** 2).sum(dim=dim, keepdim=True)
        scale = squared_norm / (1 + squared_norm)
        return scale * tensor / torch.sqrt(squared_norm + 1e-8)

    def forward(self, x):
        if self.num_route_nodes != -1:
            # x shape: [batch, num_route_nodes, in_channels]
            priors = x[None, :, :, None, :] @ self.route_weights[:, None, :, :, :]
            logits = torch.zeros(*priors.size()).to(x.device)
            for i in range(self.num_iterations):
                probs = softmax(logits, dim=2)
                outputs = self.squash((probs * priors).sum(dim=2, keepdim=True))
                if i != self.num_iterations - 1:
                    delta_logits = (priors * outputs).sum(dim=-1, keepdim=True)
                    logits = logits + delta_logits
            return outputs.squeeze(2)   # [num_caps, batch, out_channels]
        else:
            outputs = [capsule(x).view(x.size(0), -1, 1) for capsule in self.capsules]
            outputs = torch.cat(outputs, dim=-1)
            outputs = self.squash(outputs)
            return outputs

# -----------------------------------------------------------------------------
# CapFG Module (Improved)
# -----------------------------------------------------------------------------
class CapFGModule(nn.Module):
    """
    Generates a soft spatial mask from primary capsules using:
    - Activation (norm across channels)
    - Agreement (inverse variance across channels)
    Mask values are clamped to [0.1, 1.0] to avoid complete suppression.
    """
    def __init__(self, channels_per_type=32, H=8, W=8, num_types=16):
        super().__init__()
        self.channels_per_type = channels_per_type
        self.H = H
        self.W = W
        self.num_types = num_types

        # Learnable mask generator: input has 2*num_types channels
        self.mask_conv = nn.Sequential(
            nn.Conv2d(2 * num_types, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 1, kernel_size=3, padding=1)
        )
        # Initialize bias of last layer to 2.0 so initial masks are near 0.88
        nn.init.constant_(self.mask_conv[-1].bias, 2.0)
        nn.init.kaiming_normal_(self.mask_conv[-1].weight)

    def forward(self, primary_caps, debug=False):
        B = primary_caps.size(0)
        # Reshape to [B, channels_per_type, H, W, num_types]
        caps = primary_caps.view(B, self.channels_per_type, self.H, self.W, self.num_types)

        # Activation (norm across channels) -> [B, H, W, num_types]
        activations = torch.norm(caps, dim=1)

        # Agreement: low variance across channels
        variance = caps.var(dim=1)                     # [B, H, W, num_types]
        agreement = 1.0 / (variance + 1e-8)            # [B, H, W, num_types]

        # Combine and prepare for CNN
        combined = torch.cat([activations, agreement], dim=-1)          # [B, H, W, 2*num_types]
        combined = combined.permute(0, 3, 1, 2)                          # [B, 2*num_types, H, W]

        raw_mask = self.mask_conv(combined)                              # [B, 1, H, W]
        mask = 0.1 + 0.9 * torch.sigmoid(raw_mask)                       # range [0.1, 1.0]

        if debug:
            print(f"[CapFG] raw_mask mean={raw_mask.mean().item():.3f}, "
                  f"mask min={mask.min().item():.3f}, max={mask.max().item():.3f}, "
                  f"mean={mask.mean().item():.3f}")

        # Apply mask (broadcast over capsule channels)
        mask = mask[:, :, :, :, None]                                    # [B, 1, H, W, 1]
        gated = caps * mask
        return gated.view(B, -1, self.num_types)

# -----------------------------------------------------------------------------
# Larger CapFG Capsule Network with PSTanh activation and integrated binary masking
# -----------------------------------------------------------------------------
class CapFGCapsuleNet(nn.Module):
    def __init__(self, use_checkpointing=False, binary_threshold=0.2):
        super(CapFGCapsuleNet, self).__init__()
        self.use_checkpointing = use_checkpointing
        self.binary_threshold = binary_threshold   # fixed threshold for binary mask

        # Convolutional layers with PSTanh activation
        self.conv1 = nn.Conv2d(1, 128, kernel_size=5, stride=1, padding=2)
        self.act1 = PSTanh(lambda_init=0.5, beta_init=1.5)

        self.conv2 = nn.Conv2d(128, 256, kernel_size=5, stride=1, padding=2)
        self.act2 = PSTanh(lambda_init=0.5, beta_init=1.5)

        self.conv3 = nn.Conv2d(256, 256, kernel_size=3, stride=2, padding=1)  # to 16x16
        self.act3 = PSTanh(lambda_init=0.5, beta_init=1.5)

        # Primary capsules: 16 types, each producing 32 channels
        self.primary_caps = CapsuleLayer(
            num_capsules=16, num_route_nodes=-1, in_channels=256, out_channels=32,
            kernel_size=3, stride=1
        )
        # After primary_caps, spatial size becomes (16-3+1)=14 -> H=W=14
        # Total capsules = 32 * 14 * 14 = 6272, each 16D

        self.capfg = CapFGModule(channels_per_type=32, H=14, W=14, num_types=16)

        # Digit capsules: 10 types, each 32D
        self.digit_caps = CapsuleLayer(
            num_capsules=NUM_CLASSES, num_route_nodes=6272, in_channels=16, out_channels=32
        )

        # Decoder with PSTanh in hidden layers
        self.decoder = nn.Sequential(
            nn.Linear(32 * NUM_CLASSES, 1024),    # 32 * 10 = 320
            PSTanh(lambda_init=0.5, beta_init=1.5),
            nn.Linear(1024, 2048),
            PSTanh(lambda_init=0.5, beta_init=1.5),
            nn.Linear(2048, IMAGE_SIZE * IMAGE_SIZE * IMAGE_CHANNELS),   # 1024
            nn.Sigmoid()
        )

    def _generate_binary_mask(self, x):
        """
        Generate a binary foreground mask from input image x.
        For Fashion‑MNIST (grayscale, values in [0,1]), we threshold at self.binary_threshold.
        Returns a mask of same shape as x, with values 0 or 1.
        """
        # x shape: [B, 1, H, W]
        mask = (x > self.binary_threshold).float()   # binary
        return mask

    def forward(self, x, y=None, external_mask=None):
        """
        x: input image [B, 1, H, W]
        y: one‑hot labels (optional)
        external_mask: if provided, use this mask instead of the generated binary mask
        """
        # ---------- Generate binary mask (or use external) ----------
        if external_mask is not None:
            mask = external_mask
        else:
            mask = self._generate_binary_mask(x)      # [B, 1, H, W] binary

        # Apply mask to input
        x_masked = x * mask

        # ---------- Feature extraction ----------
        x = self.act1(self.conv1(x_masked))
        x = self.act2(self.conv2(x))
        x = self.act3(self.conv3(x))                 # [B, 256, 16, 16]

        # Optional gradient checkpointing to save memory
        if self.use_checkpointing and self.training:
            x = torch.utils.checkpoint.checkpoint(self.primary_caps, x)
        else:
            x = self.primary_caps(x)                  # [B, 6272, 16]

        x = self.capfg(x, debug=False)                # [B, 6272, 16]

        if self.use_checkpointing and self.training:
            x = torch.utils.checkpoint.checkpoint(self.digit_caps, x)
        else:
            x = self.digit_caps(x)                     # [10, B, 32]

        x = x.squeeze().transpose(0, 1)                 # [B, 10, 32]

        # ---------- Class probabilities ----------
        classes = (x ** 2).sum(dim=-1) ** 0.5           # [B, 10]
        classes = F.softmax(classes, dim=-1)

        # ---------- Reconstruction ----------
        if y is None:
            _, max_idx = classes.max(dim=1)
            y = torch.eye(NUM_CLASSES, device=x.device).index_select(dim=0, index=max_idx)

        reconstructions = self.decoder((x * y[:, :, None]).reshape(x.size(0), -1))

        return classes, reconstructions, mask   # also return mask for optional visualization

# -----------------------------------------------------------------------------
# Loss function
# -----------------------------------------------------------------------------
class CapsuleLoss(nn.Module):
    def __init__(self):
        super(CapsuleLoss, self).__init__()
        self.reconstruction_loss = nn.MSELoss(reduction='sum')

    def forward(self, images, labels, classes, reconstructions):
        left = F.relu(0.9 - classes, inplace=True) ** 2
        right = F.relu(classes - 0.1, inplace=True) ** 2
        margin_loss = labels * left + 0.5 * (1. - labels) * right
        margin_loss = margin_loss.sum()

        images_flat = images.view(reconstructions.size(0), -1)
        recon_loss = self.reconstruction_loss(reconstructions, images_flat)

        return (margin_loss + 0.0005 * recon_loss) / images.size(0)

# -----------------------------------------------------------------------------
# Data loading for Fashion-MNIST
# -----------------------------------------------------------------------------
def get_fashion_mnist_loaders(batch_size=BATCH_SIZE):
    transform_train = transforms.Compose([
        transforms.Resize(IMAGE_SIZE),               # Resize 28 -> 32
        transforms.RandomHorizontalFlip(),            # Optional for fashion items
        transforms.ToTensor(),
    ])
    transform_test = transforms.Compose([
        transforms.Resize(IMAGE_SIZE),
        transforms.ToTensor(),
    ])

    trainset = FashionMNIST(root='./data', train=True, download=True, transform=transform_train)
    testset = FashionMNIST(root='./data', train=False, download=True, transform=transform_test)

    train_loader = DataLoader(trainset, batch_size=batch_size, shuffle=True, num_workers=2)
    test_loader = DataLoader(testset, batch_size=batch_size, shuffle=False, num_workers=2)
    return train_loader, test_loader

# -----------------------------------------------------------------------------
# Training function with early stopping (updated to use internal masking)
# -----------------------------------------------------------------------------
def train_model(model, train_loader, test_loader, num_epochs, patience):
    model = model.to(DEVICE)
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)
    criterion = CapsuleLoss()

    train_losses, test_losses = [], []
    train_accs, test_accs = [], []
    best_test_loss = float('inf')
    best_model_state = None
    epochs_no_improve = 0

    for epoch in range(1, num_epochs + 1):
        # -------------------- Training --------------------
        model.train()
        train_loss = 0.0
        correct = 0
        total = 0
        pbar = tqdm(train_loader, desc=f'Epoch {epoch}/{num_epochs} [Train]')
        for batch_idx, (data, target) in enumerate(pbar):
            data, target = data.to(DEVICE), target.to(DEVICE)
            target_onehot = torch.eye(NUM_CLASSES, device=DEVICE)[target]

            # Apply data augmentation (shifts only; no external mask needed)
            data = augmentation(data)

            # Forward pass: model generates its own binary mask internally
            classes, reconstructions, _ = model(data, target_onehot)
            loss = criterion(data, target_onehot, classes, reconstructions)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            pred = classes.argmax(dim=1)
            correct += pred.eq(target).sum().item()
            total += target.size(0)

            pbar.set_postfix({'loss': train_loss/(batch_idx+1), 'acc': 100.*correct/total})

        train_losses.append(train_loss / len(train_loader))
        train_accs.append(100. * correct / total)

        # -------------------- Validation --------------------
        model.eval()
        test_loss = 0.0
        correct = 0
        total = 0
        with torch.no_grad():
            for data, target in tqdm(test_loader, desc=f'Epoch {epoch}/{num_epochs} [Test]'):
                data, target = data.to(DEVICE), target.to(DEVICE)
                target_onehot = torch.eye(NUM_CLASSES, device=DEVICE)[target]

                # No external mask needed
                classes, reconstructions, _ = model(data, target_onehot)
                loss = criterion(data, target_onehot, classes, reconstructions)

                test_loss += loss.item()
                pred = classes.argmax(dim=1)
                correct += pred.eq(target).sum().item()
                total += target.size(0)

        test_losses.append(test_loss / len(test_loader))
        test_acc = 100. * correct / total
        test_accs.append(test_acc)

        scheduler.step()

        print(f'Epoch {epoch}: Train Acc: {train_accs[-1]:.2f}%, Test Acc: {test_acc:.2f}%, Test Loss: {test_losses[-1]:.4f}')

        # Early stopping check
        if test_losses[-1] < best_test_loss:
            best_test_loss = test_losses[-1]
            best_model_state = copy.deepcopy(model.state_dict())
            epochs_no_improve = 0
            torch.save(model.state_dict(), 'best_capfg_pstanh_fashion.pth')
            print(f'  -> Best model saved (test loss: {best_test_loss:.4f})')
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f'Early stopping triggered after {epoch} epochs.')
                break

    # Load best model
    model.load_state_dict(best_model_state)
    return train_losses, test_losses, train_accs, test_accs

# -----------------------------------------------------------------------------
# Main execution
# -----------------------------------------------------------------------------
if __name__ == "__main__":
    print("Loading Fashion-MNIST data...")
    train_loader, test_loader = get_fashion_mnist_loaders(BATCH_SIZE)

    print("Initializing CapFG Capsule Network with PSTanh activation for Fashion-MNIST...")
    model = CapFGCapsuleNet(use_checkpointing=False, binary_threshold=0.2)
    print(f"Total parameters: {sum(p.numel() for p in model.parameters())}")

    print("Starting training with early stopping...")
    train_losses, test_losses, train_accs, test_accs = train_model(
        model, train_loader, test_loader, NUM_EPOCHS, PATIENCE
    )

    # -------------------------------------------------------------------------
    # Plot training curves
    # -------------------------------------------------------------------------
    epochs = range(1, len(train_accs) + 1)
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_accs, 'b-', label='Train Accuracy')
    plt.plot(epochs, test_accs, 'r-', label='Test Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.title('Accuracy over Epochs')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_losses, 'b-', label='Train Loss')
    plt.plot(epochs, test_losses, 'r-', label='Test Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Loss over Epochs')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig('training_curves_pstanh_fashion.png')
    plt.show()
    print("Training complete. Best test accuracy: {:.2f}%".format(max(test_accs)))
    print("Best model saved as 'best_capfg_pstanh_fashion.pth'")
    print("Training curves saved as 'training_curves_pstanh_fashion.png'")
```

```python
# -----------------------------------------------------------------------------
# CapFG‑enhanced Capsule Network with PSTanh activation for CIFAR-10
# (Binary foreground masking integrated into the model)
# -----------------------------------------------------------------------------

import os
# Enable expandable segments to reduce memory fragmentation
os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision.datasets import CIFAR10                     # <-- changed
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import random
import copy

# -----------------------------------------------------------------------------
# Hyperparameters (adjusted for CIFAR-10)
# -----------------------------------------------------------------------------
BATCH_SIZE = 50                      # Reduced to fit GPU memory
NUM_CLASSES = 10
NUM_EPOCHS = 50                       # Maximum epochs
NUM_ROUTING_ITERATIONS = 3
IMAGE_SIZE = 32                        # CIFAR-10 is already 32x32
IMAGE_CHANNELS = 3                      # <-- changed from 1 to 3 (RGB)
LEARNING_RATE = 0.0005
PATIENCE = 5                          # Early stopping patience (epochs)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

# -----------------------------------------------------------------------------
# PSTanh Activation Function (Parametric Scaled Hyperbolic Tangent)
# -----------------------------------------------------------------------------
class PSTanh(nn.Module):
    """
    Parametric Scaled Hyperbolic Tangent activation.
    f(x) = λ * x * (1 + tanh(β * x))
    λ and β are learnable parameters.
    """
    def __init__(self, lambda_init=0.5, beta_init=1.5):
        super().__init__()
        self.lambda_ = nn.Parameter(torch.tensor(lambda_init, dtype=torch.float32))
        self.beta = nn.Parameter(torch.tensor(beta_init, dtype=torch.float32))

    def forward(self, x):
        return self.lambda_ * x * (1 + torch.tanh(self.beta * x))

# -----------------------------------------------------------------------------
# Helper functions
# -----------------------------------------------------------------------------
def softmax(input, dim=1):
    """Custom softmax used in routing (from original paper)."""
    transposed = input.transpose(dim, len(input.size()) - 1)
    softmaxed = F.softmax(transposed.contiguous().view(-1, transposed.size(-1)), dim=-1)
    return softmaxed.view(*transposed.size()).transpose(dim, len(input.size()) - 1)

def augmentation(x, max_shift=2):
    """Random shift augmentation (works for any number of channels)."""
    _, _, height, width = x.size()
    h_shift, w_shift = np.random.randint(-max_shift, max_shift + 1, size=2)
    src_h = slice(max(0, h_shift), h_shift + height)
    src_w = slice(max(0, w_shift), w_shift + width)
    tgt_h = slice(max(0, -h_shift), -h_shift + height)
    tgt_w = slice(max(0, -w_shift), -w_shift + width)
    shifted = torch.zeros_like(x)
    shifted[:, :, src_h, src_w] = x[:, :, tgt_h, tgt_w]
    return shifted.float()

# -----------------------------------------------------------------------------
# Capsule Layer (standard)
# -----------------------------------------------------------------------------
class CapsuleLayer(nn.Module):
    def __init__(self, num_capsules, num_route_nodes, in_channels, out_channels,
                 kernel_size=None, stride=None, num_iterations=NUM_ROUTING_ITERATIONS):
        super(CapsuleLayer, self).__init__()
        self.num_route_nodes = num_route_nodes
        self.num_iterations = num_iterations
        self.num_capsules = num_capsules

        if num_route_nodes != -1:
            self.route_weights = nn.Parameter(
                torch.randn(num_capsules, num_route_nodes, in_channels, out_channels)
            )
        else:
            self.capsules = nn.ModuleList([
                nn.Conv2d(in_channels, out_channels, kernel_size=kernel_size, stride=stride, padding=0)
                for _ in range(num_capsules)
            ])

    def squash(self, tensor, dim=-1):
        squared_norm = (tensor ** 2).sum(dim=dim, keepdim=True)
        scale = squared_norm / (1 + squared_norm)
        return scale * tensor / torch.sqrt(squared_norm + 1e-8)

    def forward(self, x):
        if self.num_route_nodes != -1:
            # x shape: [batch, num_route_nodes, in_channels]
            priors = x[None, :, :, None, :] @ self.route_weights[:, None, :, :, :]
            logits = torch.zeros(*priors.size()).to(x.device)
            for i in range(self.num_iterations):
                probs = softmax(logits, dim=2)
                outputs = self.squash((probs * priors).sum(dim=2, keepdim=True))
                if i != self.num_iterations - 1:
                    delta_logits = (priors * outputs).sum(dim=-1, keepdim=True)
                    logits = logits + delta_logits
            return outputs.squeeze(2)   # [num_caps, batch, out_channels]
        else:
            outputs = [capsule(x).view(x.size(0), -1, 1) for capsule in self.capsules]
            outputs = torch.cat(outputs, dim=-1)
            outputs = self.squash(outputs)
            return outputs

# -----------------------------------------------------------------------------
# CapFG Module (Improved)
# -----------------------------------------------------------------------------
class CapFGModule(nn.Module):
    """
    Generates a soft spatial mask from primary capsules using:
    - Activation (norm across channels)
    - Agreement (inverse variance across channels)
    Mask values are clamped to [0.1, 1.0] to avoid complete suppression.
    """
    def __init__(self, channels_per_type=32, H=8, W=8, num_types=16):
        super().__init__()
        self.channels_per_type = channels_per_type
        self.H = H
        self.W = W
        self.num_types = num_types

        # Learnable mask generator: input has 2*num_types channels
        self.mask_conv = nn.Sequential(
            nn.Conv2d(2 * num_types, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 1, kernel_size=3, padding=1)
        )
        # Initialize bias of last layer to 2.0 so initial masks are near 0.88
        nn.init.constant_(self.mask_conv[-1].bias, 2.0)
        nn.init.kaiming_normal_(self.mask_conv[-1].weight)

    def forward(self, primary_caps, debug=False):
        B = primary_caps.size(0)
        # Reshape to [B, channels_per_type, H, W, num_types]
        caps = primary_caps.view(B, self.channels_per_type, self.H, self.W, self.num_types)

        # Activation (norm across channels) -> [B, H, W, num_types]
        activations = torch.norm(caps, dim=1)

        # Agreement: low variance across channels
        variance = caps.var(dim=1)                     # [B, H, W, num_types]
        agreement = 1.0 / (variance + 1e-8)            # [B, H, W, num_types]

        # Combine and prepare for CNN
        combined = torch.cat([activations, agreement], dim=-1)          # [B, H, W, 2*num_types]
        combined = combined.permute(0, 3, 1, 2)                          # [B, 2*num_types, H, W]

        raw_mask = self.mask_conv(combined)                              # [B, 1, H, W]
        mask = 0.1 + 0.9 * torch.sigmoid(raw_mask)                       # range [0.1, 1.0]

        if debug:
            print(f"[CapFG] raw_mask mean={raw_mask.mean().item():.3f}, "
                  f"mask min={mask.min().item():.3f}, max={mask.max().item():.3f}, "
                  f"mean={mask.mean().item():.3f}")

        # Apply mask (broadcast over capsule channels)
        mask = mask[:, :, :, :, None]                                    # [B, 1, H, W, 1]
        gated = caps * mask
        return gated.view(B, -1, self.num_types)

# -----------------------------------------------------------------------------
# Larger CapFG Capsule Network with PSTanh activation and integrated binary masking
# -----------------------------------------------------------------------------
class CapFGCapsuleNet(nn.Module):
    def __init__(self, use_checkpointing=False, binary_threshold=0.2):
        super(CapFGCapsuleNet, self).__init__()
        self.use_checkpointing = use_checkpointing
        self.binary_threshold = binary_threshold   # fixed threshold for binary mask

        # Convolutional layers with PSTanh activation
        self.conv1 = nn.Conv2d(IMAGE_CHANNELS, 128, kernel_size=5, stride=1, padding=2)   # <-- input channels changed
        self.act1 = PSTanh(lambda_init=0.5, beta_init=1.5)

        self.conv2 = nn.Conv2d(128, 256, kernel_size=5, stride=1, padding=2)
        self.act2 = PSTanh(lambda_init=0.5, beta_init=1.5)

        self.conv3 = nn.Conv2d(256, 256, kernel_size=3, stride=2, padding=1)  # to 16x16
        self.act3 = PSTanh(lambda_init=0.5, beta_init=1.5)

        # Primary capsules: 16 types, each producing 32 channels
        self.primary_caps = CapsuleLayer(
            num_capsules=16, num_route_nodes=-1, in_channels=256, out_channels=32,
            kernel_size=3, stride=1
        )
        # After primary_caps, spatial size becomes (16-3+1)=14 -> H=W=14
        # Total capsules = 32 * 14 * 14 = 6272, each 16D

        self.capfg = CapFGModule(channels_per_type=32, H=14, W=14, num_types=16)

        # Digit capsules: 10 types, each 32D
        self.digit_caps = CapsuleLayer(
            num_capsules=NUM_CLASSES, num_route_nodes=6272, in_channels=16, out_channels=32
        )

        # Decoder with PSTanh in hidden layers
        self.decoder = nn.Sequential(
            nn.Linear(32 * NUM_CLASSES, 1024),    # 32 * 10 = 320
            PSTanh(lambda_init=0.5, beta_init=1.5),
            nn.Linear(1024, 2048),
            PSTanh(lambda_init=0.5, beta_init=1.5),
            nn.Linear(2048, IMAGE_SIZE * IMAGE_SIZE * IMAGE_CHANNELS),   # <-- changed to 3072
            nn.Sigmoid()
        )

    def _generate_binary_mask(self, x):
        """
        Generate a binary foreground mask from input image x.
        For CIFAR‑10 (RGB, values in [0,1]), we threshold at self.binary_threshold.
        Returns a mask of same shape as x, with values 0 or 1.
        """
        # x shape: [B, 3, H, W]
        mask = (x > self.binary_threshold).float()   # binary per channel
        return mask

    def forward(self, x, y=None, external_mask=None):
        """
        x: input image [B, 3, H, W]
        y: one‑hot labels (optional)
        external_mask: if provided, use this mask instead of the generated binary mask
        """
        # ---------- Generate binary mask (or use external) ----------
        if external_mask is not None:
            mask = external_mask
        else:
            mask = self._generate_binary_mask(x)      # [B, 3, H, W] binary

        # Apply mask to input
        x_masked = x * mask

        # ---------- Feature extraction ----------
        x = self.act1(self.conv1(x_masked))
        x = self.act2(self.conv2(x))
        x = self.act3(self.conv3(x))                 # [B, 256, 16, 16]

        # Optional gradient checkpointing to save memory
        if self.use_checkpointing and self.training:
            x = torch.utils.checkpoint.checkpoint(self.primary_caps, x)
        else:
            x = self.primary_caps(x)                  # [B, 6272, 16]

        x = self.capfg(x, debug=False)                # [B, 6272, 16]

        if self.use_checkpointing and self.training:
            x = torch.utils.checkpoint.checkpoint(self.digit_caps, x)
        else:
            x = self.digit_caps(x)                     # [10, B, 32]

        x = x.squeeze().transpose(0, 1)                 # [B, 10, 32]

        # ---------- Class probabilities ----------
        classes = (x ** 2).sum(dim=-1) ** 0.5           # [B, 10]
        classes = F.softmax(classes, dim=-1)

        # ---------- Reconstruction ----------
        if y is None:
            _, max_idx = classes.max(dim=1)
            y = torch.eye(NUM_CLASSES, device=x.device).index_select(dim=0, index=max_idx)

        reconstructions = self.decoder((x * y[:, :, None]).reshape(x.size(0), -1))

        return classes, reconstructions, mask   # also return mask for optional visualization

# -----------------------------------------------------------------------------
# Loss function
# -----------------------------------------------------------------------------
class CapsuleLoss(nn.Module):
    def __init__(self):
        super(CapsuleLoss, self).__init__()
        self.reconstruction_loss = nn.MSELoss(reduction='sum')

    def forward(self, images, labels, classes, reconstructions):
        left = F.relu(0.9 - classes, inplace=True) ** 2
        right = F.relu(classes - 0.1, inplace=True) ** 2
        margin_loss = labels * left + 0.5 * (1. - labels) * right
        margin_loss = margin_loss.sum()

        images_flat = images.view(reconstructions.size(0), -1)
        recon_loss = self.reconstruction_loss(reconstructions, images_flat)

        return (margin_loss + 0.0005 * recon_loss) / images.size(0)

# -----------------------------------------------------------------------------
# Data loading for CIFAR-10
# -----------------------------------------------------------------------------
def get_cifar10_loaders(batch_size=BATCH_SIZE):
    transform_train = transforms.Compose([
        transforms.Resize(IMAGE_SIZE),               # already 32, but safe
        transforms.RandomHorizontalFlip(),            # common for CIFAR-10
        transforms.ToTensor(),
    ])
    transform_test = transforms.Compose([
        transforms.Resize(IMAGE_SIZE),
        transforms.ToTensor(),
    ])

    trainset = CIFAR10(root='./data', train=True, download=True, transform=transform_train)
    testset = CIFAR10(root='./data', train=False, download=True, transform=transform_test)

    train_loader = DataLoader(trainset, batch_size=batch_size, shuffle=True, num_workers=2)
    test_loader = DataLoader(testset, batch_size=batch_size, shuffle=False, num_workers=2)
    return train_loader, test_loader

# -----------------------------------------------------------------------------
# Training function with early stopping (updated to use internal masking)
# -----------------------------------------------------------------------------
def train_model(model, train_loader, test_loader, num_epochs, patience):
    model = model.to(DEVICE)
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)
    criterion = CapsuleLoss()

    train_losses, test_losses = [], []
    train_accs, test_accs = [], []
    best_test_loss = float('inf')
    best_model_state = None
    epochs_no_improve = 0

    for epoch in range(1, num_epochs + 1):
        # -------------------- Training --------------------
        model.train()
        train_loss = 0.0
        correct = 0
        total = 0
        pbar = tqdm(train_loader, desc=f'Epoch {epoch}/{num_epochs} [Train]')
        for batch_idx, (data, target) in enumerate(pbar):
            data, target = data.to(DEVICE), target.to(DEVICE)
            target_onehot = torch.eye(NUM_CLASSES, device=DEVICE)[target]

            # Apply data augmentation (shifts only; no external mask needed)
            data = augmentation(data)

            # Forward pass: model generates its own binary mask internally
            classes, reconstructions, _ = model(data, target_onehot)
            loss = criterion(data, target_onehot, classes, reconstructions)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            pred = classes.argmax(dim=1)
            correct += pred.eq(target).sum().item()
            total += target.size(0)

            pbar.set_postfix({'loss': train_loss/(batch_idx+1), 'acc': 100.*correct/total})

        train_losses.append(train_loss / len(train_loader))
        train_accs.append(100. * correct / total)

        # -------------------- Validation --------------------
        model.eval()
        test_loss = 0.0
        correct = 0
        total = 0
        with torch.no_grad():
            for data, target in tqdm(test_loader, desc=f'Epoch {epoch}/{num_epochs} [Test]'):
                data, target = data.to(DEVICE), target.to(DEVICE)
                target_onehot = torch.eye(NUM_CLASSES, device=DEVICE)[target]

                # No external mask needed
                classes, reconstructions, _ = model(data, target_onehot)
                loss = criterion(data, target_onehot, classes, reconstructions)

                test_loss += loss.item()
                pred = classes.argmax(dim=1)
                correct += pred.eq(target).sum().item()
                total += target.size(0)

        test_losses.append(test_loss / len(test_loader))
        test_acc = 100. * correct / total
        test_accs.append(test_acc)

        scheduler.step()

        print(f'Epoch {epoch}: Train Acc: {train_accs[-1]:.2f}%, Test Acc: {test_acc:.2f}%, Test Loss: {test_losses[-1]:.4f}')

        # Early stopping check
        if test_losses[-1] < best_test_loss:
            best_test_loss = test_losses[-1]
            best_model_state = copy.deepcopy(model.state_dict())
            epochs_no_improve = 0
            torch.save(model.state_dict(), 'best_capfg_pstanh_cifar10.pth')   # <-- filename changed
            print(f'  -> Best model saved (test loss: {best_test_loss:.4f})')
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f'Early stopping triggered after {epoch} epochs.')
                break

    # Load best model
    model.load_state_dict(best_model_state)
    return train_losses, test_losses, train_accs, test_accs

# -----------------------------------------------------------------------------
# Main execution
# -----------------------------------------------------------------------------
if __name__ == "__main__":
    print("Loading CIFAR-10 data...")
    train_loader, test_loader = get_cifar10_loaders(BATCH_SIZE)

    print("Initializing CapFG Capsule Network with PSTanh activation for CIFAR-10...")
    model = CapFGCapsuleNet(use_checkpointing=False, binary_threshold=0.2)
    print(f"Total parameters: {sum(p.numel() for p in model.parameters())}")

    print("Starting training with early stopping...")
    train_losses, test_losses, train_accs, test_accs = train_model(
        model, train_loader, test_loader, NUM_EPOCHS, PATIENCE
    )

    # -------------------------------------------------------------------------
    # Plot training curves
    # -------------------------------------------------------------------------
    epochs = range(1, len(train_accs) + 1)
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_accs, 'b-', label='Train Accuracy')
    plt.plot(epochs, test_accs, 'r-', label='Test Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.title('Accuracy over Epochs')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_losses, 'b-', label='Train Loss')
    plt.plot(epochs, test_losses, 'r-', label='Test Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Loss over Epochs')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig('training_curves_pstanh_cifar10.png')   # <-- filename changed
    plt.show()
    print("Training complete. Best test accuracy: {:.2f}%".format(max(test_accs)))
    print("Best model saved as 'best_capfg_pstanh_cifar10.pth'")
    print("Training curves saved as 'training_curves_pstanh_cifar10.png'")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from tqdm import tqdm

# ---------------------------
# PSTanh Activation (as before)
# ---------------------------
class PSTanh(nn.Module):
    def __init__(self, num_capsules, per_dim=False, dim=None):
        super().__init__()
        self.per_dim = per_dim
        if per_dim:
            self.a = nn.Parameter(torch.ones(num_capsules, dim))
            self.b = nn.Parameter(torch.ones(num_capsules, dim))
        else:
            self.a = nn.Parameter(torch.tensor(1.0))
            self.b = nn.Parameter(torch.tensor(1.0))

    def forward(self, x):
        return self.a * torch.tanh(self.b * x)

# ---------------------------
# PSTanh Capsule Layer (simplified routing)
# ---------------------------
class PSTanhCapsuleLayer(nn.Module):
    def __init__(self, in_capsules, out_capsules, in_dim, out_dim, routing_iters=3):
        super().__init__()
        self.in_capsules = in_capsules
        self.out_capsules = out_capsules
        self.in_dim = in_dim
        self.out_dim = out_dim
        self.routing_iters = routing_iters

        self.W = nn.Parameter(torch.randn(out_capsules, in_capsules, out_dim, in_dim))
        self.pstanh = PSTanh(out_capsules, per_dim=False)   # shared parameters

    def forward(self, x):
        batch_size = x.size(0)
        # x shape: (batch, in_capsules, in_dim)

        # Compute predictions u_hat
        x = x.unsqueeze(1).unsqueeze(-2)          # (batch, 1, in_capsules, 1, in_dim)
        W = self.W.unsqueeze(0)                   # (1, out_capsules, in_capsules, out_dim, in_dim)
        u_hat = torch.matmul(W, x).squeeze(-1)    # (batch, out_capsules, in_capsules, out_dim)

        b = torch.zeros(batch_size, self.out_capsules, self.in_capsules, 1).to(x.device)

        for i in range(self.routing_iters):
            c = F.softmax(b, dim=1)                # coupling coefficients
            s = (c * u_hat).sum(dim=2)             # (batch, out_capsules, out_dim)
            v = self.pstanh(s)                      # apply PSTanh

            # For routing, use unit‑length vectors
            with torch.no_grad():
                v_norm = v.norm(dim=-1, keepdim=True)
                v_unit = v / (v_norm + 1e-8)

            if i < self.routing_iters - 1:
                agreement = (u_hat * v_unit.unsqueeze(2)).sum(dim=-1, keepdim=True)
                b = b + agreement

        # Final output (with PSTanh applied)
        return v

# ---------------------------
# Complete Capsule Network for Fashion‑MNIST
# ---------------------------
class CapsNetPSTanh(nn.Module):
    def __init__(self):
        super().__init__()
        # Conv layer to extract features
        self.conv1 = nn.Conv2d(1, 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        # Primary capsule layer: 32×6×6 capsules of dimension 8
        self.primary = nn.Conv2d(256, 32*8, kernel_size=9, stride=2, padding=0)
        self.primary_capsules = 32 * 6 * 6   # after conv: (28-9+1)=20 → after stride 2: 6
        self.primary_dim = 8

        # Digit capsule layer: 10 capsules (one per class) of dimension 16
        self.digit_layer = PSTanhCapsuleLayer(
            in_capsules=self.primary_capsules,
            out_capsules=10,
            in_dim=self.primary_dim,
            out_dim=16,
            routing_iters=3
        )

        # Loss function (margin loss for class presence)
        self.m_loss = nn.MarginRankingLoss(margin=0.9)

    def forward(self, x):
        # x: (batch, 1, 28, 28)
        x = self.relu(self.conv1(x))               # (batch, 256, 20, 20)
        x = self.primary(x)                         # (batch, 256, 6, 6) – actually 256 = 32*8
        batch_size = x.size(0)
        # Reshape to primary capsules: (batch, num_primary, dim)
        x = x.view(batch_size, 32*8, -1).transpose(1, 2)  # (batch, 6*6, 32*8) – wait, careful
        # Better: primary capsules are 32 * 6 * 6 = 1152 capsules each of dimension 8
        x = x.view(batch_size, 32*8, 6*6).transpose(1, 2)  # (batch, 36, 256) – still not right.
        # Actually, conv output: (batch, 256, 6, 6). We want 32 capsules per spatial position (6x6) each of dim 8.
        # So reshape to (batch, 32, 8, 6, 6) then flatten spatial: (batch, 32*6*6, 8)
        x = x.view(batch_size, 32, 8, 6, 6)
        x = x.permute(0, 3, 4, 1, 2).contiguous()   # (batch, 6, 6, 32, 8)
        x = x.view(batch_size, -1, 8)                # (batch, 1152, 8)  – 1152 = 32*6*6
        primary_caps = x

        # Digit capsules
        digit_caps = self.digit_layer(primary_caps)  # (batch, 10, 16)
        return digit_caps

# ---------------------------
# Margin Loss
# ---------------------------
def margin_loss(output, target, m_plus=0.9, m_minus=0.1, lambda_=0.5):
    # output: (batch, 10, 16) – we use the vector length as class probability
    v_norm = output.norm(dim=-1)                     # (batch, 10)
    # Build target matrix: one‑hot for present classes, -1 for absent
    target_onehot = torch.zeros_like(v_norm)
    target_onehot.scatter_(1, target.unsqueeze(-1), 1.0)
    # Loss for present classes
    loss_present = target_onehot * F.relu(m_plus - v_norm).pow(2)
    # Loss for absent classes
    loss_absent = (1 - target_onehot) * F.relu(v_norm - m_minus).pow(2) * lambda_
    return (loss_present + loss_absent).sum(dim=1).mean()

# ---------------------------
# Data loading
# ---------------------------
def get_dataloaders(batch_size=128):
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.1307,), (0.3081,))   # Fashion‑MNIST stats
    ])
    train_set = datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
    test_set = datasets.FashionMNIST(root='./data', train=False, download=True, transform=transform)
    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=4)
    test_loader = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=4)
    return train_loader, test_loader

# ---------------------------
# Training
# ---------------------------
def train(model, device, train_loader, optimizer, epoch):
    model.train()
    total_loss = 0
    correct = 0
    total = 0
    for data, target in tqdm(train_loader, desc=f'Epoch {epoch}'):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad()
        output = model(data)                     # (batch, 10, 16)
        loss = margin_loss(output, target)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * data.size(0)
        # Prediction: class with longest capsule vector
        pred = output.norm(dim=-1).argmax(dim=1)
        correct += pred.eq(target).sum().item()
        total += data.size(0)

    avg_loss = total_loss / total
    acc = 100. * correct / total
    print(f'Train Epoch {epoch}: Average loss: {avg_loss:.4f}, Accuracy: {acc:.2f}%')

def test(model, device, test_loader):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for data, target in tqdm(test_loader, desc='Testing'):
            data, target = data.to(device), target.to(device)
            output = model(data)
            pred = output.norm(dim=-1).argmax(dim=1)
            correct += pred.eq(target).sum().item()
            total += data.size(0)
    acc = 100. * correct / total
    print(f'Test Accuracy: {acc:.2f}%')

# ---------------------------
# Main
# ---------------------------
def main():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    batch_size = 128
    epochs = 10
    lr = 0.001

    train_loader, test_loader = get_dataloaders(batch_size)
    model = CapsNetPSTanh().to(device)
    optimizer = optim.Adam(model.parameters(), lr=lr)

    for epoch in range(1, epochs + 1):
        train(model, device, train_loader, optimizer, epoch)
        test(model, device, test_loader)

if __name__ == '__main__':
    main()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
from time import time


def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs


class PrimaryCapsule(nn.Module):
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsule, self).__init__()
        self.num_maps = num_maps
        self.num_caps = 6 * 6 * self.num_maps
        self.num_dims = num_dims
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                                kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        # x shape: (batch, 256, 20, 20)
        out = self.conv1(x)                # -> (batch, 256, 6, 6)
        out = out.view(-1, self.num_caps, self.num_dims)  # -> (batch, 1152, 8)
        out = squash(out)
        return out


class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        # x shape: (batch, num_caps_in, num_dims_in)
        x = x[:, None, :, :, None]            # -> (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # -> (num_caps_out, batch, num_caps_in, num_dims_out)

        x_hat_detached = x_hat.detach()

        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                # last iteration, use x_hat (with gradients)
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                # earlier iterations, use detached x_hat
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)      # -> (batch, num_caps_out, num_dims_out)


class CapsuleNet(nn.Module):
    """
    A Capsule Network on MNIST.
    """
    def __init__(self, input_size, classes, routings):
        super(CapsuleNet, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings

        # Layer 1: Conventional Conv2D layer
        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)

        # Layer 2: Primary Capsules
        self.primarycaps = PrimaryCapsule()   # default: 32 maps, 8D capsules

        # Layer 3: Digit Capsules (with routing)
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,      # 1152
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )

        # Decoder network
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )

        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))          # -> (batch, 256, 20, 20)
        out = self.primarycaps(out)             # -> (batch, 1152, 8)
        out = self.digitcaps(out)               # -> (batch, 10, 16)

        length = out.norm(dim=-1)                # -> (batch, 10)

        if y is None:  # during testing, no label given → create one-hot from prediction
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )  # -> (batch, 784)
        return length, reconstruction.view(-1, *self.input_size)


def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    """
    Capsule loss = Margin loss + lam_recon * reconstruction loss.
    """
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()

    L_recon = nn.MSELoss()(x_recon, x)

    return L_margin + lam_recon * L_recon


def show_reconstruction(model, test_loader, n_images, args):
    model.eval()
    for x, _ in test_loader:
        x = Variable(x[:min(n_images, x.size(0))].cuda(), volatile=True)
        _, x_recon = model(x)
        data = np.concatenate([x.data, x_recon.data])
        img = combine_images(np.transpose(data, [0, 2, 3, 1]))
        image = img * 255
        Image.fromarray(image.astype(np.uint8)).save(args["save_dir"] + "/real_and_recon.png")
        print()
        print('Reconstructed images are saved to %s/real_and_recon.png' % args["save_dir"])
        print('-' * 70)
        plt.imshow(plt.imread(args["save_dir"] + "/real_and_recon.png"))
        plt.show()
        break


def test(model, test_loader, args):
    model.eval()
    test_loss = 0
    correct = 0
    for x, y in test_loader:
        y = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y = Variable(x.cuda(), volatile=True), Variable(y.cuda())
        y_pred, x_recon = model(x)
        test_loss += caps_loss(y, y_pred, x, x_recon, args["lam_recon"]).item() * x.size(0)
        y_pred = y_pred.data.max(1)[1]
        y_true = y.data.max(1)[1]
        correct += y_pred.eq(y_true).cpu().sum()

    test_loss /= len(test_loader.dataset)
    return test_loss, correct / len(test_loader.dataset)


def train(model, train_loader, test_loader, args):
    """
    Training a CapsuleNet
    """
    print('Begin Training' + '-' * 70)
    logfile = open(args["save_dir"] + '/log.csv', 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    t0 = time()
    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.
    for epoch in range(args["epochs"]):
        model.train()
        lr_decay.step()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y = Variable(x.cuda()), Variable(y.cuda())

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y)
            loss = caps_loss(y, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        val_loss, val_acc = test(model, test_loader, args)
        logwriter.writerow(dict(epoch=epoch, loss=training_loss / len(train_loader.dataset),
                                val_loss=val_loss, val_acc=val_acc))
        print("==> Epoch %02d: loss=%.5f, val_loss=%.5f, val_acc=%.4f, time=%ds"
              % (epoch, training_loss / len(train_loader.dataset),
                 val_loss, val_acc, time() - ti))
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), args["save_dir"] + '/epoch%d.pkl' % epoch)
            print("best val_acc increased to %.4f" % best_val_acc)
    logfile.close()
    torch.save(model.state_dict(), args["save_dir"] + '/trained_model.pkl')
    print('Trained model saved to \'%s/trained_model.h5\'' % args["save_dir"])
    print("Total time = %ds" % (time() - t0))
    print('End Training' + '-' * 70)
    return model


def load_mnist(path='./data', download=False, batch_size=100, shift_pixels=2):
    """
    Construct dataloaders for training and test data. Data augmentation is also done here.
    """
    kwargs = {'num_workers': 1, 'pin_memory': True}

    train_loader = torch.utils.data.DataLoader(
        datasets.MNIST(path, train=True, download=download,
                       transform=transforms.Compose([
                           transforms.RandomCrop(size=28, padding=shift_pixels),
                           transforms.ToTensor()
                       ])),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.MNIST(path, train=False, download=download,
                       transform=transforms.ToTensor()),
        batch_size=batch_size, shuffle=True, **kwargs)

    return train_loader, test_loader


if __name__ == "__main__":
    # setting the hyper parameters
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": .001,
        "lr_decay": .9,
        "lam_recon": .0005 * 784,
        "routings": 3,
        "shift_pixels": 2,
        "data_dir": "./MNIST",
        "save_dir": "./result",
        "weights": None,
        "testing": False
    }
    if not os.path.exists(args["save_dir"]):
        os.makedirs(args["save_dir"])

    # load data
    train_loader, test_loader = load_mnist(args["data_dir"], download=True,
                                           batch_size=args["batch_size"])

    # define model
    model = CapsuleNet(input_size=[1, 28, 28], classes=10, routings=3)
    model.cuda()
    print(model)

    # train or test
    if args["weights"] is not None:
        model.load_state_dict(torch.load(args["weights"]))
    if not args["testing"]:
        train(model, train_loader, test_loader, args)
    else:
        if args["weights"] is None:
            print('No weights are provided. Will test using random initialized weights.')
        test_loss, test_acc = test(model=model, test_loader=test_loader, args=args)
        print('test acc = %.4f, test loss = %.5f' % (test_acc, test_loss))
        show_reconstruction(model, test_loader, 50, args)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
from time import time


def squash(inputs, axis=-1):
    """Squash function as used in capsule networks."""
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs


class PrimaryCapsule(nn.Module):
    """Primary Capsule layer that outputs squashed vectors and raw activation norms."""
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsule, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        # Convolution from 256 channels to num_maps * num_dims (each capsule dim is a channel)
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        # x shape: (batch, 256, 20, 20)
        out = self.conv1(x)                     # -> (batch, 256, 6, 6)
        batch_size = out.size(0)
        # Reshape to separate maps, spatial dimensions, and capsule dims
        out = out.view(batch_size, self.num_maps, self.num_dims, 6, 6)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 6, 6, 8)

        # Raw activation norms (before squashing) – used for CapFG
        raw_norms = torch.norm(out, dim=-1)              # (batch, 32, 6, 6)

        # Apply squashing along capsule dimension
        squashed = squash(out, axis=-1)                   # (batch, 32, 6, 6, 8)

        # Flatten spatial and map dimensions for routing
        flat_squashed = squashed.view(batch_size, -1, self.num_dims)  # (batch, 1152, 8)

        return flat_squashed, raw_norms


class DenseCapsule(nn.Module):
    """Capsule layer with routing-by-agreement (digit capsules)."""
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        # x shape: (batch, num_caps_in, num_dims_in)
        x = x[:, None, :, :, None]            # -> (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # -> (num_caps_out, batch, num_caps_in, num_dims_out)

        x_hat_detached = x_hat.detach()

        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                # last iteration, use x_hat (with gradients)
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                # earlier iterations, use detached x_hat
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)      # -> (batch, num_caps_out, num_dims_out)


class CapFG(nn.Module):
    """
    Capsule-Guided Foreground Gating.
    Takes raw activation norms of primary capsules and outputs a soft spatial mask.
    """
    def __init__(self, in_maps=32, grid_size=6):
        super(CapFG, self).__init__()
        # Small CNN to produce a single-channel spatial mask
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()      # outputs values in [0,1]
        )

    def forward(self, raw_norms):
        # raw_norms: (batch, 32, 6, 6)
        mask = self.conv(raw_norms)          # (batch, 1, 6, 6)
        return mask.squeeze(1)                # (batch, 6, 6)


class CapsuleNet(nn.Module):
    """
    Capsule Network with binary foreground masking and Capsule-Guided Foreground Gating.
    """
    def __init__(self, input_size, classes, routings):
        super(CapsuleNet, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings

        # Layer 1: Conventional Conv2D
        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)

        # Layer 2: Primary Capsules
        self.primarycaps = PrimaryCapsule()          # 32 maps, 8D capsules

        # Foreground gating module
        self.capfg = CapFG(in_maps=32, grid_size=6)

        # Layer 3: Digit Capsules (with routing)
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,      # 1152
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )

        # Decoder network
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )

        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        # ----- Binary foreground masking (image level) -----
        # For MNIST: simple threshold (pixel > 0.1) to separate digit from background
        mask_img = (x > 0.1).float()        # binary mask, same size as input
        x_fg = x * mask_img                 # foreground-only image

        # ----- Convolutional feature extraction -----
        out = self.relu(self.conv1(x_fg))   # -> (batch, 256, 20, 20)

        # ----- Primary capsules -----
        prim_flat, raw_norms = self.primarycaps(out)   # prim_flat: (batch, 1152, 8)
                                                        # raw_norms: (batch, 32, 6, 6)

        # ----- Capsule‑guided foreground gating -----
        # Compute soft spatial mask from raw activation norms
        A = self.capfg(raw_norms)            # (batch, 6, 6)

        # Reshape flattened primary capsules back to spatial form
        batch_size = prim_flat.size(0)
        prim_spatial = prim_flat.view(batch_size, 32, 6, 6, 8)  # (batch, 32, 6, 6, 8)

        # Apply gating: multiply each capsule vector by the mask value of its spatial cell
        # mask A: (batch, 6, 6) -> add dimensions for maps and capsule dims
        A = A.unsqueeze(1).unsqueeze(-1)     # (batch, 1, 6, 6, 1)
        prim_gated = prim_spatial * A         # broadcast to (batch, 32, 6, 6, 8)

        # Flatten again for routing
        prim_gated_flat = prim_gated.view(batch_size, -1, 8)  # (batch, 1152, 8)

        # ----- Digit capsules (routing) -----
        out = self.digitcaps(prim_gated_flat)   # -> (batch, 10, 16)

        # ----- Output lengths and reconstruction -----
        length = out.norm(dim=-1)                # (batch, 10)

        if y is None:  # test mode: create one-hot from prediction
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )  # -> (batch, 784)
        return length, reconstruction.view(-1, *self.input_size)


def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    """
    Capsule loss = Margin loss + lam_recon * reconstruction loss.
    """
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()

    L_recon = nn.MSELoss()(x_recon, x)

    return L_margin + lam_recon * L_recon


def combine_images(images):
    """Utility to combine a batch of images into a single grid for display."""
    images = np.array(images)
    n = images.shape[0]
    h, w, c = images.shape[1:]
    rows = int(np.ceil(np.sqrt(n)))
    cols = rows
    combined = np.zeros((h * rows, w * cols, c))
    for i in range(rows):
        for j in range(cols):
            idx = i * cols + j
            if idx < n:
                combined[i*h:(i+1)*h, j*w:(j+1)*w] = images[idx]
    return combined


def show_reconstruction(model, test_loader, n_images, args):
    model.eval()
    for x, _ in test_loader:
        x = x[:min(n_images, x.size(0))].cuda()
        with torch.no_grad():
            _, x_recon = model(x)
        data = np.concatenate([x.cpu().numpy(), x_recon.cpu().numpy()])
        img = combine_images(np.transpose(data, [0, 2, 3, 1]))
        image = img * 255
        Image.fromarray(image.astype(np.uint8)).save(args["save_dir"] + "/real_and_recon.png")
        print()
        print('Reconstructed images are saved to %s/real_and_recon.png' % args["save_dir"])
        print('-' * 70)
        plt.imshow(plt.imread(args["save_dir"] + "/real_and_recon.png"))
        plt.show()
        break


def test(model, test_loader, args):
    model.eval()
    test_loss = 0
    correct = 0
    for x, y in test_loader:
        y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y_onehot = x.cuda(), y_onehot.cuda()
        with torch.no_grad():
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
        test_loss += loss.item() * x.size(0)
        y_pred_class = y_pred.data.max(1)[1]
        y_true = y.cuda()
        correct += y_pred_class.eq(y_true).cpu().sum()

    test_loss /= len(test_loader.dataset)
    return test_loss, correct.item() / len(test_loader.dataset)


def train(model, train_loader, test_loader, args):
    """
    Training a CapsuleNet with CapFG.
    """
    print('Begin Training' + '-' * 70)
    logfile = open(args["save_dir"] + '/log.csv', 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    t0 = time()
    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.
    for epoch in range(args["epochs"]):
        model.train()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y_onehot = x.cuda(), y_onehot.cuda()

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        # Step the scheduler after optimizer step (PyTorch 1.1+ recommendation)
        lr_decay.step()

        val_loss, val_acc = test(model, test_loader, args)
        logwriter.writerow(dict(epoch=epoch, loss=training_loss / len(train_loader.dataset),
                                val_loss=val_loss, val_acc=val_acc))
        print("==> Epoch %02d: loss=%.5f, val_loss=%.5f, val_acc=%.4f, time=%ds"
              % (epoch, training_loss / len(train_loader.dataset),
                 val_loss, val_acc, time() - ti))
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), args["save_dir"] + '/epoch%d.pkl' % epoch)
            print("best val_acc increased to %.4f" % best_val_acc)
    logfile.close()
    torch.save(model.state_dict(), args["save_dir"] + '/trained_model.pkl')
    print('Trained model saved to \'%s/trained_model.h5\'' % args["save_dir"])
    print("Total time = %ds" % (time() - t0))
    print('End Training' + '-' * 70)
    return model


def load_mnist(path='./data', download=False, batch_size=100, shift_pixels=2):
    """
    Construct dataloaders for training and test data. Data augmentation is also done here.
    """
    kwargs = {'num_workers': 1, 'pin_memory': True}

    train_loader = torch.utils.data.DataLoader(
        datasets.MNIST(path, train=True, download=download,
                       transform=transforms.Compose([
                           transforms.RandomCrop(size=28, padding=shift_pixels),
                           transforms.ToTensor()
                       ])),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.MNIST(path, train=False, download=download,
                       transform=transforms.ToTensor()),
        batch_size=batch_size, shuffle=True, **kwargs)

    return train_loader, test_loader


if __name__ == "__main__":
    # setting the hyper parameters
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": .001,
        "lr_decay": .9,
        "lam_recon": .0005 * 784,
        "routings": 3,
        "shift_pixels": 2,
        "data_dir": "./MNIST",
        "save_dir": "./result",
        "weights": None,
        "testing": False
    }
    if not os.path.exists(args["save_dir"]):
        os.makedirs(args["save_dir"])

    # load data
    train_loader, test_loader = load_mnist(args["data_dir"], download=True,
                                           batch_size=args["batch_size"])

    # define model
    model = CapsuleNet(input_size=[1, 28, 28], classes=10, routings=3)
    model.cuda()
    print(model)

    # train or test
    if args["weights"] is not None:
        model.load_state_dict(torch.load(args["weights"]))
    if not args["testing"]:
        train(model, train_loader, test_loader, args)
    else:
        if args["weights"] is None:
            print('No weights are provided. Will test using random initialized weights.')
        test_loss, test_acc = test(model=model, test_loader=test_loader, args=args)
        print('test acc = %.4f, test loss = %.5f' % (test_acc, test_loss))
        show_reconstruction(model, test_loader, 50, args)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
from time import time

# For confusion matrix (optional dependency)
try:
    from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
    SKLEARN_AVAILABLE = True
except ImportError:
    SKLEARN_AVAILABLE = False
    print("scikit-learn not installed. Confusion matrix will be skipped.")


def squash(inputs, axis=-1):
    """Squash function as used in capsule networks."""
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs


class PrimaryCapsule(nn.Module):
    """Primary Capsule layer that outputs squashed vectors and raw activation norms."""
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsule, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv1(x)                     # (batch, 256, 6, 6)
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 6, 6)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 6, 6, 8)

        raw_norms = torch.norm(out, dim=-1)              # (batch, 32, 6, 6)
        squashed = squash(out, axis=-1)                   # (batch, 32, 6, 6, 8)
        flat_squashed = squashed.view(batch_size, -1, self.num_dims)  # (batch, 1152, 8)

        return flat_squashed, raw_norms


class DenseCapsule(nn.Module):
    """Capsule layer with routing-by-agreement (digit capsules)."""
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        x = x[:, None, :, :, None]            # (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # (num_caps_out, batch, num_caps_in, num_dims_out)

        x_hat_detached = x_hat.detach()
        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)      # (batch, num_caps_out, num_dims_out)


class CapFG(nn.Module):
    """Capsule-Guided Foreground Gating: soft spatial mask from raw capsule norms."""
    def __init__(self, in_maps=32, grid_size=6):
        super(CapFG, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, raw_norms):
        mask = self.conv(raw_norms)          # (batch, 1, 6, 6)
        return mask.squeeze(1)                # (batch, 6, 6)


class CapsuleNet(nn.Module):
    """Capsule Network with binary foreground masking and CapFG."""
    def __init__(self, input_size, classes, routings):
        super(CapsuleNet, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings

        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsule()
        self.capfg = CapFG(in_maps=32, grid_size=6)
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        # Binary foreground masking (image level)
        mask_img = (x > 0.1).float()        # simple threshold for MNIST
        x_fg = x * mask_img

        # Feature extraction
        out = self.relu(self.conv1(x_fg))   # (batch, 256, 20, 20)

        # Primary capsules
        prim_flat, raw_norms = self.primarycaps(out)   # (batch,1152,8) , (batch,32,6,6)

        # CapFG: compute soft spatial mask and apply gating
        A = self.capfg(raw_norms)            # (batch, 6, 6)
        batch_size = prim_flat.size(0)
        prim_spatial = prim_flat.view(batch_size, 32, 6, 6, 8)
        A = A.unsqueeze(1).unsqueeze(-1)     # (batch, 1, 6, 6, 1)
        prim_gated = prim_spatial * A
        prim_gated_flat = prim_gated.view(batch_size, -1, 8)

        # Digit capsules (routing)
        out = self.digitcaps(prim_gated_flat)   # (batch, 10, 16)

        length = out.norm(dim=-1)                # (batch, 10)

        if y is None:  # test mode
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)


def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()
    L_recon = nn.MSELoss()(x_recon, x)
    return L_margin + lam_recon * L_recon


def combine_images(images):
    images = np.array(images)
    n = images.shape[0]
    h, w, c = images.shape[1:]
    rows = int(np.ceil(np.sqrt(n)))
    cols = rows
    combined = np.zeros((h * rows, w * cols, c))
    for i in range(rows):
        for j in range(cols):
            idx = i * cols + j
            if idx < n:
                combined[i*h:(i+1)*h, j*w:(j+1)*w] = images[idx]
    return combined


def show_reconstruction(model, test_loader, n_images, args):
    model.eval()
    for x, _ in test_loader:
        x = x[:min(n_images, x.size(0))].cuda()
        with torch.no_grad():
            _, x_recon = model(x)
        # x and x_recon are on GPU, move to CPU and convert to numpy
        x_np = x.cpu().numpy()
        x_recon_np = x_recon.cpu().numpy()
        # Concatenate originals and reconstructions along batch dimension
        data = np.concatenate([x_np, x_recon_np], axis=0)  # (2*batch, 1, 28, 28)
        # Transpose to (batch, height, width, channels)
        data = np.transpose(data, [0, 2, 3, 1])            # (2*batch, 28, 28, 1)
        img = combine_images(data)                         # (H, W, 1)

        # --- FIX: if last dimension is 1, squeeze it ---
        if img.shape[-1] == 1:
            img = img.squeeze(axis=-1)                     # now (H, W)

        # Convert to 0-255 uint8
        img_uint8 = (img * 255).astype(np.uint8)
        save_path = os.path.join(args["save_dir"], "real_and_recon.png")
        Image.fromarray(img_uint8).save(save_path)
        print(f"\nReconstructed images saved to {save_path}")
        print('-' * 70)
        plt.imshow(plt.imread(save_path))
        plt.show()
        break


def test(model, test_loader, args, return_predictions=False):
    """If return_predictions=True, returns (loss, accuracy, all_preds, all_labels)."""
    model.eval()
    test_loss = 0
    correct = 0
    all_preds = []
    all_labels = []
    for x, y in test_loader:
        y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y_onehot = x.cuda(), y_onehot.cuda()
        with torch.no_grad():
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
        test_loss += loss.item() * x.size(0)
        pred_class = y_pred.data.max(1)[1]
        correct += pred_class.eq(y.cuda()).cpu().sum()
        if return_predictions:
            all_preds.extend(pred_class.cpu().numpy())
            all_labels.extend(y.numpy())

    test_loss /= len(test_loader.dataset)
    accuracy = correct.item() / len(test_loader.dataset)
    if return_predictions:
        return test_loss, accuracy, np.array(all_preds), np.array(all_labels)
    return test_loss, accuracy


def plot_training_history(history, save_dir):
    """Plot training/validation loss and validation accuracy."""
    epochs = range(1, len(history['train_loss']) + 1)

    plt.figure(figsize=(12, 4))

    # Loss plot
    plt.subplot(1, 2, 1)
    plt.plot(epochs, history['train_loss'], 'b-', label='Training loss')
    plt.plot(epochs, history['val_loss'], 'r-', label='Validation loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)

    # Accuracy plot
    plt.subplot(1, 2, 2)
    plt.plot(epochs, history['val_acc'], 'g-', label='Validation accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.title('Validation Accuracy')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    save_path = os.path.join(save_dir, 'training_curves.png')
    plt.savefig(save_path)
    print(f"Training curves saved to {save_path}")
    plt.show()


def plot_confusion_matrix(y_true, y_pred, save_dir):
    """Plot confusion matrix using sklearn if available."""
    if not SKLEARN_AVAILABLE:
        print("sklearn not available, skipping confusion matrix.")
        return
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=range(10))
    fig, ax = plt.subplots(figsize=(8, 8))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    ax.set_title('Confusion Matrix on Test Set')
    save_path = os.path.join(save_dir, 'confusion_matrix.png')
    plt.savefig(save_path)
    print(f"Confusion matrix saved to {save_path}")
    plt.show()


def train(model, train_loader, test_loader, args):
    """
    Training a CapsuleNet with CapFG. Returns history dict.
    """
    print('Begin Training' + '-' * 70)
    logfile = open(os.path.join(args["save_dir"], 'log.csv'), 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    t0 = time()
    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.

    # History storage
    history = {'train_loss': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(args["epochs"]):
        model.train()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y_onehot = x.cuda(), y_onehot.cuda()

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        lr_decay.step()

        # Validation
        val_loss, val_acc = test(model, test_loader, args)
        avg_train_loss = training_loss / len(train_loader.dataset)

        # Store history
        history['train_loss'].append(avg_train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        logwriter.writerow(dict(epoch=epoch, loss=avg_train_loss,
                                val_loss=val_loss, val_acc=val_acc))
        print("==> Epoch %02d: loss=%.5f, val_loss=%.5f, val_acc=%.4f, time=%ds"
              % (epoch, avg_train_loss, val_loss, val_acc, time() - ti))

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(args["save_dir"], f'epoch{epoch}.pkl'))
            print("best val_acc increased to %.4f" % best_val_acc)

    logfile.close()
    torch.save(model.state_dict(), os.path.join(args["save_dir"], 'trained_model.pkl'))
    print('Trained model saved to', os.path.join(args["save_dir"], 'trained_model.pkl'))
    print("Total time = %ds" % (time() - t0))
    print('End Training' + '-' * 70)
    return history


def load_mnist(path='./data', download=False, batch_size=100, shift_pixels=2):
    kwargs = {'num_workers': 1, 'pin_memory': True}
    train_loader = torch.utils.data.DataLoader(
        datasets.MNIST(path, train=True, download=download,
                       transform=transforms.Compose([
                           transforms.RandomCrop(size=28, padding=shift_pixels),
                           transforms.ToTensor()
                       ])),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.MNIST(path, train=False, download=download,
                       transform=transforms.ToTensor()),
        batch_size=batch_size, shuffle=True, **kwargs)
    return train_loader, test_loader


if __name__ == "__main__":
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": .001,
        "lr_decay": .9,
        "lam_recon": .0005 * 784,
        "routings": 3,
        "shift_pixels": 2,
        "data_dir": "./MNIST",
        "save_dir": "./result",
        "weights": None,
        "testing": False
    }
    if not os.path.exists(args["save_dir"]):
        os.makedirs(args["save_dir"])

    # Load data
    train_loader, test_loader = load_mnist(args["data_dir"], download=True,
                                           batch_size=args["batch_size"])

    # Model
    model = CapsuleNet(input_size=[1, 28, 28], classes=10, routings=3)
    model.cuda()
    print(model)

    # Train or test
    if args["weights"] is not None:
        model.load_state_dict(torch.load(args["weights"]))

    if not args["testing"]:
        # Train and get history
        history = train(model, train_loader, test_loader, args)

        # Plot training curves
        plot_training_history(history, args["save_dir"])

        # Evaluate on test set and plot confusion matrix
        test_loss, test_acc, y_pred, y_true = test(model, test_loader, args, return_predictions=True)
        print(f"\nFinal Test Accuracy: {test_acc:.4f}, Test Loss: {test_loss:.5f}")
        plot_confusion_matrix(y_true, y_pred, args["save_dir"])

        # Show reconstruction examples
        show_reconstruction(model, test_loader, 50, args)

    else:  # testing mode
        if args["weights"] is None:
            print('No weights provided. Testing with random initialization.')
        test_loss, test_acc, y_pred, y_true = test(model, test_loader, args, return_predictions=True)
        print(f'Test acc = {test_acc:.4f}, test loss = {test_loss:.5f}')
        plot_confusion_matrix(y_true, y_pred, args["save_dir"])
        show_reconstruction(model, test_loader, 50, args)
```

```python
# capsule_base_fashion.py
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
from time import time

# For confusion matrix
try:
    from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
    SKLEARN_AVAILABLE = True
except ImportError:
    SKLEARN_AVAILABLE = False
    print("scikit-learn not installed. Confusion matrix will be skipped.")


def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs


class PrimaryCapsule(nn.Module):
    """Original Primary Capsule (no extra outputs)."""
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsule, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv1(x)                     # (batch, 256, 6, 6)
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 6, 6)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 6, 6, 8)
        squashed = squash(out, axis=-1)
        return squashed.view(batch_size, -1, self.num_dims)  # (batch, 1152, 8)


class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        x = x[:, None, :, :, None]            # (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # (num_caps_out, batch, num_caps_in, num_dims_out)
        x_hat_detached = x_hat.detach()
        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)


class CapsuleNet(nn.Module):
    """Base Capsule Network (no CapFG)."""
    def __init__(self, input_size, classes, routings):
        super(CapsuleNet, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings

        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsule()          # 32 maps, 8D capsules
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        # Simple binary masking (pixel > 0.1) – works for Fashion‑MNIST
        mask_img = (x > 0.1).float()
        x_fg = x * mask_img

        out = self.relu(self.conv1(x_fg))            # (batch, 256, 20, 20)
        out = self.primarycaps(out)                  # (batch, 1152, 8)
        out = self.digitcaps(out)                    # (batch, 10, 16)

        length = out.norm(dim=-1)                     # (batch, 10)

        if y is None:  # test mode
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)


def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()
    L_recon = nn.MSELoss()(x_recon, x)
    return L_margin + lam_recon * L_recon


def combine_images(images):
    images = np.array(images)
    n = images.shape[0]
    h, w, c = images.shape[1:]
    rows = int(np.ceil(np.sqrt(n)))
    cols = rows
    combined = np.zeros((h * rows, w * cols, c))
    for i in range(rows):
        for j in range(cols):
            idx = i * cols + j
            if idx < n:
                combined[i*h:(i+1)*h, j*w:(j+1)*w] = images[idx]
    return combined


def show_reconstruction(model, test_loader, n_images, args):
    model.eval()
    for x, _ in test_loader:
        x = x[:min(n_images, x.size(0))].cuda()
        with torch.no_grad():
            _, x_recon = model(x)
        x_np = x.cpu().numpy()
        x_recon_np = x_recon.cpu().numpy()
        data = np.concatenate([x_np, x_recon_np], axis=0)
        data = np.transpose(data, [0, 2, 3, 1])
        img = combine_images(data)
        if img.shape[-1] == 1:
            img = img.squeeze(axis=-1)
        img_uint8 = (img * 255).astype(np.uint8)
        save_path = os.path.join(args["save_dir"], "real_and_recon.png")
        Image.fromarray(img_uint8).save(save_path)
        print(f"\nReconstructed images saved to {save_path}")
        print('-' * 70)
        plt.imshow(plt.imread(save_path))
        plt.show()
        break


def test(model, test_loader, args, return_predictions=False):
    model.eval()
    test_loss = 0
    correct = 0
    all_preds = []
    all_labels = []
    for x, y in test_loader:
        y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y_onehot = x.cuda(), y_onehot.cuda()
        with torch.no_grad():
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
        test_loss += loss.item() * x.size(0)
        pred_class = y_pred.data.max(1)[1]
        correct += pred_class.eq(y.cuda()).cpu().sum()
        if return_predictions:
            all_preds.extend(pred_class.cpu().numpy())
            all_labels.extend(y.numpy())

    test_loss /= len(test_loader.dataset)
    accuracy = correct.item() / len(test_loader.dataset)
    if return_predictions:
        return test_loss, accuracy, np.array(all_preds), np.array(all_labels)
    return test_loss, accuracy


def plot_training_history(history, save_dir):
    epochs = range(1, len(history['train_loss']) + 1)
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, history['train_loss'], 'b-', label='Training loss')
    plt.plot(epochs, history['val_loss'], 'r-', label='Validation loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(epochs, history['val_acc'], 'g-', label='Validation accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.title('Validation Accuracy')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    save_path = os.path.join(save_dir, 'training_curves.png')
    plt.savefig(save_path)
    print(f"Training curves saved to {save_path}")
    plt.show()


def plot_confusion_matrix(y_true, y_pred, save_dir):
    if not SKLEARN_AVAILABLE:
        print("sklearn not available, skipping confusion matrix.")
        return
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=range(10))
    fig, ax = plt.subplots(figsize=(8, 8))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    ax.set_title('Confusion Matrix on Test Set')
    save_path = os.path.join(save_dir, 'confusion_matrix.png')
    plt.savefig(save_path)
    print(f"Confusion matrix saved to {save_path}")
    plt.show()


def train(model, train_loader, test_loader, args):
    print('Begin Training' + '-' * 70)
    logfile = open(os.path.join(args["save_dir"], 'log.csv'), 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    t0 = time()
    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.

    history = {'train_loss': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(args["epochs"]):
        model.train()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y_onehot = x.cuda(), y_onehot.cuda()

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        lr_decay.step()

        val_loss, val_acc = test(model, test_loader, args)
        avg_train_loss = training_loss / len(train_loader.dataset)

        history['train_loss'].append(avg_train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        logwriter.writerow(dict(epoch=epoch, loss=avg_train_loss,
                                val_loss=val_loss, val_acc=val_acc))
        print("==> Epoch %02d: loss=%.5f, val_loss=%.5f, val_acc=%.4f, time=%ds"
              % (epoch, avg_train_loss, val_loss, val_acc, time() - ti))

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(args["save_dir"], f'epoch{epoch}.pkl'))
            print("best val_acc increased to %.4f" % best_val_acc)

    logfile.close()
    torch.save(model.state_dict(), os.path.join(args["save_dir"], 'trained_model.pkl'))
    print('Trained model saved to', os.path.join(args["save_dir"], 'trained_model.pkl'))
    print("Total time = %ds" % (time() - t0))
    print('End Training' + '-' * 70)
    return history


def load_fashion_mnist(path='./data', download=False, batch_size=100, shift_pixels=2):
    kwargs = {'num_workers': 1, 'pin_memory': True}
    train_loader = torch.utils.data.DataLoader(
        datasets.FashionMNIST(path, train=True, download=download,
                              transform=transforms.Compose([
                                  transforms.RandomCrop(size=28, padding=shift_pixels),
                                  transforms.ToTensor()
                              ])),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.FashionMNIST(path, train=False, download=download,
                              transform=transforms.ToTensor()),
        batch_size=batch_size, shuffle=True, **kwargs)
    return train_loader, test_loader


if __name__ == "__main__":
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": .001,
        "lr_decay": .9,
        "lam_recon": .0005 * 784,
        "routings": 3,
        "shift_pixels": 2,
        "data_dir": "./FashionMNIST",
        "save_dir": "./result_base",
        "weights": None,
        "testing": False
    }
    if not os.path.exists(args["save_dir"]):
        os.makedirs(args["save_dir"])

    train_loader, test_loader = load_fashion_mnist(args["data_dir"], download=True,
                                                   batch_size=args["batch_size"])

    model = CapsuleNet(input_size=[1, 28, 28], classes=10, routings=3)
    model.cuda()
    print(model)

    if args["weights"] is not None:
        model.load_state_dict(torch.load(args["weights"]))

    if not args["testing"]:
        history = train(model, train_loader, test_loader, args)
        plot_training_history(history, args["save_dir"])
        test_loss, test_acc, y_pred, y_true = test(model, test_loader, args, return_predictions=True)
        print(f"\nFinal Test Accuracy: {test_acc:.4f}, Test Loss: {test_loss:.5f}")
        plot_confusion_matrix(y_true, y_pred, args["save_dir"])
        show_reconstruction(model, test_loader, 50, args)
    else:
        if args["weights"] is None:
            print('No weights provided. Testing with random initialization.')
        test_loss, test_acc, y_pred, y_true = test(model, test_loader, args, return_predictions=True)
        print(f'Test acc = {test_acc:.4f}, test loss = {test_loss:.5f}')
        plot_confusion_matrix(y_true, y_pred, args["save_dir"])
        show_reconstruction(model, test_loader, 50, args)
```

2. CapFG‑Enhanced Capsule Network (Fashion‑MNIST)

```python
# capsule_capfg_fashion.py
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
from time import time

# For confusion matrix
try:
    from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
    SKLEARN_AVAILABLE = True
except ImportError:
    SKLEARN_AVAILABLE = False
    print("scikit-learn not installed. Confusion matrix will be skipped.")


def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs


class PrimaryCapsule(nn.Module):
    """Primary Capsule that also returns raw norms for CapFG."""
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsule, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv1(x)                     # (batch, 256, 6, 6)
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 6, 6)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 6, 6, 8)
        raw_norms = torch.norm(out, dim=-1)               # (batch, 32, 6, 6)
        squashed = squash(out, axis=-1)                    # (batch, 32, 6, 6, 8)
        flat_squashed = squashed.view(batch_size, -1, self.num_dims)  # (batch, 1152, 8)
        return flat_squashed, raw_norms


class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        x = x[:, None, :, :, None]            # (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # (num_caps_out, batch, num_caps_in, num_dims_out)
        x_hat_detached = x_hat.detach()
        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)


class CapFG(nn.Module):
    """Capsule-Guided Foreground Gating module."""
    def __init__(self, in_maps=32, grid_size=6):
        super(CapFG, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, raw_norms):
        mask = self.conv(raw_norms)          # (batch, 1, 6, 6)
        return mask.squeeze(1)                # (batch, 6, 6)


class CapsuleNet(nn.Module):
    """Capsule Network with CapFG."""
    def __init__(self, input_size, classes, routings):
        super(CapsuleNet, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings

        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsule()
        self.capfg = CapFG(in_maps=32, grid_size=6)
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        # Binary foreground masking
        mask_img = (x > 0.1).float()
        x_fg = x * mask_img

        out = self.relu(self.conv1(x_fg))            # (batch, 256, 20, 20)
        prim_flat, raw_norms = self.primarycaps(out)  # (batch,1152,8) , (batch,32,6,6)

        # CapFG: compute soft mask and apply gating
        A = self.capfg(raw_norms)                     # (batch, 6, 6)
        batch_size = prim_flat.size(0)
        prim_spatial = prim_flat.view(batch_size, 32, 6, 6, 8)
        A = A.unsqueeze(1).unsqueeze(-1)               # (batch, 1, 6, 6, 1)
        prim_gated = prim_spatial * A
        prim_gated_flat = prim_gated.view(batch_size, -1, 8)

        out = self.digitcaps(prim_gated_flat)          # (batch, 10, 16)

        length = out.norm(dim=-1)                       # (batch, 10)

        if y is None:  # test mode
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)


def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()
    L_recon = nn.MSELoss()(x_recon, x)
    return L_margin + lam_recon * L_recon


def combine_images(images):
    images = np.array(images)
    n = images.shape[0]
    h, w, c = images.shape[1:]
    rows = int(np.ceil(np.sqrt(n)))
    cols = rows
    combined = np.zeros((h * rows, w * cols, c))
    for i in range(rows):
        for j in range(cols):
            idx = i * cols + j
            if idx < n:
                combined[i*h:(i+1)*h, j*w:(j+1)*w] = images[idx]
    return combined


def show_reconstruction(model, test_loader, n_images, args):
    model.eval()
    for x, _ in test_loader:
        x = x[:min(n_images, x.size(0))].cuda()
        with torch.no_grad():
            _, x_recon = model(x)
        x_np = x.cpu().numpy()
        x_recon_np = x_recon.cpu().numpy()
        data = np.concatenate([x_np, x_recon_np], axis=0)
        data = np.transpose(data, [0, 2, 3, 1])
        img = combine_images(data)
        if img.shape[-1] == 1:
            img = img.squeeze(axis=-1)
        img_uint8 = (img * 255).astype(np.uint8)
        save_path = os.path.join(args["save_dir"], "real_and_recon.png")
        Image.fromarray(img_uint8).save(save_path)
        print(f"\nReconstructed images saved to {save_path}")
        print('-' * 70)
        plt.imshow(plt.imread(save_path))
        plt.show()
        break


def test(model, test_loader, args, return_predictions=False):
    model.eval()
    test_loss = 0
    correct = 0
    all_preds = []
    all_labels = []
    for x, y in test_loader:
        y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y_onehot = x.cuda(), y_onehot.cuda()
        with torch.no_grad():
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
        test_loss += loss.item() * x.size(0)
        pred_class = y_pred.data.max(1)[1]
        correct += pred_class.eq(y.cuda()).cpu().sum()
        if return_predictions:
            all_preds.extend(pred_class.cpu().numpy())
            all_labels.extend(y.numpy())

    test_loss /= len(test_loader.dataset)
    accuracy = correct.item() / len(test_loader.dataset)
    if return_predictions:
        return test_loss, accuracy, np.array(all_preds), np.array(all_labels)
    return test_loss, accuracy


def plot_training_history(history, save_dir):
    epochs = range(1, len(history['train_loss']) + 1)
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, history['train_loss'], 'b-', label='Training loss')
    plt.plot(epochs, history['val_loss'], 'r-', label='Validation loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(epochs, history['val_acc'], 'g-', label='Validation accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.title('Validation Accuracy')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    save_path = os.path.join(save_dir, 'training_curves.png')
    plt.savefig(save_path)
    print(f"Training curves saved to {save_path}")
    plt.show()


def plot_confusion_matrix(y_true, y_pred, save_dir):
    if not SKLEARN_AVAILABLE:
        print("sklearn not available, skipping confusion matrix.")
        return
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=range(10))
    fig, ax = plt.subplots(figsize=(8, 8))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    ax.set_title('Confusion Matrix on Test Set')
    save_path = os.path.join(save_dir, 'confusion_matrix.png')
    plt.savefig(save_path)
    print(f"Confusion matrix saved to {save_path}")
    plt.show()


def train(model, train_loader, test_loader, args):
    print('Begin Training' + '-' * 70)
    logfile = open(os.path.join(args["save_dir"], 'log.csv'), 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    t0 = time()
    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.

    history = {'train_loss': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(args["epochs"]):
        model.train()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y_onehot = x.cuda(), y_onehot.cuda()

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        lr_decay.step()

        val_loss, val_acc = test(model, test_loader, args)
        avg_train_loss = training_loss / len(train_loader.dataset)

        history['train_loss'].append(avg_train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        logwriter.writerow(dict(epoch=epoch, loss=avg_train_loss,
                                val_loss=val_loss, val_acc=val_acc))
        print("==> Epoch %02d: loss=%.5f, val_loss=%.5f, val_acc=%.4f, time=%ds"
              % (epoch, avg_train_loss, val_loss, val_acc, time() - ti))

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(args["save_dir"], f'epoch{epoch}.pkl'))
            print("best val_acc increased to %.4f" % best_val_acc)

    logfile.close()
    torch.save(model.state_dict(), os.path.join(args["save_dir"], 'trained_model.pkl'))
    print('Trained model saved to', os.path.join(args["save_dir"], 'trained_model.pkl'))
    print("Total time = %ds" % (time() - t0))
    print('End Training' + '-' * 70)
    return history


def load_fashion_mnist(path='./data', download=False, batch_size=100, shift_pixels=2):
    kwargs = {'num_workers': 1, 'pin_memory': True}
    train_loader = torch.utils.data.DataLoader(
        datasets.FashionMNIST(path, train=True, download=download,
                              transform=transforms.Compose([
                                  transforms.RandomCrop(size=28, padding=shift_pixels),
                                  transforms.ToTensor()
                              ])),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.FashionMNIST(path, train=False, download=download,
                              transform=transforms.ToTensor()),
        batch_size=batch_size, shuffle=True, **kwargs)
    return train_loader, test_loader


if __name__ == "__main__":
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": .001,
        "lr_decay": .9,
        "lam_recon": .0005 * 784,
        "routings": 3,
        "shift_pixels": 2,
        "data_dir": "./FashionMNIST",
        "save_dir": "./result_capfg",
        "weights": None,
        "testing": False
    }
    if not os.path.exists(args["save_dir"]):
        os.makedirs(args["save_dir"])

    train_loader, test_loader = load_fashion_mnist(args["data_dir"], download=True,
                                                   batch_size=args["batch_size"])

    model = CapsuleNet(input_size=[1, 28, 28], classes=10, routings=3)
    model.cuda()
    print(model)

    if args["weights"] is not None:
        model.load_state_dict(torch.load(args["weights"]))

    if not args["testing"]:
        history = train(model, train_loader, test_loader, args)
        plot_training_history(history, args["save_dir"])
        test_loss, test_acc, y_pred, y_true = test(model, test_loader, args, return_predictions=True)
        print(f"\nFinal Test Accuracy: {test_acc:.4f}, Test Loss: {test_loss:.5f}")
        plot_confusion_matrix(y_true, y_pred, args["save_dir"])
        show_reconstruction(model, test_loader, 50, args)
    else:
        if args["weights"] is None:
            print('No weights provided. Testing with random initialization.')
        test_loss, test_acc, y_pred, y_true = test(model, test_loader, args, return_predictions=True)
        print(f'Test acc = {test_acc:.4f}, test loss = {test_loss:.5f}')
        plot_confusion_matrix(y_true, y_pred, args["save_dir"])
        show_reconstruction(model, test_loader, 50, args)
```

```python
# full_comparison.py
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
import pandas as pd
from time import time

# ---------------------------
#  Common utilities
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()
    L_recon = nn.MSELoss()(x_recon, x)
    return L_margin + lam_recon * L_recon

def combine_images(images):
    images = np.array(images)
    n = images.shape[0]
    h, w, c = images.shape[1:]
    rows = int(np.ceil(np.sqrt(n)))
    cols = rows
    combined = np.zeros((h * rows, w * cols, c))
    for i in range(rows):
        for j in range(cols):
            idx = i * cols + j
            if idx < n:
                combined[i*h:(i+1)*h, j*w:(j+1)*w] = images[idx]
    return combined

def load_fashion_mnist(path='./FashionMNIST', download=False, batch_size=100, shift_pixels=2):
    kwargs = {'num_workers': 1, 'pin_memory': True}
    train_loader = torch.utils.data.DataLoader(
        datasets.FashionMNIST(path, train=True, download=download,
                              transform=transforms.Compose([
                                  transforms.RandomCrop(size=28, padding=shift_pixels),
                                  transforms.ToTensor()
                              ])),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.FashionMNIST(path, train=False, download=download,
                              transform=transforms.ToTensor()),
        batch_size=batch_size, shuffle=True, **kwargs)
    return train_loader, test_loader

# ---------------------------
#  Base Capsule Network
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsuleBase, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv1(x)                     # (batch, 256, 6, 6)
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 6, 6)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 6, 6, 8)
        squashed = squash(out, axis=-1)
        return squashed.view(batch_size, -1, self.num_dims)  # (batch, 1152, 8)

class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        x = x[:, None, :, :, None]            # (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # (num_caps_out, batch, num_caps_in, num_dims_out)
        x_hat_detached = x_hat.detach()
        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)

class CapsuleNetBase(nn.Module):
    def __init__(self, input_size, classes, routings):
        super(CapsuleNetBase, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings
        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsuleBase()
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        mask_img = (x > 0.1).float()
        x_fg = x * mask_img
        out = self.relu(self.conv1(x_fg))
        out = self.primarycaps(out)
        out = self.digitcaps(out)
        length = out.norm(dim=-1)
        if y is None:
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())
        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)

# ---------------------------
#  CapFG Capsule Network
# ---------------------------
class PrimaryCapsuleCapFG(nn.Module):
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsuleCapFG, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv1 = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv1(x)                     # (batch, 256, 6, 6)
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 6, 6)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 6, 6, 8)
        raw_norms = torch.norm(out, dim=-1)               # (batch, 32, 6, 6)
        squashed = squash(out, axis=-1)                    # (batch, 32, 6, 6, 8)
        flat_squashed = squashed.view(batch_size, -1, self.num_dims)  # (batch, 1152, 8)
        return flat_squashed, raw_norms

class CapFG(nn.Module):
    def __init__(self, in_maps=32, grid_size=6):
        super(CapFG, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, raw_norms):
        mask = self.conv(raw_norms)          # (batch, 1, 6, 6)
        return mask.squeeze(1)                # (batch, 6, 6)

class CapsuleNetCapFG(nn.Module):
    def __init__(self, input_size, classes, routings):
        super(CapsuleNetCapFG, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings
        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsuleCapFG()
        self.capfg = CapFG(in_maps=32, grid_size=6)
        self.digitcaps = DenseCapsule(
            num_caps_in=32*6*6,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        mask_img = (x > 0.1).float()
        x_fg = x * mask_img
        out = self.relu(self.conv1(x_fg))
        prim_flat, raw_norms = self.primarycaps(out)

        A = self.capfg(raw_norms)                     # (batch, 6, 6)
        batch_size = prim_flat.size(0)
        prim_spatial = prim_flat.view(batch_size, 32, 6, 6, 8)
        A = A.unsqueeze(1).unsqueeze(-1)               # (batch, 1, 6, 6, 1)
        prim_gated = prim_spatial * A
        prim_gated_flat = prim_gated.view(batch_size, -1, 8)

        out = self.digitcaps(prim_gated_flat)
        length = out.norm(dim=-1)
        if y is None:
            index = length.max(dim=1)[1]
            y = Variable(torch.zeros(length.size()).scatter_(
                1, index.view(-1, 1).cpu().data, 1.).cuda())
        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)

# ---------------------------
#  Training and evaluation functions (same for both models)
# ---------------------------
def test(model, test_loader, lam_recon, return_predictions=False):
    model.eval()
    test_loss = 0
    correct = 0
    all_preds = []
    all_labels = []
    for x, y in test_loader:
        y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y_onehot = x.cuda(), y_onehot.cuda()
        with torch.no_grad():
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)
        test_loss += loss.item() * x.size(0)
        pred_class = y_pred.data.max(1)[1]
        correct += pred_class.eq(y.cuda()).cpu().sum()
        if return_predictions:
            all_preds.extend(pred_class.cpu().numpy())
            all_labels.extend(y.numpy())

    test_loss /= len(test_loader.dataset)
    accuracy = correct.item() / len(test_loader.dataset)
    if return_predictions:
        return test_loss, accuracy, np.array(all_preds), np.array(all_labels)
    return test_loss, accuracy

def train_model(model, train_loader, test_loader, args, model_name):
    save_dir = args["save_dir"]
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)

    print(f"\n--- Training {model_name} ---")
    logfile = open(os.path.join(save_dir, 'log.csv'), 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.

    history = {'train_loss': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(args["epochs"]):
        model.train()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y_onehot = x.cuda(), y_onehot.cuda()

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        lr_decay.step()

        val_loss, val_acc = test(model, test_loader, args["lam_recon"])
        avg_train_loss = training_loss / len(train_loader.dataset)

        history['train_loss'].append(avg_train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        logwriter.writerow(dict(epoch=epoch, loss=avg_train_loss,
                                val_loss=val_loss, val_acc=val_acc))
        print(f"Epoch {epoch:02d}: loss={avg_train_loss:.5f}, val_loss={val_loss:.5f}, val_acc={val_acc:.4f}, time={int(time()-ti)}s")

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(save_dir, f'epoch{epoch}.pkl'))
            print(f"  best val_acc increased to {best_val_acc:.4f}")

    logfile.close()
    torch.save(model.state_dict(), os.path.join(save_dir, 'trained_model.pkl'))
    print(f"{model_name} training finished. Model saved to {save_dir}/trained_model.pkl")
    return history

# ---------------------------
#  Plot comparison
# ---------------------------
def plot_comparison(log_base, log_capfg, save_path='comparison_plots.png'):
    df_base = pd.read_csv(log_base)
    df_capfg = pd.read_csv(log_capfg)

    epochs_base = df_base['epoch'] + 1
    epochs_capfg = df_capfg['epoch'] + 1

    plt.style.use('seaborn-v0_8-darkgrid')
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    # Training loss
    axes[0].plot(epochs_base, df_base['loss'], 'b-', label='Base')
    axes[0].plot(epochs_capfg, df_capfg['loss'], 'r-', label='CapFG')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title('Training Loss')
    axes[0].legend()
    axes[0].grid(True)

    # Validation loss
    axes[1].plot(epochs_base, df_base['val_loss'], 'b--', label='Base')
    axes[1].plot(epochs_capfg, df_capfg['val_loss'], 'r--', label='CapFG')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].set_title('Validation Loss')
    axes[1].legend()
    axes[1].grid(True)

    # Validation accuracy
    axes[2].plot(epochs_base, df_base['val_acc'], 'b-.', label='Base')
    axes[2].plot(epochs_capfg, df_capfg['val_acc'], 'r-.', label='CapFG')
    axes[2].set_xlabel('Epoch')
    axes[2].set_ylabel('Accuracy')
    axes[2].set_title('Validation Accuracy')
    axes[2].legend()
    axes[2].grid(True)

    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.show()
    print(f"Comparison plots saved as '{save_path}'")

# ---------------------------
#  Main script
# ---------------------------
if __name__ == "__main__":
    # Settings
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": .001,
        "lr_decay": .9,
        "lam_recon": .0005 * 784,
        "routings": 3,
        "shift_pixels": 2,
        "data_dir": "./FashionMNIST",
    }

    # Directories for results
    base_dir = "./result_base"
    capfg_dir = "./result_capfg"
    args_base = args.copy()
    args_base["save_dir"] = base_dir
    args_capfg = args.copy()
    args_capfg["save_dir"] = capfg_dir

    # Load data
    print("Loading Fashion-MNIST...")
    train_loader, test_loader = load_fashion_mnist(args["data_dir"], download=True,
                                                   batch_size=args["batch_size"])

    # Option to skip training if logs already exist (set to False to force retrain)
    skip_if_exists = True

    # Train base model if needed
    log_base_path = os.path.join(base_dir, 'log.csv')
    if not skip_if_exists or not os.path.exists(log_base_path):
        print("\n--- Initializing Base Model ---")
        model_base = CapsuleNetBase(input_size=[1, 28, 28], classes=10, routings=args["routings"])
        model_base.cuda()
        train_model(model_base, train_loader, test_loader, args_base, "Base Capsule")
    else:
        print(f"Base model log found at {log_base_path}, skipping training.")

    # Train CapFG model if needed
    log_capfg_path = os.path.join(capfg_dir, 'log.csv')
    if not skip_if_exists or not os.path.exists(log_capfg_path):
        print("\n--- Initializing CapFG Model ---")
        model_capfg = CapsuleNetCapFG(input_size=[1, 28, 28], classes=10, routings=args["routings"])
        model_capfg.cuda()
        train_model(model_capfg, train_loader, test_loader, args_capfg, "CapFG Capsule")
    else:
        print(f"CapFG model log found at {log_capfg_path}, skipping training.")

    # Plot comparison
    if os.path.exists(log_base_path) and os.path.exists(log_capfg_path):
        plot_comparison(log_base_path, log_capfg_path)
    else:
        print("One or both log files are missing. Cannot generate comparison plots.")
```

```python
# cifar10_comparison.py
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.autograd import Variable
from torch.optim import Adam, lr_scheduler
from torchvision import transforms, datasets
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import csv
import os
import pandas as pd
from time import time

# For confusion matrix (optional)
try:
    from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
    SKLEARN_AVAILABLE = True
except ImportError:
    SKLEARN_AVAILABLE = False
    print("scikit-learn not installed. Confusion matrix will be skipped.")

# ---------------------------
#  Common utilities
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = norm**2 / (1 + norm**2) / (norm + 1e-8)
    return scale * inputs

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    L = y_true * torch.clamp(0.9 - y_pred, min=0.) ** 2 + \
        0.5 * (1 - y_true) * torch.clamp(y_pred - 0.1, min=0.) ** 2
    L_margin = L.sum(dim=1).mean()
    L_recon = nn.MSELoss()(x_recon, x)
    return L_margin + lam_recon * L_recon

def combine_images(images):
    """Combine a batch of images (H,W,C) into a grid."""
    images = np.array(images)
    n = images.shape[0]
    h, w, c = images.shape[1:]
    rows = int(np.ceil(np.sqrt(n)))
    cols = rows
    combined = np.zeros((h * rows, w * cols, c))
    for i in range(rows):
        for j in range(cols):
            idx = i * cols + j
            if idx < n:
                combined[i*h:(i+1)*h, j*w:(j+1)*w] = images[idx]
    return combined

def show_reconstruction(model, test_loader, n_images, save_dir, model_name):
    """Display and save original vs reconstructed images."""
    model.eval()
    for x, _ in test_loader:
        x = x[:min(n_images, x.size(0))].cuda()
        with torch.no_grad():
            _, x_recon = model(x)
        # Move to CPU and convert to numpy
        x_np = x.cpu().numpy()                # (N, C, H, W)
        x_recon_np = x_recon.cpu().numpy()
        # Concatenate originals and reconstructions
        data = np.concatenate([x_np, x_recon_np], axis=0)
        # Transpose to (N, H, W, C)
        data = np.transpose(data, [0, 2, 3, 1])
        img = combine_images(data)             # (H_grid, W_grid, C)
        # If grayscale (C=1), squeeze; else keep as RGB
        if img.shape[-1] == 1:
            img = img.squeeze(axis=-1)
        img_uint8 = (img * 255).astype(np.uint8)
        save_path = os.path.join(save_dir, f"recon_{model_name}.png")
        Image.fromarray(img_uint8).save(save_path)
        print(f"Reconstructed images saved to {save_path}")
        plt.imshow(img_uint8)
        plt.title(f"{model_name} - Original (top) vs Reconstructed (bottom)")
        plt.axis('off')
        plt.show()
        break

def plot_confusion_matrix(y_true, y_pred, save_dir, model_name):
    """Plot and save confusion matrix."""
    if not SKLEARN_AVAILABLE:
        return
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=range(10))
    fig, ax = plt.subplots(figsize=(8, 8))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    ax.set_title(f'Confusion Matrix - {model_name}')
    save_path = os.path.join(save_dir, f"confusion_{model_name}.png")
    plt.savefig(save_path)
    print(f"Confusion matrix saved to {save_path}")
    plt.show()

def load_cifar10(path='./CIFAR10', download=False, batch_size=100, shift_pixels=4):
    """Data loaders for CIFAR-10 with optional random crop augmentation."""
    kwargs = {'num_workers': 1, 'pin_memory': True}
    transform_train = transforms.Compose([
        transforms.RandomCrop(32, padding=shift_pixels),
        transforms.ToTensor(),
    ])
    transform_test = transforms.ToTensor()

    train_loader = torch.utils.data.DataLoader(
        datasets.CIFAR10(path, train=True, download=download, transform=transform_train),
        batch_size=batch_size, shuffle=True, **kwargs)
    test_loader = torch.utils.data.DataLoader(
        datasets.CIFAR10(path, train=False, download=download, transform=transform_test),
        batch_size=batch_size, shuffle=True, **kwargs)
    return train_loader, test_loader

# ---------------------------
#  Base Capsule Network (CIFAR-10 version)
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsuleBase, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        # Input: 256 channels from conv1, output: num_maps * num_dims channels
        self.conv = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        # x shape: (batch, 256, 24, 24) after conv1
        out = self.conv(x)                     # (batch, 256, 8, 8)   [256 = 32*8]
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 8, 8)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 8, 8, 8)
        squashed = squash(out, axis=-1)
        # Flatten spatial and map dimensions
        return squashed.view(batch_size, -1, self.num_dims)  # (batch, 32*8*8=2048, 8)

class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super(DenseCapsule, self).__init__()
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in,
                                                       num_dims_out, num_dims_in))
        self.routings = routings
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out

    def forward(self, x):
        x = x[:, None, :, :, None]            # (batch, 1, num_caps_in, num_dims_in, 1)
        x_hat = torch.squeeze(torch.matmul(self.weight, x), dim=-1)  # (num_caps_out, batch, num_caps_in, num_dims_out)
        x_hat_detached = x_hat.detach()
        b = Variable(torch.zeros(x.shape[0], self.num_caps_out, self.num_caps_in).cuda())

        assert self.routings > 0
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            if i == self.routings - 1:
                out = squash(torch.sum(c[:, :, :, None] * x_hat, dim=-2, keepdim=True))
            else:
                out = squash(torch.sum(c[:, :, :, None] * x_hat_detached, dim=-2, keepdim=True))
                b = b + torch.sum(out * x_hat_detached, dim=-1)
        return torch.squeeze(out, dim=-2)

class CapsuleNetBase(nn.Module):
    def __init__(self, input_size, classes, routings):
        super(CapsuleNetBase, self).__init__()
        self.input_size = input_size   # (C, H, W)
        self.classes = classes
        self.routings = routings
        # Conv1: 3x32x32 -> 256x24x24
        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsuleBase()
        # Number of primary capsules: 32 maps * 8*8 spatial = 2048
        self.digitcaps = DenseCapsule(
            num_caps_in=32*8*8,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        # Decoder: input 16*10 = 160, output 3*32*32 = 3072
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        # Binary foreground masking (simple threshold)
        mask_img = (x > 0.1).float()
        x_fg = x * mask_img

        out = self.relu(self.conv1(x_fg))        # (batch, 256, 24, 24)
        out = self.primarycaps(out)               # (batch, 2048, 8)
        out = self.digitcaps(out)                 # (batch, 10, 16)

        length = out.norm(dim=-1)                 # (batch, 10)

        if y is None:  # test mode
            index = length.max(dim=1)[1]
            y = torch.zeros(length.size()).scatter_(1, index.view(-1, 1).cpu(), 1.).cuda()

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)

# ---------------------------
#  CapFG Capsule Network (CIFAR-10 version)
# ---------------------------
class PrimaryCapsuleCapFG(nn.Module):
    def __init__(self, num_maps=32, num_dims=8):
        super(PrimaryCapsuleCapFG, self).__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(256, self.num_maps * self.num_dims,
                               kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)                         # (batch, 256, 8, 8)
        batch_size = out.size(0)
        out = out.view(batch_size, self.num_maps, self.num_dims, 8, 8)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (batch, 32, 8, 8, 8)
        raw_norms = torch.norm(out, dim=-1)                # (batch, 32, 8, 8)
        squashed = squash(out, axis=-1)                     # (batch, 32, 8, 8, 8)
        flat_squashed = squashed.view(batch_size, -1, self.num_dims)  # (batch, 2048, 8)
        return flat_squashed, raw_norms

class CapFG(nn.Module):
    def __init__(self, in_maps=32, grid_size=8):
        super(CapFG, self).__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, raw_norms):
        mask = self.conv(raw_norms)          # (batch, 1, grid_size, grid_size)
        return mask.squeeze(1)                # (batch, grid_size, grid_size)

class CapsuleNetCapFG(nn.Module):
    def __init__(self, input_size, classes, routings):
        super(CapsuleNetCapFG, self).__init__()
        self.input_size = input_size
        self.classes = classes
        self.routings = routings
        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.primarycaps = PrimaryCapsuleCapFG()
        self.capfg = CapFG(in_maps=32, grid_size=8)
        self.digitcaps = DenseCapsule(
            num_caps_in=32*8*8,
            num_dims_in=8,
            num_caps_out=classes,
            num_dims_out=16,
            routings=routings
        )
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU()

    def forward(self, x, y=None):
        # Binary foreground masking
        mask_img = (x > 0.1).float()
        x_fg = x * mask_img

        out = self.relu(self.conv1(x_fg))           # (batch, 256, 24, 24)
        prim_flat, raw_norms = self.primarycaps(out)  # (batch,2048,8) , (batch,32,8,8)

        # CapFG: compute soft spatial mask and apply gating
        A = self.capfg(raw_norms)                     # (batch, 8, 8)
        batch_size = prim_flat.size(0)
        prim_spatial = prim_flat.view(batch_size, 32, 8, 8, 8)
        A = A.unsqueeze(1).unsqueeze(-1)               # (batch, 1, 8, 8, 1)
        prim_gated = prim_spatial * A
        prim_gated_flat = prim_gated.view(batch_size, -1, 8)

        out = self.digitcaps(prim_gated_flat)          # (batch, 10, 16)

        length = out.norm(dim=-1)

        if y is None:
            index = length.max(dim=1)[1]
            y = torch.zeros(length.size()).scatter_(1, index.view(-1, 1).cpu(), 1.).cuda()

        reconstruction = self.decoder(
            (out * y[:, :, None]).view(out.size(0), -1)
        )
        return length, reconstruction.view(-1, *self.input_size)

# ---------------------------
#  Training and evaluation functions
# ---------------------------
def test(model, test_loader, lam_recon, return_predictions=False):
    model.eval()
    test_loss = 0
    correct = 0
    all_preds = []
    all_labels = []
    for x, y in test_loader:
        y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
        x, y_onehot = x.cuda(), y_onehot.cuda()
        with torch.no_grad():
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)
        test_loss += loss.item() * x.size(0)
        pred_class = y_pred.data.max(1)[1]
        correct += pred_class.eq(y.cuda()).cpu().sum()
        if return_predictions:
            all_preds.extend(pred_class.cpu().numpy())
            all_labels.extend(y.numpy())

    test_loss /= len(test_loader.dataset)
    accuracy = correct.item() / len(test_loader.dataset)
    if return_predictions:
        return test_loss, accuracy, np.array(all_preds), np.array(all_labels)
    return test_loss, accuracy

def train_model(model, train_loader, test_loader, args, model_name):
    save_dir = args["save_dir"]
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)

    print(f"\n--- Training {model_name} on CIFAR-10 ---")
    logfile = open(os.path.join(save_dir, 'log.csv'), 'w')
    logwriter = csv.DictWriter(logfile, fieldnames=['epoch', 'loss', 'val_loss', 'val_acc'])
    logwriter.writeheader()

    optimizer = Adam(model.parameters(), lr=args["lr"])
    lr_decay = lr_scheduler.ExponentialLR(optimizer, gamma=args["lr_decay"])
    best_val_acc = 0.

    history = {'train_loss': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(args["epochs"]):
        model.train()
        ti = time()
        training_loss = 0.0
        for i, (x, y) in enumerate(train_loader):
            y_onehot = torch.zeros(y.size(0), 10).scatter_(1, y.view(-1, 1), 1.)
            x, y_onehot = x.cuda(), y_onehot.cuda()

            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, args["lam_recon"])
            loss.backward()
            training_loss += loss.item() * x.size(0)
            optimizer.step()

        lr_decay.step()

        val_loss, val_acc = test(model, test_loader, args["lam_recon"])
        avg_train_loss = training_loss / len(train_loader.dataset)

        history['train_loss'].append(avg_train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        logwriter.writerow(dict(epoch=epoch, loss=avg_train_loss,
                                val_loss=val_loss, val_acc=val_acc))
        print(f"Epoch {epoch:02d}: loss={avg_train_loss:.5f}, val_loss={val_loss:.5f}, val_acc={val_acc:.4f}, time={int(time()-ti)}s")

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(save_dir, f'epoch{epoch}.pkl'))
            print(f"  best val_acc increased to {best_val_acc:.4f}")

    logfile.close()
    torch.save(model.state_dict(), os.path.join(save_dir, 'trained_model.pkl'))
    print(f"{model_name} training finished. Model saved to {save_dir}/trained_model.pkl")
    return history

def plot_comparison(log_base, log_capfg, save_path='comparison_plots.png'):
    df_base = pd.read_csv(log_base)
    df_capfg = pd.read_csv(log_capfg)

    epochs_base = df_base['epoch'] + 1
    epochs_capfg = df_capfg['epoch'] + 1

    plt.style.use('seaborn-v0_8-darkgrid')
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    # Training loss
    axes[0].plot(epochs_base, df_base['loss'], 'b-', label='Base')
    axes[0].plot(epochs_capfg, df_capfg['loss'], 'r-', label='CapFG')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title('Training Loss')
    axes[0].legend()
    axes[0].grid(True)

    # Validation loss
    axes[1].plot(epochs_base, df_base['val_loss'], 'b--', label='Base')
    axes[1].plot(epochs_capfg, df_capfg['val_loss'], 'r--', label='CapFG')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].set_title('Validation Loss')
    axes[1].legend()
    axes[1].grid(True)

    # Validation accuracy
    axes[2].plot(epochs_base, df_base['val_acc'], 'b-.', label='Base')
    axes[2].plot(epochs_capfg, df_capfg['val_acc'], 'r-.', label='CapFG')
    axes[2].set_xlabel('Epoch')
    axes[2].set_ylabel('Accuracy')
    axes[2].set_title('Validation Accuracy')
    axes[2].legend()
    axes[2].grid(True)

    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.show()
    print(f"Comparison plots saved as '{save_path}'")

# ---------------------------
#  Main script
# ---------------------------
if __name__ == "__main__":
    # Settings
    input_size = (3, 32, 32)
    args = {
        "epochs": 50,
        "batch_size": 100,
        "lr": 0.001,
        "lr_decay": 0.9,
        "lam_recon": 0.0005 * np.prod(input_size),  # scale with number of pixels
        "routings": 3,
        "shift_pixels": 4,
        "data_dir": "./CIFAR10",
    }

    # Directories for results
    base_dir = "./cifar10_result_base"
    capfg_dir = "./cifar10_result_capfg"
    args_base = args.copy()
    args_base["save_dir"] = base_dir
    args_capfg = args.copy()
    args_capfg["save_dir"] = capfg_dir

    # Load data
    print("Loading CIFAR-10...")
    train_loader, test_loader = load_cifar10(args["data_dir"], download=True,
                                              batch_size=args["batch_size"],
                                              shift_pixels=args["shift_pixels"])

    # Option to skip training if logs already exist (set to False to force retrain)
    skip_if_exists = True

    # --- Train Base Model ---
    log_base_path = os.path.join(base_dir, 'log.csv')
    if not skip_if_exists or not os.path.exists(log_base_path):
        print("\nInitializing Base Model...")
        model_base = CapsuleNetBase(input_size, classes=10, routings=args["routings"])
        model_base.cuda()
        train_model(model_base, train_loader, test_loader, args_base, "Base Capsule")

        # After training, generate confusion matrix and reconstructions for base model
        print("\n--- Evaluating Base Model ---")
        _, _, y_pred, y_true = test(model_base, test_loader, args["lam_recon"], return_predictions=True)
        plot_confusion_matrix(y_true, y_pred, base_dir, "Base")
        show_reconstruction(model_base, test_loader, 50, base_dir, "Base")
    else:
        print(f"Base model log found at {log_base_path}, skipping training.")

    # --- Train CapFG Model ---
    log_capfg_path = os.path.join(capfg_dir, 'log.csv')
    if not skip_if_exists or not os.path.exists(log_capfg_path):
        print("\nInitializing CapFG Model...")
        model_capfg = CapsuleNetCapFG(input_size, classes=10, routings=args["routings"])
        model_capfg.cuda()
        train_model(model_capfg, train_loader, test_loader, args_capfg, "CapFG Capsule")

        # After training, generate confusion matrix and reconstructions for CapFG model
        print("\n--- Evaluating CapFG Model ---")
        _, _, y_pred, y_true = test(model_capfg, test_loader, args["lam_recon"], return_predictions=True)
        plot_confusion_matrix(y_true, y_pred, capfg_dir, "CapFG")
        show_reconstruction(model_capfg, test_loader, 50, capfg_dir, "CapFG")
    else:
        print(f"CapFG model log found at {log_capfg_path}, skipping training.")

    # --- Comparison plots ---
    if os.path.exists(log_base_path) and os.path.exists(log_capfg_path):
        plot_comparison(log_base_path, log_capfg_path, save_path='cifar10_comparison.png')
    else:
        print("One or both log files are missing. Cannot generate comparison plots.")
```

```python
# cifar10_baseline_vs_capfg.py

import os
import csv
import math
import random
from time import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam, lr_scheduler
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms

# ---------------------------
# Reproducibility
# ---------------------------
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ---------------------------
# Common utilities
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = (norm ** 2) / (1.0 + norm ** 2)
    return scale * inputs / (norm + 1e-8)

def to_onehot(y, num_classes):
    return torch.eye(num_classes, device=y.device)[y]

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    margin = (
        y_true * torch.clamp(0.9 - y_pred, min=0.0) ** 2 +
        0.5 * (1.0 - y_true) * torch.clamp(y_pred - 0.1, min=0.0) ** 2
    )
    margin_loss = margin.sum(dim=1).mean()
    recon_loss = F.mse_loss(x_recon, x)
    return margin_loss + lam_recon * recon_loss

# ---------------------------
# CIFAR-10 loaders
# ---------------------------
def load_cifar10(data_dir="./CIFAR10", batch_size=128, val_size=5000):
    train_transform = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ])

    test_transform = transforms.Compose([
        transforms.ToTensor(),
    ])

    full_train = datasets.CIFAR10(
        root=data_dir,
        train=True,
        download=True,
        transform=train_transform
    )

    test_set = datasets.CIFAR10(
        root=data_dir,
        train=False,
        download=True,
        transform=test_transform
    )

    train_len = len(full_train) - val_size
    val_len = val_size
    train_set, val_set = random_split(
        full_train,
        [train_len, val_len],
        generator=torch.Generator().manual_seed(42)
    )

    # validation should not use random augmentation
    val_set.dataset = datasets.CIFAR10(
        root=data_dir,
        train=True,
        download=False,
        transform=test_transform
    )

    # keep same indices from split
    val_set = torch.utils.data.Subset(val_set.dataset, val_set.indices)

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
    val_loader   = DataLoader(val_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
    test_loader  = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)

    return train_loader, val_loader, test_loader

# ---------------------------
# Capsule blocks
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)  # (B, 256, 8, 8) for CIFAR-10 after conv1
        bsz = out.size(0)
        h, w = out.size(2), out.size(3)  # expected 8, 8
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()  # (B, 32, H, W, 8)
        out = squash(out, axis=-1)
        return out.view(bsz, -1, self.num_dims), h, w  # (B, 32*H*W, 8)

class PrimaryCapsuleCapFG(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)
        bsz = out.size(0)
        h, w = out.size(2), out.size(3)
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (B, 32, H, W, 8)
        raw_norms = torch.norm(out, dim=-1)             # (B, 32, H, W)
        squashed = squash(out, axis=-1)                 # (B, 32, H, W, 8)
        return squashed, raw_norms, h, w

class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super().__init__()
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out
        self.num_dims_in = num_dims_in
        self.num_dims_out = num_dims_out
        self.routings = routings

        self.weight = nn.Parameter(
            0.01 * torch.randn(num_caps_out, num_caps_in, num_dims_out, num_dims_in)
        )

    def forward(self, x):
        # x: (B, num_caps_in, num_dims_in)
        # result: (B, num_caps_out, num_caps_in, num_dims_out)
        x = x[:, None, :, :, None]
        x_hat = torch.matmul(self.weight[None, :, :, :, :], x).squeeze(-1)

        b = torch.zeros(x.size(0), self.num_caps_out, self.num_caps_in, device=x.device)

        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            s = (c[:, :, :, None] * x_hat).sum(dim=2)
            v = squash(s, axis=-1)
            if i < self.routings - 1:
                b = b + (x_hat * v[:, :, None, :]).sum(dim=-1)

        return v  # (B, num_caps_out, num_dims_out)

# ---------------------------
# Baseline CapsNet (TRUE baseline: no binary mask)
# ---------------------------
class CapsuleNetBase(nn.Module):
    def __init__(self, input_size=(3, 32, 32), classes=10, routings=3):
        super().__init__()
        self.input_size = input_size
        self.classes = classes

        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)
        self.primarycaps = PrimaryCapsuleBase(in_channels=256, num_maps=32, num_dims=8)

        # CIFAR-10 geometry:
        # 32x32 -> conv1(k9,s1) => 24x24
        # primarycaps conv(k9,s2) => 8x8
        self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))  # NO binary threshold mask here
        prim_flat, h, w = self.primarycaps(out)
        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.input_size)
        return length, recon

# ---------------------------
# CapFG model
# ---------------------------
class CapFG(nn.Module):
    def __init__(self, in_maps=32):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, raw_norms):
        mask = self.conv(raw_norms)   # (B, 1, H, W)
        return mask

class CapsuleNetCapFG(nn.Module):
    def __init__(self, input_size=(3, 32, 32), classes=10, routings=3):
        super().__init__()
        self.input_size = input_size
        self.classes = classes

        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        self.primarycaps = PrimaryCapsuleCapFG(in_channels=256, num_maps=32, num_dims=8)
        self.capfg = CapFG(in_maps=32)

        self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))
        prim_spatial, raw_norms, h, w = self.primarycaps(out)  # prim_spatial: (B,32,H,W,8)

        A = self.capfg(raw_norms)  # (B,1,H,W)
        A = A.unsqueeze(-1)        # (B,1,H,W,1)

        prim_gated = prim_spatial * A
        prim_flat = prim_gated.view(x.size(0), -1, 8)

        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.input_size)
        return length, recon

# ---------------------------
# Evaluation
# ---------------------------
def evaluate(model, loader, lam_recon, num_classes=10):
    model.eval()
    total_loss = 0.0
    total_correct = 0
    total = 0

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, num_classes)

            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)

            total_loss += loss.item() * x.size(0)
            preds = y_pred.argmax(dim=1)
            total_correct += (preds == y).sum().item()
            total += x.size(0)

    return total_loss / total, total_correct / total

# ---------------------------
# Training
# ---------------------------
def train_model(model, train_loader, val_loader, config, save_dir, model_name):
    os.makedirs(save_dir, exist_ok=True)

    optimizer = Adam(model.parameters(), lr=config["lr"])
    scheduler = lr_scheduler.ExponentialLR(optimizer, gamma=config["lr_decay"])

    best_val_acc = 0.0
    history = []

    log_path = os.path.join(save_dir, "log.csv")
    with open(log_path, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["epoch", "train_loss", "val_loss", "val_acc"])
        writer.writeheader()

        for epoch in range(config["epochs"]):
            model.train()
            t0 = time()
            running_loss = 0.0
            total = 0

            for x, y in train_loader:
                x, y = x.to(device), y.to(device)
                y_onehot = to_onehot(y, config["classes"])

                optimizer.zero_grad()
                y_pred, x_recon = model(x, y_onehot)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"])
                loss.backward()
                optimizer.step()

                running_loss += loss.item() * x.size(0)
                total += x.size(0)

            scheduler.step()

            train_loss = running_loss / total
            val_loss, val_acc = evaluate(model, val_loader, config["lam_recon"], config["classes"])

            row = {
                "epoch": epoch + 1,
                "train_loss": train_loss,
                "val_loss": val_loss,
                "val_acc": val_acc
            }
            writer.writerow(row)
            history.append(row)

            print(
                f"{model_name} | Epoch {epoch+1:02d}/{config['epochs']} | "
                f"train_loss={train_loss:.4f} | val_loss={val_loss:.4f} | "
                f"val_acc={val_acc:.4f} | time={int(time()-t0)}s"
            )

            if val_acc > best_val_acc:
                best_val_acc = val_acc
                torch.save(model.state_dict(), os.path.join(save_dir, "best_model.pt"))

    print(f"{model_name} best val_acc = {best_val_acc:.4f}")
    return pd.DataFrame(history)

# ---------------------------
# Plotting
# ---------------------------
def plot_comparison(base_log, capfg_log, save_path="cifar10_baseline_vs_capfg.png"):
    df_base = pd.read_csv(base_log)
    df_capfg = pd.read_csv(capfg_log)

    fig, axes = plt.subplots(1, 3, figsize=(16, 5))

    axes[0].plot(df_base["epoch"], df_base["train_loss"], label="Baseline")
    axes[0].plot(df_capfg["epoch"], df_capfg["train_loss"], label="CapFG")
    axes[0].set_title("Training Loss")
    axes[0].set_xlabel("Epoch")

    axes[1].plot(df_base["epoch"], df_base["val_loss"], label="Baseline")
    axes[1].plot(df_capfg["epoch"], df_capfg["val_loss"], label="CapFG")
    axes[1].set_title("Validation Loss")
    axes[1].set_xlabel("Epoch")

    axes[2].plot(df_base["epoch"], df_base["val_acc"], label="Baseline")
    axes[2].plot(df_capfg["epoch"], df_capfg["val_acc"], label="CapFG")
    axes[2].set_title("Validation Accuracy")
    axes[2].set_xlabel("Epoch")

    for ax in axes:
        ax.legend()
        ax.grid(True)

    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.show()
    print(f"Saved plot to {save_path}")

# ---------------------------
# Main
# ---------------------------
if __name__ == "__main__":
    config = {
        "epochs": 50,
        "batch_size": 128,
        "lr": 0.001,
        "lr_decay": 0.9,
        "lam_recon": 0.0005 * (3 * 32 * 32),
        "routings": 3,
        "classes": 10,
        "data_dir": "./CIFAR10"
    }

    train_loader, val_loader, test_loader = load_cifar10(
        data_dir=config["data_dir"],
        batch_size=config["batch_size"],
        val_size=5000
    )

    # ---- TRUE baseline ----
    base_dir = "./result_cifar10_base_true"
    model_base = CapsuleNetBase(
        input_size=(3, 32, 32),
        classes=config["classes"],
        routings=config["routings"]
    ).to(device)

    train_model(
        model=model_base,
        train_loader=train_loader,
        val_loader=val_loader,
        config=config,
        save_dir=base_dir,
        model_name="Baseline CapsNet"
    )

    model_base.load_state_dict(torch.load(os.path.join(base_dir, "best_model.pt"), map_location=device))
    test_loss_base, test_acc_base = evaluate(model_base, test_loader, config["lam_recon"], config["classes"])
    print(f"\nBaseline CapsNet TEST: loss={test_loss_base:.4f}, acc={test_acc_base:.4f}")

    # ---- CapFG ----
    capfg_dir = "./result_cifar10_capfg_only"
    model_capfg = CapsuleNetCapFG(
        input_size=(3, 32, 32),
        classes=config["classes"],
        routings=config["routings"]
    ).to(device)

    train_model(
        model=model_capfg,
        train_loader=train_loader,
        val_loader=val_loader,
        config=config,
        save_dir=capfg_dir,
        model_name="CapFG CapsNet"
    )

    model_capfg.load_state_dict(torch.load(os.path.join(capfg_dir, "best_model.pt"), map_location=device))
    test_loss_capfg, test_acc_capfg = evaluate(model_capfg, test_loader, config["lam_recon"], config["classes"])
    print(f"CapFG CapsNet TEST: loss={test_loss_capfg:.4f}, acc={test_acc_capfg:.4f}")

    # ---- Save summary ----
    summary = pd.DataFrame([
        {"model": "Baseline CapsNet", "test_loss": test_loss_base, "test_acc": test_acc_base},
        {"model": "CapFG CapsNet", "test_loss": test_loss_capfg, "test_acc": test_acc_capfg},
    ])
    summary.to_csv("cifar10_test_summary.csv", index=False)
    print("\nSaved cifar10_test_summary.csv")

    plot_comparison(
        os.path.join(base_dir, "log.csv"),
        os.path.join(capfg_dir, "log.csv"),
        save_path="cifar10_baseline_vs_capfg.png"
    )
```

```python
# improved_capsule_experiments.py
import os
import csv
import math
import random
import argparse
from time import time
from collections import defaultdict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam, lr_scheduler
from torch.utils.data import DataLoader, random_split, Dataset
from torchvision import datasets, transforms, models
from torchvision.datasets import ImageFolder
from torchvision.transforms.functional import to_pil_image

# For complexity (optional)
try:
    from thop import profile, clever_format
    THOP_AVAILABLE = True
except ImportError:
    THOP_AVAILABLE = False

# ---------------------------
# Reproducibility
# ---------------------------
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

# ---------------------------
# Device
# ---------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ---------------------------
# Helper functions & losses
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = (norm ** 2) / (1.0 + norm ** 2)
    return scale * inputs / (norm + 1e-8)

def to_onehot(y, num_classes):
    return torch.eye(num_classes, device=y.device)[y]

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    margin = (
        y_true * torch.clamp(0.9 - y_pred, min=0.0) ** 2 +
        0.5 * (1.0 - y_true) * torch.clamp(y_pred - 0.1, min=0.0) ** 2
    )
    margin_loss = margin.sum(dim=1).mean()
    recon_loss = F.mse_loss(x_recon, x)
    return margin_loss + lam_recon * recon_loss

# ---------------------------
# Datasets (including synthetic and CIFAR-10-C)
# ---------------------------
def load_dataset(dataset_name, data_dir="./data", batch_size=128, val_size=5000):
    """Returns train_loader, val_loader, test_loader for given dataset."""
    if dataset_name == "cifar10":
        train_transform = transforms.Compose([
            transforms.RandomCrop(32, padding=4),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
        ])
        test_transform = transforms.Compose([transforms.ToTensor()])
        full_train = datasets.CIFAR10(root=data_dir, train=True, download=True, transform=train_transform)
        test_set = datasets.CIFAR10(root=data_dir, train=False, download=True, transform=test_transform)
        n_classes = 10
        input_shape = (3, 32, 32)
    elif dataset_name == "fashion_mnist":
        train_transform = transforms.Compose([
            transforms.Resize(32),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
        ])
        test_transform = transforms.Compose([transforms.Resize(32), transforms.ToTensor()])
        full_train = datasets.FashionMNIST(root=data_dir, train=True, download=True, transform=train_transform)
        test_set = datasets.FashionMNIST(root=data_dir, train=False, download=True, transform=test_transform)
        n_classes = 10
        input_shape = (1, 32, 32)
    elif dataset_name == "svhn":
        train_transform = transforms.Compose([
            transforms.RandomCrop(32, padding=4),
            transforms.ToTensor(),
        ])
        test_transform = transforms.Compose([transforms.ToTensor()])
        full_train = datasets.SVHN(root=data_dir, split='train', download=True, transform=train_transform)
        test_set = datasets.SVHN(root=data_dir, split='test', download=True, transform=test_transform)
        n_classes = 10
        input_shape = (3, 32, 32)
    elif dataset_name == "cifar10_c":
        # Corruption level 3 as a simple domain shift test
        # Download from https://github.com/hendrycks/robustness
        # Here we just create a wrapper that loads corrupted images
        raise NotImplementedError("Please manually download CIFAR-10-C and point to it.")
    else:
        raise ValueError(f"Unsupported dataset: {dataset_name}")

    # Create train/val split
    train_len = len(full_train) - val_size
    val_len = val_size
    train_set, val_set = random_split(full_train, [train_len, val_len], generator=torch.Generator().manual_seed(42))
    # Fix validation transform (no augmentation)
    if hasattr(val_set, 'dataset'):
        val_set.dataset.transform = test_transform
    else:
        val_set.dataset.transform = test_transform

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
    val_loader   = DataLoader(val_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
    test_loader  = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
    return train_loader, val_loader, test_loader, n_classes, input_shape

class SyntheticForegroundDataset(Dataset):
    """Creates synthetic data with known foreground/background mask."""
    def __init__(self, num_samples=5000, size=32, num_classes=10, bg_noise_std=0.5):
        self.num_samples = num_samples
        self.size = size
        self.num_classes = num_classes
        self.bg_noise_std = bg_noise_std
        # Use a small set of random shapes as "foreground"
        self.images = []
        self.labels = []
        self.masks = []
        for _ in range(num_samples):
            img = torch.randn(3, size, size) * bg_noise_std  # random background
            label = random.randint(0, num_classes-1)
            # create a square foreground at random position
            fg_size = random.randint(8, 16)
            fg_x = random.randint(0, size - fg_size)
            fg_y = random.randint(0, size - fg_size)
            mask = torch.zeros(size, size)
            mask[fg_y:fg_y+fg_size, fg_x:fg_x+fg_size] = 1.0
            fg_value = torch.randn(3, fg_size, fg_size) * 0.2 + 0.8  # brighter foreground
            img[:, fg_y:fg_y+fg_size, fg_x:fg_x+fg_size] = fg_value
            self.images.append(img)
            self.labels.append(label)
            self.masks.append(mask)
    def __len__(self):
        return self.num_samples
    def __getitem__(self, idx):
        return self.images[idx], self.labels[idx], self.masks[idx].unsqueeze(0)  # (1, H, W)

# ---------------------------
# Capsule building blocks (unchanged from your original, but with improved masks)
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)
    def forward(self, x):
        out = self.conv(x)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()
        out = squash(out, axis=-1)
        return out.view(bsz, -1, self.num_dims), h, w

class PrimaryCapsuleCapFG(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)
    def forward(self, x):
        out = self.conv(x)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()
        raw_norms = torch.norm(out, dim=-1)
        squashed = squash(out, axis=-1)
        return squashed, raw_norms, h, w

class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super().__init__()
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out
        self.routings = routings
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in, num_dims_out, num_dims_in))
    def forward(self, x):
        x = x[:, None, :, :, None]
        x_hat = torch.matmul(self.weight[None, :, :, :, :], x).squeeze(-1)
        b = torch.zeros(x.size(0), self.num_caps_out, self.num_caps_in, device=x.device)
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            s = (c[:, :, :, None] * x_hat).sum(dim=2)
            v = squash(s, axis=-1)
            if i < self.routings - 1:
                b = b + (x_hat * v[:, :, None, :]).sum(dim=-1)
        return v

class CapFG(nn.Module):
    """Learnable foreground mask generator (no fixed threshold)."""
    def __init__(self, in_maps=32):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )
    def forward(self, raw_norms):
        return self.conv(raw_norms)  # (B,1,H,W)

class CapsuleNetBase(nn.Module):
    def __init__(self, input_shape=(3,32,32), classes=10, routings=3):
        super().__init__()
        self.input_shape = input_shape
        self.classes = classes
        self.conv1 = nn.Conv2d(input_shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)
        self.primarycaps = PrimaryCapsuleBase(in_channels=256, num_maps=32, num_dims=8)
        self.num_caps_in = 32 * 8 * 8
        self.digitcaps = DenseCapsule(self.num_caps_in, classes, 8, 16, routings)
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512), nn.ReLU(inplace=True),
            nn.Linear(512, 1024), nn.ReLU(inplace=True),
            nn.Linear(1024, input_shape[0]*input_shape[1]*input_shape[2]), nn.Sigmoid()
        )
    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))
        prim_flat, h, w = self.primarycaps(out)
        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)
        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]
        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.input_shape)
        return length, recon

class CapsuleNetCapFG(nn.Module):
    def __init__(self, input_shape=(3,32,32), classes=10, routings=3):
        super().__init__()
        self.input_shape = input_shape
        self.classes = classes
        self.conv1 = nn.Conv2d(input_shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)
        self.primarycaps = PrimaryCapsuleCapFG(in_channels=256, num_maps=32, num_dims=8)
        self.capfg = CapFG(in_maps=32)
        self.num_caps_in = 32 * 8 * 8
        self.digitcaps = DenseCapsule(self.num_caps_in, classes, 8, 16, routings)
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512), nn.ReLU(inplace=True),
            nn.Linear(512, 1024), nn.ReLU(inplace=True),
            nn.Linear(1024, input_shape[0]*input_shape[1]*input_shape[2]), nn.Sigmoid()
        )
    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))
        prim_spatial, raw_norms, h, w = self.primarycaps(out)  # (B,32,H,W,8)
        A = self.capfg(raw_norms)  # (B,1,H,W)
        A = A.unsqueeze(-1)        # (B,1,H,W,1)
        prim_gated = prim_spatial * A
        prim_flat = prim_gated.view(x.size(0), -1, 8)
        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)
        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]
        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.input_shape)
        return length, recon

# ---------------------------
# Benchmark models (CNN+attention, ViT, modern capsule variants)
# ---------------------------
class ResNetWithCBAM(nn.Module):
    """ResNet18 with Convolutional Block Attention Module."""
    def __init__(self, num_classes=10, input_channels=3):
        super().__init__()
        self.resnet = models.resnet18(weights=None)
        self.resnet.conv1 = nn.Conv2d(input_channels, 64, kernel_size=7, stride=2, padding=3, bias=False)
        self.resnet.fc = nn.Linear(512, num_classes)
        # Insert CBAM after each residual block (simplified)
        from torchvision.ops import Conv2dNormActivation
        class CBAM(nn.Module):
            def __init__(self, channels):
                super().__init__()
                self.avg_pool = nn.AdaptiveAvgPool2d(1)
                self.max_pool = nn.AdaptiveMaxPool2d(1)
                self.fc = nn.Sequential(
                    nn.Linear(channels, channels//16, bias=False),
                    nn.ReLU(),
                    nn.Linear(channels//16, channels, bias=False)
                )
                self.spatial = nn.Conv2d(2, 1, kernel_size=7, padding=3, bias=False)
            def forward(self, x):
                avg_out = self.fc(self.avg_pool(x).view(x.size(0), -1)).unsqueeze(-1).unsqueeze(-1)
                max_out = self.fc(self.max_pool(x).view(x.size(0), -1)).unsqueeze(-1).unsqueeze(-1)
                channel_att = torch.sigmoid(avg_out + max_out)
                x = x * channel_att
                spatial_in = torch.cat([torch.mean(x, dim=1, keepdim=True), torch.max(x, dim=1, keepdim=True)[0]], dim=1)
                spatial_att = torch.sigmoid(self.spatial(spatial_in))
                return x * spatial_att
        # Insert CBAM after each residual block
        for name, module in self.resnet.named_children():
            if isinstance(module, nn.Sequential) and len(module) > 0 and isinstance(module[0], nn.Conv2d):
                module.add_module('cbam', CBAM(module[-1].bn2.num_features if hasattr(module[-1], 'bn2') else 512))
    def forward(self, x):
        return self.resnet(x)

class SimpleViT(nn.Module):
    """Minimal ViT for small images, using torchvision's ViT."""
    def __init__(self, num_classes=10, input_channels=3, image_size=32):
        super().__init__()
        from torchvision.models.vision_transformer import vit_b_16
        self.vit = vit_b_16(weights=None, image_size=image_size)
        self.vit.conv_proj = nn.Conv2d(input_channels, 768, kernel_size=16, stride=16)
        self.vit.heads = nn.Linear(768, num_classes)
    def forward(self, x):
        return self.vit(x)

class EfficientCapsule(nn.Module):
    """Modern capsule variant from 'Efficient-CapsNet' style."""
    def __init__(self, input_shape=(3,32,32), classes=10):
        super().__init__()
        self.conv1 = nn.Conv2d(input_shape[0], 64, kernel_size=5, stride=2, padding=2)
        self.bn1 = nn.BatchNorm2d(64)
        self.conv2 = nn.Conv2d(64, 128, kernel_size=5, stride=2, padding=2)
        self.bn2 = nn.BatchNorm2d(128)
        self.primarycaps = PrimaryCapsuleBase(in_channels=128, num_maps=16, num_dims=8)
        self.digitcaps = DenseCapsule(num_caps_in=16*8*8, num_caps_out=classes, num_dims_in=8, num_dims_out=16)
        self.decoder = nn.Sequential(nn.Linear(16*classes, 512), nn.ReLU(), nn.Linear(512, input_shape[0]*input_shape[1]*input_shape[2]), nn.Sigmoid())
    def forward(self, x, y=None):
        x = F.relu(self.bn1(self.conv1(x)))
        x = F.relu(self.bn2(self.conv2(x)))
        prim, _, _ = self.primarycaps(x)
        out = self.digitcaps(prim)
        length = out.norm(dim=-1)
        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]
        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        return length, recon.view(-1, *input_shape)

# ---------------------------
# Training & evaluation (with multiple runs)
# ---------------------------
def evaluate(model, loader, lam_recon, num_classes):
    model.eval()
    total_loss, total_correct, total = 0.0, 0, 0
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, num_classes)
            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)
            total_loss += loss.item() * x.size(0)
            preds = y_pred.argmax(dim=1)
            total_correct += (preds == y).sum().item()
            total += x.size(0)
    return total_loss / total, total_correct / total

def train_one_run(model, train_loader, val_loader, config, run_id=0, model_name="model"):
    optimizer = Adam(model.parameters(), lr=config["lr"])
    scheduler = lr_scheduler.ExponentialLR(optimizer, gamma=config["lr_decay"])
    best_val_acc = 0.0
    for epoch in range(config["epochs"]):
        model.train()
        running_loss, total = 0.0, 0
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, config["classes"])
            optimizer.zero_grad()
            y_pred, x_recon = model(x, y_onehot)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"])
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * x.size(0)
            total += x.size(0)
        scheduler.step()
        val_loss, val_acc = evaluate(model, val_loader, config["lam_recon"], config["classes"])
        if val_acc > best_val_acc:
            best_val_acc = val_acc
    return best_val_acc

def run_multiple_runs(model_class, model_kwargs, config, train_loader, val_loader, n_runs=5):
    accuracies = []
    for run in range(n_runs):
        seed = 42 + run
        set_seed(seed)
        model = model_class(**model_kwargs).to(device)
        acc = train_one_run(model, train_loader, val_loader, config, run, model_class.__name__)
        accuracies.append(acc)
        print(f"  Run {run+1}: val_acc = {acc:.4f}")
    mean_acc = np.mean(accuracies)
    std_acc = np.std(accuracies)
    return accuracies, mean_acc, std_acc

# ---------------------------
# Direct background suppression evaluation
# ---------------------------
def evaluate_background_suppression(model, synthetic_loader, device, lam_recon, num_classes):
    """Computes foreground IOU, background leakage, and accuracy drop under background noise."""
    model.eval()
    ious = []
    bg_leakage = []
    correct_clean = 0
    total = 0
    with torch.no_grad():
        for x, y, true_mask in synthetic_loader:
            x, y = x.to(device), y.to(device)
            true_mask = true_mask.to(device)
            # Get predicted mask from CapFG model if available
            if hasattr(model, 'capfg'):
                _, raw_norms, _, _ = model.primarycaps(model.relu(model.conv1(x)))
                pred_mask = model.capfg(raw_norms)  # (B,1,H,W)
                pred_mask_bin = (pred_mask > 0.5).float()
                intersection = (pred_mask_bin * true_mask).sum(dim=(1,2,3))
                union = ((pred_mask_bin + true_mask) > 0.5).float().sum(dim=(1,2,3))
                iou = (intersection + 1e-6) / (union + 1e-6)
                ious.extend(iou.cpu().numpy())
                bg_leak = (pred_mask_bin * (1 - true_mask)).sum(dim=(1,2,3)) / ((1 - true_mask).sum(dim=(1,2,3)) + 1e-6)
                bg_leakage.extend(bg_leak.cpu().numpy())
            # Accuracy
            y_pred, _ = model(x)
            correct_clean += (y_pred.argmax(dim=1) == y).sum().item()
            total += x.size(0)
    clean_acc = correct_clean / total
    # Also test with background replaced by noise
    return {"foreground_iou": np.mean(ious) if ious else None,
            "background_leakage": np.mean(bg_leakage) if bg_leakage else None,
            "clean_accuracy": clean_acc}

# ---------------------------
# Complexity analysis
# ---------------------------
def measure_complexity(model, input_shape, device):
    model.eval()
    dummy_input = torch.randn(1, *input_shape).to(device)
    # Parameters
    params = sum(p.numel() for p in model.parameters())
    # FLOPs
    flops = None
    if THOP_AVAILABLE:
        flops, _ = profile(model, inputs=(dummy_input,), verbose=False)
        flops = flops / 1e6  # MFLOPs
    # Inference time
    start = time()
    with torch.no_grad():
        for _ in range(100):
            _ = model(dummy_input)
    infer_time = (time() - start) / 100 * 1000  # ms per sample
    return params, flops, infer_time

# ---------------------------
# Ablation study
# ---------------------------
def run_ablation_study(config, train_loader, val_loader):
    results = {}
    # 1. CapFG alone (mask without routing modification)
    # 2. Binary mask (threshold=0.1)
    # 3. Remove gating (i.e., no mask)
    # 4. Threshold sweep
    thresholds = [0.05, 0.1, 0.2, 0.3]
    for th in thresholds:
        # For binary mask we modify forward of PrimaryCapsuleCapFG to apply static threshold on raw_norms
        results[f"binary_mask_th{th}"] = run_multiple_runs(...)  # omitted for brevity - implement similarly
    # 5. Controlled background noise (add noise to CIFAR-10 background using synthetic dataset)
    # 6. Adversarial background shift (replace background with random patches)
    # This function would be long but the structure is clear.
    return results

# ---------------------------
# Main experiment dispatcher
# ---------------------------
def main():
    # Allow unknown arguments (like -f from Jupyter)
    parser = argparse.ArgumentParser()
    parser.add_argument("--exp", type=str, default="baseline_vs_capfg",
                        choices=["baseline_vs_capfg", "benchmark", "ablation", "robustness", "complexity", "full"])
    parser.add_argument("--dataset", type=str, default="cifar10",
                        choices=["cifar10", "fashion_mnist", "svhn"])
    parser.add_argument("--n_runs", type=int, default=5, help="Number of runs for statistical validation")
    # Parse known args only, ignore extra (e.g., -f)
    args, unknown = parser.parse_known_args()
    if unknown:
        print(f"Ignoring unknown arguments: {unknown}")

    config = {
        "epochs": 50,
        "batch_size": 128,
        "lr": 0.001,
        "lr_decay": 0.9,
        "lam_recon": 0.0005 * (3 * 32 * 32),  # placeholder, will be updated per dataset
        "routings": 3,
        "classes": 10,
    }

    # Load dataset
    train_loader, val_loader, test_loader, n_classes, input_shape = load_dataset(
        args.dataset, batch_size=config["batch_size"])
    config["classes"] = n_classes
    config["lam_recon"] = 0.0005 * (input_shape[0] * input_shape[1] * input_shape[2])

    if args.exp == "baseline_vs_capfg":
        print("=== Baseline vs CapFG with multiple runs ===")
        base_accs, base_mean, base_std = run_multiple_runs(
            CapsuleNetBase, {"input_shape": input_shape, "classes": n_classes, "routings": config["routings"]},
            config, train_loader, val_loader, n_runs=args.n_runs)
        capfg_accs, capfg_mean, capfg_std = run_multiple_runs(
            CapsuleNetCapFG, {"input_shape": input_shape, "classes": n_classes, "routings": config["routings"]},
            config, train_loader, val_loader, n_runs=args.n_runs)
        # Statistical test
        t_stat, p_val = stats.ttest_ind(base_accs, capfg_accs)
        print(f"\nBaseline: {base_mean:.4f} ± {base_std:.4f}")
        print(f"CapFG:    {capfg_mean:.4f} ± {capfg_std:.4f}")
        print(f"Independent t-test: t={t_stat:.4f}, p={p_val:.4f} -> {'significant' if p_val<0.05 else 'not significant'}")
        # Save results
        pd.DataFrame({"Baseline": base_accs, "CapFG": capfg_accs}).to_csv("statistical_runs.csv", index=False)

    elif args.exp == "benchmark":
        print("=== Benchmark against SOTA models ===")
        # For capsule-based models we can use capsule loss; for CNN/ViT we need a standard cross-entropy trainer.
        # For brevity, we only show the structure. A full implementation would require separate training loops.
        # I'll implement a simple wrapper that uses standard loss for non-capsule models.
        print("Benchmarking requires separate training loops for CNNs/ViT. Exiting for now.")
        # You can extend this as needed.

    elif args.exp == "robustness":
        print("=== Direct background suppression evaluation ===")
        model = CapsuleNetCapFG(input_shape=input_shape, classes=n_classes).to(device)
        # Train model briefly (or load pretrained) – here we train for a few epochs for demonstration.
        # In practice, you should load a fully trained model.
        print("Training CapFG model for few epochs...")
        train_one_run(model, train_loader, val_loader, config, run_id=0, model_name="capfg")
        # Create synthetic dataset with ground truth masks
        synth_dataset = SyntheticForegroundDataset(num_samples=1000)
        synth_loader = DataLoader(synth_dataset, batch_size=32, shuffle=False)
        metrics = evaluate_background_suppression(model, synth_loader, device, config["lam_recon"], n_classes)
        print(f"Foreground IoU: {metrics['foreground_iou']:.4f}")
        print(f"Background leakage: {metrics['background_leakage']:.4f}")
        print(f"Clean accuracy on synthetic: {metrics['clean_accuracy']:.4f}")

    elif args.exp == "complexity":
        print("=== Complexity analysis ===")
        model = CapsuleNetCapFG(input_shape=input_shape, classes=n_classes).to(device)
        params, flops, infer_time = measure_complexity(model, input_shape, device)
        print(f"Parameters: {params/1e6:.2f}M")
        if flops:
            print(f"FLOPs: {flops:.2f} M")
        print(f"Inference time per sample: {infer_time:.2f} ms")

    elif args.exp == "full":
        # Run all major experiments sequentially
        print("Full experiment suite not fully implemented yet.")
        pass

if __name__ == "__main__":
    main()
```

```python
# cifar10_ablation_4models.py

import os
import csv
import random
from time import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam, lr_scheduler
from torch.utils.data import DataLoader, random_split, Subset
from torchvision import datasets, transforms


# =========================================================
# 1. Reproducibility and device
# =========================================================
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)


# =========================================================
# 2. Common utilities
# =========================================================
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = (norm ** 2) / (1.0 + norm ** 2)
    return scale * inputs / (norm + 1e-8)

def to_onehot(y, num_classes):
    return torch.eye(num_classes, device=y.device)[y]

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    margin = (
        y_true * torch.clamp(0.9 - y_pred, min=0.0) ** 2
        + 0.5 * (1.0 - y_true) * torch.clamp(y_pred - 0.1, min=0.0) ** 2
    )
    margin_loss = margin.sum(dim=1).mean()
    recon_loss = F.mse_loss(x_recon, x)
    return margin_loss + lam_recon * recon_loss


# =========================================================
# 3. CIFAR-10 loaders
# =========================================================
def load_cifar10(data_dir="./CIFAR10", batch_size=128, val_size=5000):
    train_transform = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ])

    eval_transform = transforms.Compose([
        transforms.ToTensor(),
    ])

    full_train_aug = datasets.CIFAR10(
        root=data_dir,
        train=True,
        download=True,
        transform=train_transform
    )

    full_train_eval = datasets.CIFAR10(
        root=data_dir,
        train=True,
        download=False,
        transform=eval_transform
    )

    test_set = datasets.CIFAR10(
        root=data_dir,
        train=False,
        download=True,
        transform=eval_transform
    )

    train_len = len(full_train_aug) - val_size
    val_len = val_size

    train_split, val_split_indices = random_split(
        range(len(full_train_aug)),
        [train_len, val_len],
        generator=torch.Generator().manual_seed(42)
    )

    train_indices = train_split.indices
    val_indices = val_split_indices.indices

    train_set = Subset(full_train_aug, train_indices)
    val_set = Subset(full_train_eval, val_indices)

    pin_memory = torch.cuda.is_available()

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=pin_memory)
    val_loader = DataLoader(val_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    test_loader = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)

    return train_loader, val_loader, test_loader


# =========================================================
# 4. Capsule blocks
# =========================================================
class PrimaryCapsule(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(
            in_channels, num_maps * num_dims,
            kernel_size=9, stride=2, padding=0
        )

    def forward(self, x):
        out = self.conv(x)  # expected shape: (B, 256, 8, 8)
        bsz = out.size(0)
        h, w = out.size(2), out.size(3)
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()  # (B, 32, H, W, 8)
        raw_norms = torch.norm(out, dim=-1)            # (B, 32, H, W)
        squashed = squash(out, axis=-1)                # (B, 32, H, W, 8)
        return squashed, raw_norms, h, w

class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super().__init__()
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out
        self.num_dims_in = num_dims_in
        self.num_dims_out = num_dims_out
        self.routings = routings

        self.weight = nn.Parameter(
            0.01 * torch.randn(num_caps_out, num_caps_in, num_dims_out, num_dims_in)
        )

    def forward(self, x):
        # x: (B, num_caps_in, num_dims_in)
        x = x[:, None, :, :, None]  # (B,1,Nin,Din,1)
        x_hat = torch.matmul(self.weight[None, :, :, :, :], x).squeeze(-1)  # (B,Nout,Nin,Dout)

        b = torch.zeros(x.size(0), self.num_caps_out, self.num_caps_in, device=x.device)

        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            s = (c[:, :, :, None] * x_hat).sum(dim=2)  # (B,Nout,Dout)
            v = squash(s, axis=-1)
            if i < self.routings - 1:
                b = b + (x_hat * v[:, :, None, :]).sum(dim=-1)

        return v

class CapFG(nn.Module):
    def __init__(self, in_maps=32):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_maps, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, raw_norms):
        return self.conv(raw_norms)  # (B,1,H,W)


# =========================================================
# 5. Unified model with switches
# =========================================================
class CapsuleNetAblation(nn.Module):
    def __init__(
        self,
        input_size=(3, 32, 32),
        classes=10,
        routings=3,
        use_input_mask=False,
        use_capfg=False,
        mask_threshold=0.1
    ):
        super().__init__()
        self.input_size = input_size
        self.classes = classes
        self.use_input_mask = use_input_mask
        self.use_capfg = use_capfg
        self.mask_threshold = mask_threshold

        self.conv1 = nn.Conv2d(input_size[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)
        self.primarycaps = PrimaryCapsule(in_channels=256, num_maps=32, num_dims=8)

        # CIFAR-10 geometry: 32 -> 24 after conv1 -> 8 after primarycaps
        self.num_caps_in = 32 * 8 * 8

        if self.use_capfg:
            self.capfg = CapFG(in_maps=32)

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, input_size[0] * input_size[1] * input_size[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None):
        # Optional binary input mask
        if self.use_input_mask:
            mask_img = (x > self.mask_threshold).float()
            x = x * mask_img

        out = self.relu(self.conv1(x))
        prim_spatial, raw_norms, h, w = self.primarycaps(out)   # prim_spatial: (B,32,H,W,8)

        # Optional CapFG gating
        if self.use_capfg:
            A = self.capfg(raw_norms)      # (B,1,H,W)
            A = A.unsqueeze(-1)            # (B,1,H,W,1)
            prim_spatial = prim_spatial * A

        prim_flat = prim_spatial.view(x.size(0), -1, 8)
        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.input_size)
        return length, recon


# =========================================================
# 6. Train / eval
# =========================================================
def evaluate(model, loader, lam_recon, num_classes=10):
    model.eval()
    total_loss = 0.0
    total_correct = 0
    total = 0

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, num_classes)

            y_pred, x_recon = model(x)
            loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)

            total_loss += loss.item() * x.size(0)
            preds = y_pred.argmax(dim=1)
            total_correct += (preds == y).sum().item()
            total += x.size(0)

    return total_loss / total, total_correct / total

def train_model(model, train_loader, val_loader, config, save_dir, model_name):
    os.makedirs(save_dir, exist_ok=True)

    optimizer = Adam(model.parameters(), lr=config["lr"])
    scheduler = lr_scheduler.ExponentialLR(optimizer, gamma=config["lr_decay"])

    best_val_acc = 0.0
    history = []

    log_path = os.path.join(save_dir, "log.csv")
    with open(log_path, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["epoch", "train_loss", "val_loss", "val_acc"])
        writer.writeheader()

        for epoch in range(config["epochs"]):
            model.train()
            running_loss = 0.0
            total = 0
            t0 = time()

            for x, y in train_loader:
                x, y = x.to(device), y.to(device)
                y_onehot = to_onehot(y, config["classes"])

                optimizer.zero_grad()
                y_pred, x_recon = model(x, y_onehot)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"])
                loss.backward()
                optimizer.step()

                running_loss += loss.item() * x.size(0)
                total += x.size(0)

            scheduler.step()

            train_loss = running_loss / total
            val_loss, val_acc = evaluate(model, val_loader, config["lam_recon"], config["classes"])

            row = {
                "epoch": epoch + 1,
                "train_loss": train_loss,
                "val_loss": val_loss,
                "val_acc": val_acc,
            }
            writer.writerow(row)
            history.append(row)

            print(
                f"{model_name} | Epoch {epoch+1:02d}/{config['epochs']} | "
                f"train_loss={train_loss:.4f} | val_loss={val_loss:.4f} | "
                f"val_acc={val_acc:.4f} | time={int(time()-t0)}s"
            )

            if val_acc > best_val_acc:
                best_val_acc = val_acc
                torch.save(model.state_dict(), os.path.join(save_dir, "best_model.pt"))

    print(f"{model_name} best val_acc = {best_val_acc:.4f}")
    return pd.DataFrame(history)


# =========================================================
# 7. Plotting
# =========================================================
def plot_ablation(log_paths, labels, save_path="cifar10_ablation_4models.png"):
    dfs = [pd.read_csv(p) for p in log_paths]

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    for df, label in zip(dfs, labels):
        axes[0].plot(df["epoch"], df["train_loss"], label=label)
        axes[1].plot(df["epoch"], df["val_loss"], label=label)
        axes[2].plot(df["epoch"], df["val_acc"], label=label)

    axes[0].set_title("Training Loss")
    axes[1].set_title("Validation Loss")
    axes[2].set_title("Validation Accuracy")

    for ax in axes:
        ax.set_xlabel("Epoch")
        ax.legend()
        ax.grid(True)

    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.show()
    print(f"Saved plot to {save_path}")


# =========================================================
# 8. Main
# =========================================================
if __name__ == "__main__":
    config = {
        "epochs": 50,
        "batch_size": 128,
        "lr": 0.001,
        "lr_decay": 0.9,
        "lam_recon": 0.0005 * (3 * 32 * 32),
        "routings": 3,
        "classes": 10,
        "data_dir": "./CIFAR10",
        "mask_threshold": 0.1,
    }

    train_loader, val_loader, test_loader = load_cifar10(
        data_dir=config["data_dir"],
        batch_size=config["batch_size"],
        val_size=5000
    )

    experiments = [
        {
            "name": "Baseline",
            "save_dir": "./result_ablation_baseline",
            "use_input_mask": False,
            "use_capfg": False,
        },
        {
            "name": "MaskOnly",
            "save_dir": "./result_ablation_mask_only",
            "use_input_mask": True,
            "use_capfg": False,
        },
        {
            "name": "CapFGOnly",
            "save_dir": "./result_ablation_capfg_only",
            "use_input_mask": False,
            "use_capfg": True,
        },
        {
            "name": "MaskPlusCapFG",
            "save_dir": "./result_ablation_mask_plus_capfg",
            "use_input_mask": True,
            "use_capfg": True,
        },
    ]

    summary_rows = []
    log_paths = []
    labels = []

    for exp in experiments:
        print("\n" + "=" * 70)
        print(f"Running experiment: {exp['name']}")
        print("=" * 70)

        model = CapsuleNetAblation(
            input_size=(3, 32, 32),
            classes=config["classes"],
            routings=config["routings"],
            use_input_mask=exp["use_input_mask"],
            use_capfg=exp["use_capfg"],
            mask_threshold=config["mask_threshold"],
        ).to(device)

        train_model(
            model=model,
            train_loader=train_loader,
            val_loader=val_loader,
            config=config,
            save_dir=exp["save_dir"],
            model_name=exp["name"]
        )

        model.load_state_dict(torch.load(os.path.join(exp["save_dir"], "best_model.pt"), map_location=device))
        test_loss, test_acc = evaluate(model, test_loader, config["lam_recon"], config["classes"])

        print(f"{exp['name']} TEST: loss={test_loss:.4f}, acc={test_acc:.4f}")

        summary_rows.append({
            "model": exp["name"],
            "use_input_mask": exp["use_input_mask"],
            "use_capfg": exp["use_capfg"],
            "mask_threshold": config["mask_threshold"] if exp["use_input_mask"] else None,
            "test_loss": test_loss,
            "test_acc": test_acc,
        })

        log_paths.append(os.path.join(exp["save_dir"], "log.csv"))
        labels.append(exp["name"])

    summary_df = pd.DataFrame(summary_rows)
    summary_df.to_csv("cifar10_ablation_summary.csv", index=False)
    print("\nSaved cifar10_ablation_summary.csv")
    print(summary_df)

    plot_ablation(
        log_paths=log_paths,
        labels=labels,
        save_path="cifar10_ablation_4models.png"
    )
```

```python
# ============================================================
# improved_capfg_complete_fixed_v2.py
# Full corrected version for CIFAR-10
# - Fixes DenseCapsule shape mismatch using einsum
# - Uses improved pre-squash CapFG
# - Fixes background suppression mask size mismatch by upsampling 8x8 -> 32x32
# - Uses paired t-test across runs
# ============================================================

import random
import time
import argparse
import numpy as np
import pandas as pd
from scipy import stats

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam, lr_scheduler
from torch.utils.data import DataLoader, random_split, Subset, Dataset
from torchvision import datasets, transforms


# ---------------------------
# Reproducibility
# ---------------------------
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")


# ---------------------------
# Helper functions
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = (norm ** 2) / (1.0 + norm ** 2)
    return scale * inputs / (norm + 1e-8)

def to_onehot(y, num_classes):
    return torch.eye(num_classes, device=y.device)[y]

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    margin = (
        y_true * torch.clamp(0.9 - y_pred, min=0.0) ** 2
        + 0.5 * (1.0 - y_true) * torch.clamp(y_pred - 0.1, min=0.0) ** 2
    )
    margin_loss = margin.sum(dim=1).mean()
    recon_loss = F.mse_loss(x_recon, x)
    return margin_loss + lam_recon * recon_loss


# ---------------------------
# Data loaders (CIFAR-10)
# ---------------------------
def load_cifar10(data_dir="./data", batch_size=128, val_size=5000):
    train_tf = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ])
    eval_tf = transforms.Compose([
        transforms.ToTensor()
    ])

    full_train_aug = datasets.CIFAR10(root=data_dir, train=True, download=True, transform=train_tf)
    full_train_eval = datasets.CIFAR10(root=data_dir, train=True, download=False, transform=eval_tf)
    test_set = datasets.CIFAR10(root=data_dir, train=False, download=True, transform=eval_tf)

    total_len = len(full_train_aug)
    train_len = total_len - val_size

    train_subset, val_subset = random_split(
        range(total_len),
        [train_len, val_size],
        generator=torch.Generator().manual_seed(42)
    )

    train_set = Subset(full_train_aug, train_subset.indices)
    val_set = Subset(full_train_eval, val_subset.indices)

    pin_memory = torch.cuda.is_available()

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=pin_memory)
    val_loader = DataLoader(val_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    test_loader = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)

    return train_loader, val_loader, test_loader


# ---------------------------
# Synthetic dataset for background suppression
# ---------------------------
class SyntheticForegroundDataset(Dataset):
    def __init__(self, num_samples=2000, size=32, num_classes=10):
        self.num_samples = num_samples
        self.size = size
        self.num_classes = num_classes
        self.images = []
        self.labels = []
        self.masks = []

        for _ in range(num_samples):
            img = torch.randn(3, size, size) * 0.5
            label = random.randint(0, num_classes - 1)

            fg_size = random.randint(8, 16)
            fx = random.randint(0, size - fg_size)
            fy = random.randint(0, size - fg_size)

            mask = torch.zeros(1, size, size)
            mask[:, fy:fy+fg_size, fx:fx+fg_size] = 1.0

            fg = torch.randn(3, fg_size, fg_size) * 0.2 + 0.8
            img[:, fy:fy+fg_size, fx:fx+fg_size] = fg

            self.images.append(img)
            self.labels.append(label)
            self.masks.append(mask)

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        return self.images[idx], self.labels[idx], self.masks[idx]


# ---------------------------
# Capsule network components
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    """Baseline version: returns squashed capsules."""
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)  # (B, 256, 8, 8)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (B, 32, H, W, 8)
        out = squash(out, axis=-1)
        return out.view(bsz, -1, self.num_dims), h, w


class PrimaryCapsuleUnsquashed(nn.Module):
    """Improved CapFG version: returns unsquashed capsules + raw norms."""
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)  # (B, 256, 8, 8)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()   # (B, 32, H, W, 8)
        raw_norms = torch.norm(out, dim=-1)             # (B, 32, H, W)
        return out, raw_norms, h, w


class DenseCapsule(nn.Module):
    """
    Corrected DenseCapsule using einsum.
    weight shape: (Nout, Nin, Dout, Din)
    x shape:      (B, Nin, Din)
    output:       (B, Nout, Dout)
    """
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super().__init__()
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out
        self.num_dims_in = num_dims_in
        self.num_dims_out = num_dims_out
        self.routings = routings

        self.weight = nn.Parameter(
            0.01 * torch.randn(num_caps_out, num_caps_in, num_dims_out, num_dims_in)
        )

    def forward(self, x):
        # x: (B, Nin, Din)
        # weight: (Nout, Nin, Dout, Din)
        # u_hat: (B, Nout, Nin, Dout)
        u_hat = torch.einsum('bid,oijd->boij', x, self.weight)

        b = torch.zeros(
            x.size(0), self.num_caps_out, self.num_caps_in,
            device=x.device
        )

        for i in range(self.routings):
            c = F.softmax(b, dim=1)                       # (B, Nout, Nin)
            s = (c.unsqueeze(-1) * u_hat).sum(dim=2)     # (B, Nout, Dout)
            v = squash(s, axis=-1)                       # (B, Nout, Dout)

            if i < self.routings - 1:
                b = b + (u_hat * v.unsqueeze(2)).sum(dim=-1)

        return v


class ImprovedCapFG(nn.Module):
    """
    Improved mask generator:
    - local conv stack
    - optional global modulation
    - learnable temperature
    - entropy regularization
    """
    def __init__(self, in_maps=32, beta=0.2, entropy_weight=0.01, use_global=True):
        super().__init__()
        self.beta = beta
        self.entropy_weight = entropy_weight
        self.use_global = use_global

        self.local = nn.Sequential(
            nn.Conv2d(in_maps, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),

            nn.Conv2d(32, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True),

            nn.Conv2d(16, 1, kernel_size=1)
        )

        if use_global:
            self.global_pool = nn.AdaptiveAvgPool2d(1)
            self.fc = nn.Linear(in_maps, 1)

        self.temperature = nn.Parameter(torch.tensor(1.0))

    def forward(self, raw_norms):
        logits = self.local(raw_norms)   # (B,1,H,W)

        if self.use_global:
            g = torch.sigmoid(self.fc(self.global_pool(raw_norms).flatten(1)))  # (B,1)
            logits = logits * g[:, :, None, None]

        mask = torch.sigmoid(logits / self.temperature)
        return mask

    def entropy_loss(self, mask):
        m = mask.view(mask.size(0), -1)
        entropy = -torch.mean(m * torch.log(m + 1e-8) + (1 - m) * torch.log(1 - m + 1e-8))
        return self.entropy_weight * entropy


# ---------------------------
# Full models
# ---------------------------
class CapsuleNetBaseline(nn.Module):
    def __init__(self, shape=(3, 32, 32), classes=10, routings=3):
        super().__init__()
        self.shape = shape
        self.classes = classes

        self.conv1 = nn.Conv2d(shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        self.primarycaps = PrimaryCapsuleBase(in_channels=256, num_maps=32, num_dims=8)
        self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, shape[0] * shape[1] * shape[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))
        prim_flat, h, w = self.primarycaps(out)
        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        return length, recon.view(-1, *self.shape)


class CapsuleNetImprovedCapFG(nn.Module):
    def __init__(self, shape=(3, 32, 32), classes=10, routings=3, beta=0.2, entropy_weight=0.01, use_global=True):
        super().__init__()
        self.shape = shape
        self.classes = classes
        self.beta = beta

        self.conv1 = nn.Conv2d(shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        self.primary = PrimaryCapsuleUnsquashed(in_channels=256, num_maps=32, num_dims=8)
        self.mask_gen = ImprovedCapFG(in_maps=32, beta=beta, entropy_weight=entropy_weight, use_global=use_global)

        self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, shape[0] * shape[1] * shape[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None, return_mask=False):
        out = self.relu(self.conv1(x))
        prim_unsq, raw_norms, h, w = self.primary(out)     # (B,32,H,W,8), (B,32,H,W)

        mask = self.mask_gen(raw_norms)                    # (B,1,H,W)
        mask_exp = mask.unsqueeze(-1)                      # (B,1,H,W,1)

        # Residual gating
        prim_masked = prim_unsq * mask_exp + prim_unsq * self.beta * (1 - mask_exp)

        prim_squashed = squash(prim_masked, axis=-1)
        prim_flat = prim_squashed.view(x.size(0), -1, 8)

        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.shape)

        if return_mask:
            return length, recon, mask

        return length, recon

    def mask_entropy_loss(self, mask):
        return self.mask_gen.entropy_loss(mask)


# ---------------------------
# Training and evaluation
# ---------------------------
def evaluate(model, loader, lam_recon, num_classes, use_mask_loss=False):
    model.eval()
    total_loss = 0.0
    total_correct = 0
    total = 0

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, num_classes)

            if use_mask_loss and hasattr(model, "mask_entropy_loss"):
                y_pred, x_recon, mask = model(x, return_mask=True)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon) + model.mask_entropy_loss(mask)
            else:
                y_pred, x_recon = model(x)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)

            total_loss += loss.item() * x.size(0)
            preds = y_pred.argmax(dim=1)
            total_correct += (preds == y).sum().item()
            total += x.size(0)

    return total_loss / total, total_correct / total


def train_one_run(model, train_loader, val_loader, config, use_mask_loss=False):
    optimizer = Adam(model.parameters(), lr=config["lr"])
    scheduler = lr_scheduler.ExponentialLR(optimizer, gamma=config["lr_decay"])

    best_val_acc = 0.0

    for epoch in range(config["epochs"]):
        model.train()

        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, config["classes"])

            optimizer.zero_grad()

            if use_mask_loss and hasattr(model, "mask_entropy_loss"):
                y_pred, x_recon, mask = model(x, y_onehot, return_mask=True)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"]) + model.mask_entropy_loss(mask)
            else:
                y_pred, x_recon = model(x, y_onehot)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"])

            loss.backward()
            optimizer.step()

        scheduler.step()
        _, val_acc = evaluate(model, val_loader, config["lam_recon"], config["classes"], use_mask_loss)

        if val_acc > best_val_acc:
            best_val_acc = val_acc

        print(f"  Epoch {epoch+1:02d}/{config['epochs']} | best_val_acc={best_val_acc:.4f}")

    return best_val_acc


def multi_run(model_class, model_kwargs, config, train_loader, val_loader, n_runs=3, use_mask_loss=False):
    accuracies = []

    for run in range(n_runs):
        print(f"\nRun {run+1}/{n_runs}")
        set_seed(42 + run)

        model = model_class(**model_kwargs).to(device)
        acc = train_one_run(model, train_loader, val_loader, config, use_mask_loss=use_mask_loss)
        accuracies.append(acc)

        print(f"  Final best val_acc = {acc:.4f}")

    mean = np.mean(accuracies)
    std = np.std(accuracies)
    return mean, std, accuracies


# ---------------------------
# Background suppression metrics
# ---------------------------
def background_suppression_metrics(model, synth_loader):
    model.eval()
    ious = []
    leakages = []

    with torch.no_grad():
        for x, y, true_mask in synth_loader:
            x = x.to(device)
            true_mask = true_mask.to(device)       # (B,1,32,32)

            _, _, pred_mask = model(x, return_mask=True)  # (B,1,8,8)

            # Upsample predicted mask to image size
            pred_mask_up = F.interpolate(
                pred_mask,
                size=true_mask.shape[-2:],   # (32,32)
                mode="bilinear",
                align_corners=False
            )

            pred_bin = (pred_mask_up > 0.5).float()

            inter = (pred_bin * true_mask).sum(dim=(1, 2, 3))
            union = ((pred_bin + true_mask) > 0.5).float().sum(dim=(1, 2, 3))
            iou = (inter + 1e-6) / (union + 1e-6)
            ious.extend(iou.cpu().numpy())

            bg_leak = (pred_bin * (1 - true_mask)).sum(dim=(1, 2, 3)) / (
                (1 - true_mask).sum(dim=(1, 2, 3)) + 1e-6
            )
            leakages.extend(bg_leak.cpu().numpy())

    return np.mean(ious), np.mean(leakages)


# ---------------------------
# Complexity
# ---------------------------
def measure_complexity(model, input_shape):
    model.eval()
    dummy = torch.randn(1, *input_shape).to(device)

    params = sum(p.numel() for p in model.parameters())

    start = time.time()
    with torch.no_grad():
        for _ in range(100):
            _ = model(dummy)
    infer_ms = (time.time() - start) / 100 * 1000.0

    return params, infer_ms


# ---------------------------
# Main
# ---------------------------
def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--n_runs", type=int, default=3)
    parser.add_argument("--epochs", type=int, default=30)
    parser.add_argument("--batch_size", type=int, default=128)

    args, unknown = parser.parse_known_args()
    if unknown:
        print(f"Ignoring unknown arguments: {unknown}")

    config = {
        "epochs": args.epochs,
        "batch_size": args.batch_size,
        "lr": 0.001,
        "lr_decay": 0.9,
        "lam_recon": 0.0005 * (3 * 32 * 32),
        "classes": 10,
    }

    print("\nLoading CIFAR-10...")
    train_loader, val_loader, test_loader = load_cifar10(batch_size=config["batch_size"])

    print("\n" + "=" * 70)
    print(f"Baseline vs Improved CapFG (n_runs={args.n_runs}, epochs={args.epochs})")
    print("=" * 70)

    print("\nTraining Baseline CapsNet...")
    base_mean, base_std, base_accs = multi_run(
        CapsuleNetBaseline,
        {"shape": (3, 32, 32), "classes": 10},
        config,
        train_loader,
        val_loader,
        n_runs=args.n_runs,
        use_mask_loss=False
    )

    print("\nTraining Improved CapFG...")
    capfg_mean, capfg_std, capfg_accs = multi_run(
        CapsuleNetImprovedCapFG,
        {"shape": (3, 32, 32), "classes": 10, "beta": 0.2, "entropy_weight": 0.01, "use_global": True},
        config,
        train_loader,
        val_loader,
        n_runs=args.n_runs,
        use_mask_loss=True
    )

    t_stat, p_val = stats.ttest_rel(base_accs, capfg_accs)

    print("\nResults:")
    print(f"Baseline CapsNet : {base_mean:.4f} ± {base_std:.4f}")
    print(f"Improved CapFG   : {capfg_mean:.4f} ± {capfg_std:.4f}")
    print(f"Paired t-test    : t={t_stat:.4f}, p={p_val:.4f}")

    print("\nTraining final Improved CapFG model for background analysis...")
    final_model = CapsuleNetImprovedCapFG(
        shape=(3, 32, 32),
        classes=10,
        beta=0.2,
        entropy_weight=0.01,
        use_global=True
    ).to(device)

    train_one_run(final_model, train_loader, val_loader, config, use_mask_loss=True)

    print("\nEvaluating background suppression on synthetic dataset...")
    synth_dataset = SyntheticForegroundDataset(num_samples=1000, size=32, num_classes=10)
    synth_loader = DataLoader(synth_dataset, batch_size=64, shuffle=False)

    iou, leakage = background_suppression_metrics(final_model, synth_loader)
    print(f"Foreground IoU     : {iou:.4f}")
    print(f"Background Leakage : {leakage:.4f}")

    print("\nComplexity analysis...")
    base_model_for_complexity = CapsuleNetBaseline(shape=(3, 32, 32), classes=10).to(device)
    capfg_model_for_complexity = CapsuleNetImprovedCapFG(shape=(3, 32, 32), classes=10, beta=0.2, entropy_weight=0.01, use_global=True).to(device)

    base_params, base_infer_ms = measure_complexity(base_model_for_complexity, (3, 32, 32))
    capfg_params, capfg_infer_ms = measure_complexity(capfg_model_for_complexity, (3, 32, 32))

    print(f"Baseline Params      : {base_params/1e6:.2f} M | {base_infer_ms:.2f} ms/sample")
    print(f"Improved CapFG Params: {capfg_params/1e6:.2f} M | {capfg_infer_ms:.2f} ms/sample")

    summary = pd.DataFrame({
        "Model": ["Baseline CapsNet", "Improved CapFG"],
        "Mean Val Acc": [base_mean, capfg_mean],
        "Std Val Acc": [base_std, capfg_std],
        "Foreground IoU": [None, iou],
        "Background Leakage": [None, leakage],
        "Parameters (M)": [base_params / 1e6, capfg_params / 1e6],
        "Inference (ms)": [base_infer_ms, capfg_infer_ms],
    })

    summary.to_csv("improved_capfg_results.csv", index=False)
    print("\nSaved results to improved_capfg_results.csv")
    print("\n✅ All experiments completed.")


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# improved_capfg_with_ablation.py
# - Full ablation: Baseline, MaskOnly, CapFGOnly, Full
# - Uses all improvements (einsum, pre-squash, residual, entropy, global)
# - Reports mean±std, paired t-tests, background IoU/leakage, complexity
# ============================================================

import random
import time
import argparse
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam, lr_scheduler
from torch.utils.data import DataLoader, random_split, Subset, Dataset
from torchvision import datasets, transforms

# ---------------------------
# Reproducibility
# ---------------------------
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ---------------------------
# Helper functions
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = (norm ** 2) / (1.0 + norm ** 2)
    return scale * inputs / (norm + 1e-8)

def to_onehot(y, num_classes):
    return torch.eye(num_classes, device=y.device)[y]

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    margin = (y_true * torch.clamp(0.9 - y_pred, min=0.0)**2 +
              0.5 * (1.0 - y_true) * torch.clamp(y_pred - 0.1, min=0.0)**2)
    margin_loss = margin.sum(dim=1).mean()
    recon_loss = F.mse_loss(x_recon, x)
    return margin_loss + lam_recon * recon_loss

# ---------------------------
# Data loaders (CIFAR-10)
# ---------------------------
def load_cifar10(data_dir="./data", batch_size=128, val_size=5000):
    train_tf = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ])
    eval_tf = transforms.Compose([transforms.ToTensor()])

    full_train_aug = datasets.CIFAR10(root=data_dir, train=True, download=True, transform=train_tf)
    full_train_eval = datasets.CIFAR10(root=data_dir, train=True, download=False, transform=eval_tf)
    test_set = datasets.CIFAR10(root=data_dir, train=False, download=True, transform=eval_tf)

    total_len = len(full_train_aug)
    train_len = total_len - val_size
    train_subset, val_subset = random_split(range(total_len), [train_len, val_size],
                                            generator=torch.Generator().manual_seed(42))
    train_set = Subset(full_train_aug, train_subset.indices)
    val_set = Subset(full_train_eval, val_subset.indices)

    pin_memory = torch.cuda.is_available()
    train_loader = DataLoader(train_set, batch_size, shuffle=True, num_workers=2, pin_memory=pin_memory)
    val_loader   = DataLoader(val_set, batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    test_loader  = DataLoader(test_set, batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    return train_loader, val_loader, test_loader

# ---------------------------
# Synthetic dataset for background suppression
# ---------------------------
class SyntheticForegroundDataset(Dataset):
    def __init__(self, num_samples=2000, size=32, num_classes=10):
        self.num_samples = num_samples
        self.size = size
        self.num_classes = num_classes
        self.images = []
        self.labels = []
        self.masks = []
        for _ in range(num_samples):
            img = torch.randn(3, size, size) * 0.5
            label = random.randint(0, num_classes-1)
            fg_size = random.randint(8, 16)
            fx = random.randint(0, size - fg_size)
            fy = random.randint(0, size - fg_size)
            mask = torch.zeros(1, size, size)
            mask[:, fy:fy+fg_size, fx:fx+fg_size] = 1.0
            fg = torch.randn(3, fg_size, fg_size) * 0.2 + 0.8
            img[:, fy:fy+fg_size, fx:fx+fg_size] = fg
            self.images.append(img)
            self.labels.append(label)
            self.masks.append(mask)
    def __len__(self):
        return self.num_samples
    def __getitem__(self, idx):
        return self.images[idx], self.labels[idx], self.masks[idx]

# ---------------------------
# Capsule network components
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    """Returns squashed capsules."""
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)
    def forward(self, x):
        out = self.conv(x)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()
        out = squash(out, axis=-1)
        return out.view(bsz, -1, self.num_dims), h, w

class PrimaryCapsuleUnsquashed(nn.Module):
    """Returns unsquashed capsules + raw norms."""
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)
    def forward(self, x):
        out = self.conv(x)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()
        raw_norms = torch.norm(out, dim=-1)
        return out, raw_norms, h, w

class DenseCapsule(nn.Module):
    """Einsum-based DenseCapsule."""
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super().__init__()
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out
        self.routings = routings
        self.weight = nn.Parameter(0.01 * torch.randn(num_caps_out, num_caps_in, num_dims_out, num_dims_in))
    def forward(self, x):
        u_hat = torch.einsum('bid,oijd->boij', x, self.weight)
        b = torch.zeros(x.size(0), self.num_caps_out, self.num_caps_in, device=x.device)
        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            s = (c.unsqueeze(-1) * u_hat).sum(dim=2)
            v = squash(s, axis=-1)
            if i < self.routings - 1:
                b = b + (u_hat * v.unsqueeze(2)).sum(dim=-1)
        return v

class ImprovedCapFG(nn.Module):
    def __init__(self, in_maps=32, beta=0.2, entropy_weight=0.01, use_global=True):
        super().__init__()
        self.beta = beta
        self.entropy_weight = entropy_weight
        self.use_global = use_global
        self.local = nn.Sequential(
            nn.Conv2d(in_maps, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(inplace=True),
            nn.Conv2d(32, 16, 3, padding=1), nn.BatchNorm2d(16), nn.ReLU(inplace=True),
            nn.Conv2d(16, 1, 1)
        )
        if use_global:
            self.global_pool = nn.AdaptiveAvgPool2d(1)
            self.fc = nn.Linear(in_maps, 1)
        self.temperature = nn.Parameter(torch.tensor(1.0))
    def forward(self, raw_norms):
        logits = self.local(raw_norms)
        if self.use_global:
            g = torch.sigmoid(self.fc(self.global_pool(raw_norms).flatten(1)))
            logits = logits * g[:, :, None, None]
        return torch.sigmoid(logits / self.temperature)
    def entropy_loss(self, mask):
        m = mask.view(mask.size(0), -1)
        entropy = -torch.mean(m * torch.log(m+1e-8) + (1-m)*torch.log(1-m+1e-8))
        return self.entropy_weight * entropy

# ---------------------------
# Unified ablative model
# ---------------------------
class CapsuleNetAblation(nn.Module):
    def __init__(self, shape=(3,32,32), classes=10, routings=3,
                 use_input_mask=False, use_capfg=False, mask_threshold=0.1,
                 beta=0.2, entropy_weight=0.01, use_global=True):
        super().__init__()
        self.shape = shape
        self.classes = classes
        self.use_input_mask = use_input_mask
        self.use_capfg = use_capfg
        self.mask_threshold = mask_threshold

        self.conv1 = nn.Conv2d(shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        if use_capfg:
            self.primary = PrimaryCapsuleUnsquashed(256, 32, 8)
            self.mask_gen = ImprovedCapFG(32, beta=beta, entropy_weight=entropy_weight, use_global=use_global)
            self.num_caps_in = 32 * 8 * 8
        else:
            self.primary = PrimaryCapsuleBase(256, 32, 8)   # baseline primary
            self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(self.num_caps_in, classes, 8, 16, routings)
        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512), nn.ReLU(inplace=True),
            nn.Linear(512, 1024), nn.ReLU(inplace=True),
            nn.Linear(1024, shape[0]*shape[1]*shape[2]), nn.Sigmoid()
        )

    def forward(self, x, y=None, return_mask=False):
        # Optional input binary mask
        if self.use_input_mask:
            mask_img = (x > self.mask_threshold).float()
            x = x * mask_img

        out = self.relu(self.conv1(x))

        if self.use_capfg:
            prim_unsq, raw_norms, h, w = self.primary(out)   # unsquashed
            mask = self.mask_gen(raw_norms)                  # (B,1,H,W)
            mask_exp = mask.unsqueeze(-1)                    # (B,1,H,W,1)
            prim_masked = prim_unsq * mask_exp + prim_unsq * self.mask_gen.beta * (1 - mask_exp)
            prim_squashed = squash(prim_masked, axis=-1)
            prim_flat = prim_squashed.view(x.size(0), -1, 8)
        else:
            prim_flat, h, w = self.primary(out)              # already squashed and flattened

        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.shape)

        if return_mask and self.use_capfg:
            return length, recon, mask
        return length, recon

    def mask_entropy_loss(self, mask):
        if self.use_capfg:
            return self.mask_gen.entropy_loss(mask)
        return torch.tensor(0.0, device=device)

# ---------------------------
# Training and evaluation (reused)
# ---------------------------
def evaluate(model, loader, lam_recon, num_classes, use_mask_loss=False):
    model.eval()
    total_loss, total_correct, total = 0.0, 0, 0
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, num_classes)
            if use_mask_loss and hasattr(model, 'mask_entropy_loss'):
                y_pred, x_recon, mask = model(x, return_mask=True)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon) + model.mask_entropy_loss(mask)
            else:
                y_pred, x_recon = model(x)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)
            total_loss += loss.item() * x.size(0)
            total_correct += (y_pred.argmax(1) == y).sum().item()
            total += x.size(0)
    return total_loss/total, total_correct/total

def train_one_run(model, train_loader, val_loader, config, use_mask_loss=False):
    optimizer = Adam(model.parameters(), lr=config['lr'])
    scheduler = lr_scheduler.ExponentialLR(optimizer, gamma=config['lr_decay'])
    best_val_acc = 0.0
    for epoch in range(config['epochs']):
        model.train()
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, config['classes'])
            optimizer.zero_grad()
            if use_mask_loss and hasattr(model, 'mask_entropy_loss'):
                y_pred, x_recon, mask = model(x, y_onehot, return_mask=True)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config['lam_recon']) + model.mask_entropy_loss(mask)
            else:
                y_pred, x_recon = model(x, y_onehot)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config['lam_recon'])
            loss.backward()
            optimizer.step()
        scheduler.step()
        _, val_acc = evaluate(model, val_loader, config['lam_recon'], config['classes'], use_mask_loss)
        if val_acc > best_val_acc:
            best_val_acc = val_acc
        print(f"  Epoch {epoch+1:02d}/{config['epochs']} | best_val_acc={best_val_acc:.4f}")
    return best_val_acc

def multi_run(model_kwargs_list, config, train_loader, val_loader, n_runs=3):
    """model_kwargs_list: list of dicts with keys 'name', 'model_kwargs', 'use_mask_loss'"""
    results = {}
    for exp in model_kwargs_list:
        name = exp['name']
        print(f"\n--- {name} ---")
        accs = []
        for run in range(n_runs):
            set_seed(42+run)
            model = CapsuleNetAblation(**exp['model_kwargs']).to(device)
            acc = train_one_run(model, train_loader, val_loader, config, use_mask_loss=exp['use_mask_loss'])
            accs.append(acc)
            print(f"  Run {run+1} final val_acc = {acc:.4f}")
        mean_acc = np.mean(accs)
        std_acc = np.std(accs)
        results[name] = (mean_acc, std_acc, accs)
        print(f"{name}: {mean_acc:.4f} ± {std_acc:.4f}")
    return results

# ---------------------------
# Background suppression (for full model)
# ---------------------------
def background_suppression_metrics(model, synth_loader):
    model.eval()
    ious, leakages = [], []
    with torch.no_grad():
        for x, y, true_mask in synth_loader:
            x = x.to(device)
            true_mask = true_mask.to(device)
            _, _, pred_mask = model(x, return_mask=True)
            pred_mask_up = F.interpolate(pred_mask, size=true_mask.shape[-2:], mode='bilinear', align_corners=False)
            pred_bin = (pred_mask_up > 0.5).float()
            inter = (pred_bin * true_mask).sum(dim=(1,2,3))
            union = ((pred_bin + true_mask) > 0.5).float().sum(dim=(1,2,3))
            iou = (inter+1e-6)/(union+1e-6)
            ious.extend(iou.cpu().numpy())
            bg_leak = (pred_bin * (1-true_mask)).sum(dim=(1,2,3)) / ((1-true_mask).sum(dim=(1,2,3))+1e-6)
            leakages.extend(bg_leak.cpu().numpy())
    return np.mean(ious), np.mean(leakages)

# ---------------------------
# Complexity measurement
# ---------------------------
def measure_complexity(model, input_shape):
    model.eval()
    dummy = torch.randn(1, *input_shape).to(device)
    params = sum(p.numel() for p in model.parameters())
    start = time.time()
    with torch.no_grad():
        for _ in range(100):
            _ = model(dummy)
    infer_ms = (time.time() - start) / 100 * 1000
    return params, infer_ms

# ---------------------------
# Plotting
# ---------------------------
def plot_ablation(results, save_path='ablation_comparison.png'):
    names = list(results.keys())
    means = [results[n][0] for n in names]
    stds  = [results[n][1] for n in names]
    plt.figure(figsize=(8,5))
    plt.bar(names, means, yerr=stds, capsize=5, color=['gray','blue','green','red'])
    plt.ylabel('Validation Accuracy')
    plt.title('Ablation Study: Baseline vs MaskOnly vs CapFGOnly vs Full')
    plt.ylim(0.65, 0.75)
    for i, (m, s) in enumerate(zip(means, stds)):
        plt.text(i, m+0.003, f'{m:.3f}', ha='center')
    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.show()
    print(f"Saved plot to {save_path}")

# ---------------------------
# Main ablation run
# ---------------------------
def main():
    parser = argparse.ArgumentParser()
    parser.add_argument('--n_runs', type=int, default=3)
    parser.add_argument('--epochs', type=int, default=30)
    parser.add_argument('--batch_size', type=int, default=128)
    args, unknown = parser.parse_known_args()
    if unknown:
        print(f"Ignoring unknown arguments: {unknown}")

    config = {
        'epochs': args.epochs,
        'batch_size': args.batch_size,
        'lr': 0.001,
        'lr_decay': 0.9,
        'lam_recon': 0.0005 * (3*32*32),
        'classes': 10,
    }

    print("\nLoading CIFAR-10...")
    train_loader, val_loader, test_loader = load_cifar10(batch_size=config['batch_size'])

    # Define four ablative configurations
    experiments = [
        {
            'name': 'Baseline',
            'model_kwargs': {'use_input_mask': False, 'use_capfg': False},
            'use_mask_loss': False
        },
        {
            'name': 'MaskOnly',
            'model_kwargs': {'use_input_mask': True, 'use_capfg': False, 'mask_threshold': 0.1},
            'use_mask_loss': False
        },
        {
            'name': 'CapFGOnly',
            'model_kwargs': {'use_input_mask': False, 'use_capfg': True,
                             'beta': 0.2, 'entropy_weight': 0.01, 'use_global': True},
            'use_mask_loss': True
        },
        {
            'name': 'Full (Mask+CapFG)',
            'model_kwargs': {'use_input_mask': True, 'use_capfg': True,
                             'mask_threshold': 0.1,
                             'beta': 0.2, 'entropy_weight': 0.01, 'use_global': True},
            'use_mask_loss': True
        }
    ]

    print("\n" + "="*70)
    print(f"Ablation Study (n_runs={args.n_runs}, epochs={args.epochs})")
    print("="*70)

    results = multi_run(experiments, config, train_loader, val_loader, n_runs=args.n_runs)

    # Print summary table
    print("\n" + "="*70)
    print("Summary of Ablation Results:")
    print(f"{'Model':<20} {'Mean Val Acc':<15} {'Std':<10}")
    for name, (mean, std, _) in results.items():
        print(f"{name:<20} {mean:.4f}        ±{std:.4f}")

    # Train final full model for background suppression and complexity
    print("\nTraining final Full model for background analysis...")
    final_model = CapsuleNetAblation(**experiments[3]['model_kwargs']).to(device)
    train_one_run(final_model, train_loader, val_loader, config, use_mask_loss=True)

    print("\nEvaluating background suppression on synthetic dataset...")
    synth_dataset = SyntheticForegroundDataset(num_samples=1000, size=32, num_classes=10)
    synth_loader = DataLoader(synth_dataset, batch_size=64, shuffle=False)
    iou, leakage = background_suppression_metrics(final_model, synth_loader)
    print(f"Foreground IoU     : {iou:.4f}")
    print(f"Background Leakage : {leakage:.4f}")

    print("\nComplexity analysis...")
    base_model = CapsuleNetAblation(**experiments[0]['model_kwargs']).to(device)
    full_model = CapsuleNetAblation(**experiments[3]['model_kwargs']).to(device)
    base_params, base_infer = measure_complexity(base_model, (3,32,32))
    full_params, full_infer = measure_complexity(full_model, (3,32,32))
    print(f"Baseline Params      : {base_params/1e6:.2f} M | {base_infer:.2f} ms/sample")
    print(f"Full Model Params    : {full_params/1e6:.2f} M | {full_infer:.2f} ms/sample")

    # Plot
    plot_ablation(results)

    # Save results CSV
    summary_rows = []
    for name, (mean, std, _) in results.items():
        summary_rows.append({
            'Model': name,
            'Mean Val Acc': mean,
            'Std Val Acc': std,
        })
    summary_rows.append({'Model': 'Full Model (extra)', 'Foreground IoU': iou, 'Background Leakage': leakage})
    summary_rows.append({'Model': 'Baseline Complexity', 'Params (M)': base_params/1e6, 'Inference (ms)': base_infer})
    summary_rows.append({'Model': 'Full Complexity', 'Params (M)': full_params/1e6, 'Inference (ms)': full_infer})
    df = pd.DataFrame(summary_rows)
    df.to_csv('ablation_results.csv', index=False)
    print("\nSaved results to ablation_results.csv")
    print("\n✅ Ablation study completed.")

if __name__ == '__main__':
    main()
```

```python
# ============================================================
# final_result1_capfg_extended.py
# Extended Result 1 pipeline for CIFAR-10
#
# Includes:
# 1. Baseline vs Improved CapFG comparison
# 2. 95% confidence intervals
# 3. Paired t-test across runs
# 4. Synthetic foreground IoU + background leakage
# 5. Complexity analysis
# 6. Threshold sensitivity
# 7. Background-shift robustness test
# 8. Thesis-ready export files
# ============================================================

import os
import random
import time
import argparse
import numpy as np
import pandas as pd
from scipy import stats

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam, lr_scheduler
from torch.utils.data import DataLoader, random_split, Subset, Dataset
from torchvision import datasets, transforms


# ---------------------------
# Reproducibility
# ---------------------------
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")


# ---------------------------
# Helper functions
# ---------------------------
def squash(inputs, axis=-1):
    norm = torch.norm(inputs, p=2, dim=axis, keepdim=True)
    scale = (norm ** 2) / (1.0 + norm ** 2)
    return scale * inputs / (norm + 1e-8)

def to_onehot(y, num_classes):
    return torch.eye(num_classes, device=y.device)[y]

def caps_loss(y_true, y_pred, x, x_recon, lam_recon):
    margin = (
        y_true * torch.clamp(0.9 - y_pred, min=0.0) ** 2
        + 0.5 * (1.0 - y_true) * torch.clamp(y_pred - 0.1, min=0.0) ** 2
    )
    margin_loss = margin.sum(dim=1).mean()
    recon_loss = F.mse_loss(x_recon, x)
    return margin_loss + lam_recon * recon_loss

def mean_ci(values, confidence=0.95):
    arr = np.array(values, dtype=float)
    n = len(arr)
    mean = np.mean(arr)
    std = np.std(arr, ddof=1) if n > 1 else 0.0
    if n > 1:
        h = stats.t.ppf((1 + confidence) / 2.0, n - 1) * (std / np.sqrt(n))
    else:
        h = 0.0
    return mean, std, mean - h, mean + h


# ---------------------------
# Data transforms
# ---------------------------
class ThresholdMaskTransform:
    def __init__(self, threshold):
        self.threshold = threshold

    def __call__(self, x):
        return x * (x > self.threshold).float()


class BackgroundNoiseTransform:
    def __init__(self, noise_std=0.25):
        self.noise_std = noise_std

    def __call__(self, x):
        noise = torch.randn_like(x) * self.noise_std
        return torch.clamp(x + noise, 0.0, 1.0)


# ---------------------------
# Data loaders
# ---------------------------
def load_cifar10(data_dir="./data", batch_size=128, val_size=5000):
    train_tf = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ])
    eval_tf = transforms.Compose([
        transforms.ToTensor()
    ])

    full_train_aug = datasets.CIFAR10(root=data_dir, train=True, download=True, transform=train_tf)
    full_train_eval = datasets.CIFAR10(root=data_dir, train=True, download=False, transform=eval_tf)
    test_set = datasets.CIFAR10(root=data_dir, train=False, download=True, transform=eval_tf)

    total_len = len(full_train_aug)
    train_len = total_len - val_size

    train_subset, val_subset = random_split(
        range(total_len),
        [train_len, val_size],
        generator=torch.Generator().manual_seed(42)
    )

    train_set = Subset(full_train_aug, train_subset.indices)
    val_set = Subset(full_train_eval, val_subset.indices)

    pin_memory = torch.cuda.is_available()

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=pin_memory)
    val_loader = DataLoader(val_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    test_loader = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)

    return train_loader, val_loader, test_loader


def load_cifar10_with_optional_mask(data_dir="./data", batch_size=128, val_size=5000, threshold=None):
    train_ops = [
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ]
    eval_ops = [
        transforms.ToTensor(),
    ]

    if threshold is not None:
        train_ops.append(ThresholdMaskTransform(threshold))
        eval_ops.append(ThresholdMaskTransform(threshold))

    train_tf = transforms.Compose(train_ops)
    eval_tf = transforms.Compose(eval_ops)

    full_train_aug = datasets.CIFAR10(root=data_dir, train=True, download=True, transform=train_tf)
    full_train_eval = datasets.CIFAR10(root=data_dir, train=True, download=False, transform=eval_tf)
    test_set = datasets.CIFAR10(root=data_dir, train=False, download=True, transform=eval_tf)

    total_len = len(full_train_aug)
    train_len = total_len - val_size

    train_subset, val_subset = random_split(
        range(total_len),
        [train_len, val_size],
        generator=torch.Generator().manual_seed(42)
    )

    train_set = Subset(full_train_aug, train_subset.indices)
    val_set = Subset(full_train_eval, val_subset.indices)

    pin_memory = torch.cuda.is_available()

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=pin_memory)
    val_loader = DataLoader(val_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    test_loader = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)

    return train_loader, val_loader, test_loader


def load_cifar10_corrupted_test(data_dir="./data", batch_size=128, noise_std=0.25):
    eval_tf = transforms.Compose([
        transforms.ToTensor(),
        BackgroundNoiseTransform(noise_std=noise_std)
    ])

    test_set = datasets.CIFAR10(root=data_dir, train=False, download=True, transform=eval_tf)

    pin_memory = torch.cuda.is_available()
    test_loader = DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=pin_memory)
    return test_loader


# ---------------------------
# Synthetic dataset
# ---------------------------
class SyntheticForegroundDataset(Dataset):
    def __init__(self, num_samples=2000, size=32, num_classes=10):
        self.num_samples = num_samples
        self.size = size
        self.num_classes = num_classes
        self.images = []
        self.labels = []
        self.masks = []

        for _ in range(num_samples):
            img = torch.randn(3, size, size) * 0.5
            label = random.randint(0, num_classes - 1)

            fg_size = random.randint(8, 16)
            fx = random.randint(0, size - fg_size)
            fy = random.randint(0, size - fg_size)

            mask = torch.zeros(1, size, size)
            mask[:, fy:fy+fg_size, fx:fx+fg_size] = 1.0

            fg = torch.randn(3, fg_size, fg_size) * 0.2 + 0.8
            img[:, fy:fy+fg_size, fx:fx+fg_size] = fg

            self.images.append(img)
            self.labels.append(label)
            self.masks.append(mask)

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        return self.images[idx], self.labels[idx], self.masks[idx]


# ---------------------------
# Capsule components
# ---------------------------
class PrimaryCapsuleBase(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()
        out = squash(out, axis=-1)
        return out.view(bsz, -1, self.num_dims), h, w


class PrimaryCapsuleUnsquashed(nn.Module):
    def __init__(self, in_channels=256, num_maps=32, num_dims=8):
        super().__init__()
        self.num_maps = num_maps
        self.num_dims = num_dims
        self.conv = nn.Conv2d(in_channels, num_maps * num_dims, kernel_size=9, stride=2, padding=0)

    def forward(self, x):
        out = self.conv(x)
        bsz, _, h, w = out.shape
        out = out.view(bsz, self.num_maps, self.num_dims, h, w)
        out = out.permute(0, 1, 3, 4, 2).contiguous()
        raw_norms = torch.norm(out, dim=-1)
        return out, raw_norms, h, w


class DenseCapsule(nn.Module):
    def __init__(self, num_caps_in, num_caps_out, num_dims_in, num_dims_out, routings=3):
        super().__init__()
        self.num_caps_in = num_caps_in
        self.num_caps_out = num_caps_out
        self.num_dims_in = num_dims_in
        self.num_dims_out = num_dims_out
        self.routings = routings

        self.weight = nn.Parameter(
            0.01 * torch.randn(num_caps_out, num_caps_in, num_dims_out, num_dims_in)
        )

    def forward(self, x):
        u_hat = torch.einsum('bid,oijd->boij', x, self.weight)

        b = torch.zeros(
            x.size(0), self.num_caps_out, self.num_caps_in,
            device=x.device
        )

        for i in range(self.routings):
            c = F.softmax(b, dim=1)
            s = (c.unsqueeze(-1) * u_hat).sum(dim=2)
            v = squash(s, axis=-1)

            if i < self.routings - 1:
                b = b + (u_hat * v.unsqueeze(2)).sum(dim=-1)

        return v


class ImprovedCapFG(nn.Module):
    def __init__(self, in_maps=32, beta=0.2, entropy_weight=0.01, use_global=True):
        super().__init__()
        self.beta = beta
        self.entropy_weight = entropy_weight
        self.use_global = use_global

        self.local = nn.Sequential(
            nn.Conv2d(in_maps, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),

            nn.Conv2d(32, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True),

            nn.Conv2d(16, 1, kernel_size=1)
        )

        if use_global:
            self.global_pool = nn.AdaptiveAvgPool2d(1)
            self.fc = nn.Linear(in_maps, 1)

        self.temperature = nn.Parameter(torch.tensor(1.0))

    def forward(self, raw_norms):
        logits = self.local(raw_norms)

        if self.use_global:
            g = torch.sigmoid(self.fc(self.global_pool(raw_norms).flatten(1)))
            logits = logits * g[:, :, None, None]

        mask = torch.sigmoid(logits / self.temperature)
        return mask

    def entropy_loss(self, mask):
        m = mask.view(mask.size(0), -1)
        entropy = -torch.mean(m * torch.log(m + 1e-8) + (1 - m) * torch.log(1 - m + 1e-8))
        return self.entropy_weight * entropy


# ---------------------------
# Full models
# ---------------------------
class CapsuleNetBaseline(nn.Module):
    def __init__(self, shape=(3, 32, 32), classes=10, routings=3):
        super().__init__()
        self.shape = shape
        self.classes = classes

        self.conv1 = nn.Conv2d(shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        self.primarycaps = PrimaryCapsuleBase(in_channels=256, num_maps=32, num_dims=8)
        self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, shape[0] * shape[1] * shape[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None):
        out = self.relu(self.conv1(x))
        prim_flat, h, w = self.primarycaps(out)
        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        return length, recon.view(-1, *self.shape)


class CapsuleNetImprovedCapFG(nn.Module):
    def __init__(self, shape=(3, 32, 32), classes=10, routings=3, beta=0.2, entropy_weight=0.01, use_global=True):
        super().__init__()
        self.shape = shape
        self.classes = classes
        self.beta = beta

        self.conv1 = nn.Conv2d(shape[0], 256, kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)

        self.primary = PrimaryCapsuleUnsquashed(in_channels=256, num_maps=32, num_dims=8)
        self.mask_gen = ImprovedCapFG(in_maps=32, beta=beta, entropy_weight=entropy_weight, use_global=use_global)

        self.num_caps_in = 32 * 8 * 8

        self.digitcaps = DenseCapsule(
            num_caps_in=self.num_caps_in,
            num_caps_out=classes,
            num_dims_in=8,
            num_dims_out=16,
            routings=routings
        )

        self.decoder = nn.Sequential(
            nn.Linear(16 * classes, 512),
            nn.ReLU(inplace=True),
            nn.Linear(512, 1024),
            nn.ReLU(inplace=True),
            nn.Linear(1024, shape[0] * shape[1] * shape[2]),
            nn.Sigmoid()
        )

    def forward(self, x, y=None, return_mask=False):
        out = self.relu(self.conv1(x))
        prim_unsq, raw_norms, h, w = self.primary(out)

        mask = self.mask_gen(raw_norms)
        mask_exp = mask.unsqueeze(-1)

        prim_masked = prim_unsq * mask_exp + prim_unsq * self.beta * (1 - mask_exp)

        prim_squashed = squash(prim_masked, axis=-1)
        prim_flat = prim_squashed.view(x.size(0), -1, 8)

        out = self.digitcaps(prim_flat)
        length = out.norm(dim=-1)

        if y is None:
            idx = length.max(dim=1)[1]
            y = torch.eye(self.classes, device=x.device)[idx]

        recon = self.decoder((out * y[:, :, None]).view(out.size(0), -1))
        recon = recon.view(-1, *self.shape)

        if return_mask:
            return length, recon, mask

        return length, recon

    def mask_entropy_loss(self, mask):
        return self.mask_gen.entropy_loss(mask)


# ---------------------------
# Train / eval
# ---------------------------
def evaluate(model, loader, lam_recon, num_classes, use_mask_loss=False):
    model.eval()
    total_loss = 0.0
    total_correct = 0
    total = 0

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, num_classes)

            if use_mask_loss and hasattr(model, "mask_entropy_loss"):
                y_pred, x_recon, mask = model(x, return_mask=True)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon) + model.mask_entropy_loss(mask)
            else:
                y_pred, x_recon = model(x)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, lam_recon)

            total_loss += loss.item() * x.size(0)
            preds = y_pred.argmax(dim=1)
            total_correct += (preds == y).sum().item()
            total += x.size(0)

    return total_loss / total, total_correct / total


def train_one_run(model, train_loader, val_loader, config, use_mask_loss=False):
    optimizer = Adam(model.parameters(), lr=config["lr"])
    scheduler = lr_scheduler.ExponentialLR(optimizer, gamma=config["lr_decay"])

    best_val_acc = 0.0

    for epoch in range(config["epochs"]):
        model.train()

        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            y_onehot = to_onehot(y, config["classes"])

            optimizer.zero_grad()

            if use_mask_loss and hasattr(model, "mask_entropy_loss"):
                y_pred, x_recon, mask = model(x, y_onehot, return_mask=True)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"]) + model.mask_entropy_loss(mask)
            else:
                y_pred, x_recon = model(x, y_onehot)
                loss = caps_loss(y_onehot, y_pred, x, x_recon, config["lam_recon"])

            loss.backward()
            optimizer.step()

        scheduler.step()
        _, val_acc = evaluate(model, val_loader, config["lam_recon"], config["classes"], use_mask_loss)

        if val_acc > best_val_acc:
            best_val_acc = val_acc

        print(f"  Epoch {epoch+1:02d}/{config['epochs']} | best_val_acc={best_val_acc:.4f}")

    return best_val_acc


def multi_run(model_class, model_kwargs, config, train_loader, val_loader, n_runs=3, use_mask_loss=False):
    accuracies = []

    for run in range(n_runs):
        print(f"\nRun {run+1}/{n_runs}")
        set_seed(42 + run)

        model = model_class(**model_kwargs).to(device)
        acc = train_one_run(model, train_loader, val_loader, config, use_mask_loss=use_mask_loss)
        accuracies.append(acc)

        print(f"  Final best val_acc = {acc:.4f}")

    mean = np.mean(accuracies)
    std = np.std(accuracies)
    return mean, std, accuracies


# ---------------------------
# Extra evaluation functions
# ---------------------------
def background_suppression_metrics(model, synth_loader):
    model.eval()
    ious = []
    leakages = []

    with torch.no_grad():
        for x, y, true_mask in synth_loader:
            x = x.to(device)
            true_mask = true_mask.to(device)

            _, _, pred_mask = model(x, return_mask=True)

            pred_mask_up = F.interpolate(
                pred_mask,
                size=true_mask.shape[-2:],
                mode="bilinear",
                align_corners=False
            )

            pred_bin = (pred_mask_up > 0.5).float()

            inter = (pred_bin * true_mask).sum(dim=(1, 2, 3))
            union = ((pred_bin + true_mask) > 0.5).float().sum(dim=(1, 2, 3))
            iou = (inter + 1e-6) / (union + 1e-6)
            ious.extend(iou.cpu().numpy())

            bg_leak = (pred_bin * (1 - true_mask)).sum(dim=(1, 2, 3)) / (
                (1 - true_mask).sum(dim=(1, 2, 3)) + 1e-6
            )
            leakages.extend(bg_leak.cpu().numpy())

    return np.mean(ious), np.mean(leakages)


def measure_complexity(model, input_shape):
    model.eval()
    dummy = torch.randn(1, *input_shape).to(device)

    params = sum(p.numel() for p in model.parameters())

    start = time.time()
    with torch.no_grad():
        for _ in range(100):
            _ = model(dummy)
    infer_ms = (time.time() - start) / 100 * 1000.0

    return params, infer_ms


def evaluate_background_shift(model, batch_size=128, noise_levels=(0.10, 0.20, 0.30)):
    rows = []

    for nl in noise_levels:
        test_loader = load_cifar10_corrupted_test(batch_size=batch_size, noise_std=nl)
        test_loss, test_acc = evaluate(model, test_loader, 0.0005 * (3 * 32 * 32), 10, use_mask_loss=True)
        rows.append({
            "noise_std": nl,
            "test_loss": test_loss,
            "test_acc": test_acc
        })

    return pd.DataFrame(rows)


def run_threshold_sensitivity(config, thresholds=(0.05, 0.10, 0.15, 0.20), n_runs=3):
    rows = []

    for th in thresholds:
        print("\n" + "=" * 70)
        print(f"Threshold sensitivity @ threshold={th}")
        print("=" * 70)

        train_loader, val_loader, test_loader = load_cifar10_with_optional_mask(
            batch_size=config["batch_size"],
            threshold=th
        )

        mask_accs = []
        for run in range(n_runs):
            set_seed(100 + run)
            model = CapsuleNetBaseline(shape=(3, 32, 32), classes=10).to(device)
            acc = train_one_run(model, train_loader, val_loader, config, use_mask_loss=False)
            mask_accs.append(acc)

        full_accs = []
        for run in range(n_runs):
            set_seed(200 + run)
            model = CapsuleNetImprovedCapFG(
                shape=(3, 32, 32),
                classes=10,
                beta=0.2,
                entropy_weight=0.01,
                use_global=True
            ).to(device)
            acc = train_one_run(model, train_loader, val_loader, config, use_mask_loss=True)
            full_accs.append(acc)

        rows.append({
            "threshold": th,
            "mask_only_mean": np.mean(mask_accs),
            "mask_only_std": np.std(mask_accs),
            "full_mean": np.mean(full_accs),
            "full_std": np.std(full_accs),
        })

    df = pd.DataFrame(rows)
    df.to_csv("threshold_sensitivity_results.csv", index=False)
    print("\nSaved threshold_sensitivity_results.csv")
    print(df)
    return df


# ---------------------------
# Main
# ---------------------------
def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--n_runs", type=int, default=3)
    parser.add_argument("--epochs", type=int, default=30)
    parser.add_argument("--batch_size", type=int, default=128)
    args, unknown = parser.parse_known_args()

    if unknown:
        print(f"Ignoring unknown arguments: {unknown}")

    config = {
        "epochs": args.epochs,
        "batch_size": args.batch_size,
        "lr": 0.001,
        "lr_decay": 0.9,
        "lam_recon": 0.0005 * (3 * 32 * 32),
        "classes": 10,
    }

    print("\nLoading CIFAR-10...")
    train_loader, val_loader, test_loader = load_cifar10(batch_size=config["batch_size"])

    print("\n" + "=" * 70)
    print(f"Baseline vs Improved CapFG (n_runs={args.n_runs}, epochs={args.epochs})")
    print("=" * 70)

    print("\nTraining Baseline CapsNet...")
    base_mean_raw, base_std_raw, base_accs = multi_run(
        CapsuleNetBaseline,
        {"shape": (3, 32, 32), "classes": 10},
        config,
        train_loader,
        val_loader,
        n_runs=args.n_runs,
        use_mask_loss=False
    )

    print("\nTraining Improved CapFG...")
    capfg_mean_raw, capfg_std_raw, capfg_accs = multi_run(
        CapsuleNetImprovedCapFG,
        {"shape": (3, 32, 32), "classes": 10, "beta": 0.2, "entropy_weight": 0.01, "use_global": True},
        config,
        train_loader,
        val_loader,
        n_runs=args.n_runs,
        use_mask_loss=True
    )

    t_stat, p_val = stats.ttest_rel(base_accs, capfg_accs)

    base_mean, base_std, base_lo, base_hi = mean_ci(base_accs)
    capfg_mean, capfg_std, capfg_lo, capfg_hi = mean_ci(capfg_accs)

    print("\nResults:")
    print(f"Baseline CapsNet : {base_mean:.4f} ± {base_std:.4f} | 95% CI [{base_lo:.4f}, {base_hi:.4f}]")
    print(f"Improved CapFG   : {capfg_mean:.4f} ± {capfg_std:.4f} | 95% CI [{capfg_lo:.4f}, {capfg_hi:.4f}]")
    print(f"Paired t-test    : t={t_stat:.4f}, p={p_val:.4f}")

    print("\nTraining final Improved CapFG model for background analysis...")
    final_model = CapsuleNetImprovedCapFG(
        shape=(3, 32, 32),
        classes=10,
        beta=0.2,
        entropy_weight=0.01,
        use_global=True
    ).to(device)

    train_one_run(final_model, train_loader, val_loader, config, use_mask_loss=True)

    print("\nEvaluating background suppression on synthetic dataset...")
    synth_dataset = SyntheticForegroundDataset(num_samples=1000, size=32, num_classes=10)
    synth_loader = DataLoader(synth_dataset, batch_size=64, shuffle=False)

    iou, leakage = background_suppression_metrics(final_model, synth_loader)
    print(f"Foreground IoU     : {iou:.4f}")
    print(f"Background Leakage : {leakage:.4f}")

    print("\nEvaluating background-shift robustness...")
    bg_df = evaluate_background_shift(final_model, batch_size=config["batch_size"], noise_levels=(0.10, 0.20, 0.30))
    bg_df.to_csv("background_shift_results.csv", index=False)
    print(bg_df)

    print("\nComplexity analysis...")
    base_model_for_complexity = CapsuleNetBaseline(shape=(3, 32, 32), classes=10).to(device)
    capfg_model_for_complexity = CapsuleNetImprovedCapFG(shape=(3, 32, 32), classes=10, beta=0.2, entropy_weight=0.01, use_global=True).to(device)

    base_params, base_infer_ms = measure_complexity(base_model_for_complexity, (3, 32, 32))
    capfg_params, capfg_infer_ms = measure_complexity(capfg_model_for_complexity, (3, 32, 32))

    print(f"Baseline Params      : {base_params/1e6:.2f} M | {base_infer_ms:.2f} ms/sample")
    print(f"Improved CapFG Params: {capfg_params/1e6:.2f} M | {capfg_infer_ms:.2f} ms/sample")

    summary = pd.DataFrame({
        "Model": ["Baseline CapsNet", "Improved CapFG"],
        "Mean Val Acc": [base_mean, capfg_mean],
        "Std Val Acc": [base_std, capfg_std],
        "CI Lower": [base_lo, capfg_lo],
        "CI Upper": [base_hi, capfg_hi],
        "Foreground IoU": [None, iou],
        "Background Leakage": [None, leakage],
        "Parameters (M)": [base_params / 1e6, capfg_params / 1e6],
        "Inference (ms)": [base_infer_ms, capfg_infer_ms],
    })

    summary.to_csv("improved_capfg_results.csv", index=False)
    print("\nSaved improved_capfg_results.csv")

    with open("thesis_result_1_summary.txt", "w") as f:
        f.write("Result 1 Summary\n")
        f.write("================\n")
        f.write(f"Baseline mean val acc: {base_mean:.4f} ± {base_std:.4f}\n")
        f.write(f"Improved CapFG mean val acc: {capfg_mean:.4f} ± {capfg_std:.4f}\n")
        f.write(f"95% CI baseline: [{base_lo:.4f}, {base_hi:.4f}]\n")
        f.write(f"95% CI improved CapFG: [{capfg_lo:.4f}, {capfg_hi:.4f}]\n")
        f.write(f"Paired t-test: t={t_stat:.4f}, p={p_val:.4f}\n")
        f.write(f"Foreground IoU: {iou:.4f}\n")
        f.write(f"Background Leakage: {leakage:.4f}\n")
        f.write(f"Baseline params (M): {base_params / 1e6:.4f}\n")
        f.write(f"Improved CapFG params (M): {capfg_params / 1e6:.4f}\n")
        f.write(f"Baseline inference (ms): {base_infer_ms:.4f}\n")
        f.write(f"Improved CapFG inference (ms): {capfg_infer_ms:.4f}\n")

    print("Saved thesis_result_1_summary.txt")

    print("\nRunning threshold sensitivity...")
    run_threshold_sensitivity(config, thresholds=(0.05, 0.10, 0.15, 0.20), n_runs=args.n_runs)

    print("\n✅ All experiments completed.")


if __name__ == "__main__":
    main()
```
