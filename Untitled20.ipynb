```python
"""
Result 2, Phase 1 (revised, with parameter-count sweep).

STATUS
------
Revision of the Phase 1 baseline script. Adds:
  (a) a parameter-count sweep so you can see which (conv1_channels, K, D)
      combination matches the paper's reported 105.5K, and
  (b) architecture presets so you can switch between configurations by
      changing one line.

Everything else is carried over from the previous revision, which fixed:
  - primary capsule stride (2 -> 1, paper-faithful)
  - capacity confound (pruning and reduced-capacity in the same run)
  - pruning collapse safety floor (min_keep_frac)
  - mid-train theta calibration
"""

import math
import time
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, random_split, Subset
from torchvision import datasets, transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")


# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------
USE_FG_ROUTING = False        # Phase 1: OFF. Set True for Phase 2 (FG-OSR).

# Number of channels output by conv1.
CAPACITY_CHANNELS = {
    "full":     256,   # current diagnostic -- ~11.6M params total
    "reduced":  32,    # the earlier 1.42M run (56.92% test)
    "tiny":     16,    # paper-like regime
    "minimal":  8,     # paper-like regime, even smaller
}

# ------------------------------------------------------------
# Architecture presets. Change ACTIVE_PRESET to switch configs.
# ------------------------------------------------------------
PRESETS = {
    "current": {
        "K": 32, "D": 16,
        "capacity_mode": "full",
        "primary_stride": 1,
        "note": "Current diagnostic (~11.6M params). Reference point.",
    },
    "paper_k8_d8": {
        "K": 8, "D": 8,
        "capacity_mode": "tiny",
        "primary_stride": 1,
        "note": "Paper-like. Expected ~90K params.",
    },
    "paper_k8_d16": {
        "K": 8, "D": 16,
        "capacity_mode": "tiny",
        "primary_stride": 1,
        "note": "Paper-like. Larger capsule dim. Expected ~180K params.",
    },
    "paper_k16_d8": {
        "K": 16, "D": 8,
        "capacity_mode": "minimal",
        "primary_stride": 1,
        "note": "Paper-like. More types, tiny conv1. Expected ~90K params.",
    },
    "paper_k16_d16": {
        "K": 16, "D": 16,
        "capacity_mode": "minimal",
        "primary_stride": 1,
        "note": "Paper-like. Expected ~180K params.",
    },
}

# ------------------------------------------------------------
# Which preset to run. Change this one line to switch.
# ------------------------------------------------------------
ACTIVE_PRESET = "current"

_PRESET = PRESETS[ACTIVE_PRESET]

CONFIG = {
    "data_dir": "./data",
    "batch_size": 256,
    "K": _PRESET["K"],
    "D": _PRESET["D"],
    "capacity_mode": _PRESET["capacity_mode"],
    "primary_stride": _PRESET["primary_stride"],

    # training
    "epochs": 60,
    "seeds": [42, 1, 7],
    "lr": 1e-3,
    "lr_floor": 1e-5,
    "warmup_epochs": 5,
    "weight_decay": 5e-4,
    "grad_clip": 1.0,

    # routing
    "alpha": 1.5,
    "entmax_n_iter": 30,

    # pruning
    "pruning_enabled": True,
    "prune_theta_start": 1.0,
    "prune_theta_final": 0.7,
    "prune_anneal_start": 30,
    "prune_min_keep_frac": 0.5,
    "similarity_mode": "flatten",
    "run_calibration": True,

    # FG-OSR (Phase 2)
    "relevance_mode": "learned",
    "relevance_hidden": 32,
    "fg_prune_thresh": 0.35,
    "fg_prune_warmup": 8,
    "fg_temp_beta": 0.6,
    "fg_temp_warmup": 8,
    "lam_relevance_sparse": 0.0,
}


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


# ------------------------------------------------------------
# Core ops
# ------------------------------------------------------------
def squash(x, dim=-1, axis=None, eps=1e-8):
    if axis is not None:
        dim = axis
    norm_sq = (x ** 2).sum(dim=dim, keepdim=True)
    norm = torch.sqrt(norm_sq + eps)
    scale = norm_sq / (1.0 + norm_sq)
    return scale * x / norm


class HouseholderOrthogonal(nn.Module):
    def __init__(self, dim, num_reflections=None):
        super().__init__()
        self.dim = dim
        self.num_reflections = num_reflections or dim
        self.v = nn.Parameter(torch.randn(self.num_reflections, dim) * 0.5)

    def forward(self):
        device = self.v.device
        dtype = self.v.dtype
        W = torch.eye(self.dim, device=device, dtype=dtype)
        for k in range(self.num_reflections):
            v = self.v[k]
            v_norm_sq = (v @ v).clamp(min=1e-8)
            W = W - 2.0 * torch.outer(v, v @ W) / v_norm_sq
        return W


def entmax_bisect(scores, alpha=1.5, dim=-1, n_iter=30, mask=None):
    if alpha == 1.0:
        if mask is not None:
            scores = scores.masked_fill(mask == 0, float("-inf"))
        return F.softmax(scores, dim=dim)

    if mask is not None:
        scores = scores.masked_fill(mask == 0, float("-inf"))

    a1 = alpha - 1.0
    x = scores * a1
    x = torch.where(torch.isneginf(scores), torch.full_like(x, float("-inf")), x)

    max_val = x.amax(dim=dim, keepdim=True)
    max_val = torch.nan_to_num(max_val, neginf=0.0)

    tau_lo = max_val - 1.0
    tau_hi = max_val.clone()

    with torch.no_grad():
        for _ in range(n_iter):
            tau = (tau_lo + tau_hi) / 2.0
            p = torch.clamp(x - tau, min=0.0) ** (1.0 / a1)
            p = torch.nan_to_num(p, nan=0.0)
            f = p.sum(dim=dim, keepdim=True)
            too_big = f > 1.0
            tau_lo = torch.where(too_big, tau, tau_lo)
            tau_hi = torch.where(too_big, tau_hi, tau)
        tau_final = ((tau_lo + tau_hi) / 2.0).detach()

    p = torch.clamp(x - tau_final, min=0.0) ** (1.0 / a1)
    p = torch.nan_to_num(p, nan=0.0)
    denom = p.sum(dim=dim, keepdim=True).clamp(min=1e-8)
    return p / denom


# ------------------------------------------------------------
# Primary capsules
# ------------------------------------------------------------
class PrimaryCapsules(nn.Module):
    def __init__(self, in_c, num_caps, dim_caps, kernel=9, stride=1):
        super().__init__()
        self.num_caps = num_caps
        self.dim_caps = dim_caps
        self.conv = nn.Conv2d(in_c, num_caps * dim_caps,
                              kernel_size=kernel, stride=stride)

    def forward(self, x):
        out = self.conv(x)
        b, _, h, w = out.shape
        out = out.view(b, self.num_caps, self.dim_caps, h, w)
        return out, h, w


# ------------------------------------------------------------
# Relevance head (placeholder for CapFG, Phase 2 only)
# ------------------------------------------------------------
class RelevanceHead(nn.Module):
    def __init__(self, in_ch=1, hidden=32):
        super().__init__()
        self.conv1 = nn.Conv2d(in_ch, hidden, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(hidden)
        self.conv2 = nn.Conv2d(hidden, hidden, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(hidden)
        self.conv3 = nn.Conv2d(hidden, 1, 1)

    def forward(self, x):
        h = F.relu(self.bn1(self.conv1(x)))
        h = F.relu(self.bn2(self.conv2(h)))
        return torch.sigmoid(self.conv3(h)).squeeze(1)


# ------------------------------------------------------------
# Pruned capsule layer
# ------------------------------------------------------------
class PrunedCapsuleLayer(nn.Module):
    def __init__(self, theta=1.0, fg_mode=False, fg_thresh=0.35,
                 min_keep_frac=0.5, similarity_mode="flatten"):
        super().__init__()
        self.theta = theta
        self.fg_mode = fg_mode
        self.fg_thresh_target = fg_thresh
        self.fg_thresh_current = 0.0
        self.min_keep_frac = min_keep_frac
        assert similarity_mode in ("flatten", "pooled")
        self.similarity_mode = similarity_mode

    def set_theta(self, t):
        self.theta = t

    def set_fg_thresh(self, t):
        self.fg_thresh_current = t

    def _descriptor(self, u):
        B, K, D, H, W = u.shape
        if self.similarity_mode == "pooled":
            return u.mean(dim=(3, 4))
        return u.reshape(B, K, D * H * W)

    def _min_keep(self, K):
        return max(1, int(math.ceil(self.min_keep_frac * K)))

    def forward(self, u, relevance=None):
        B, K, D, H, W = u.shape
        desc = self._descriptor(u)
        norms = torch.norm(desc, p=2, dim=-1)
        order = torch.argsort(norms, dim=1, descending=False)
        desc_ordered = torch.gather(
            desc, 1, order.unsqueeze(-1).expand(-1, -1, desc.shape[-1]))
        desc_unit = F.normalize(desc_ordered, p=2, dim=-1, eps=1e-8)
        T = torch.bmm(desc_unit, desc_unit.transpose(1, 2))

        upper_tri = torch.triu(
            torch.ones(K, K, device=u.device, dtype=torch.bool), diagonal=1)
        T_masked = T.masked_fill(~upper_tri.unsqueeze(0), -1.0)
        redundant = (T_masked > self.theta).any(dim=2)
        keep_ordered = (~redundant).float()

        min_keep = self._min_keep(K)
        n_kept = keep_ordered.sum(dim=1)
        needs_topup = n_kept < min_keep
        if needs_topup.any():
            topup_mask = torch.zeros_like(keep_ordered)
            topup_mask[:, -min_keep:] = 1.0
            keep_ordered = torch.where(
                needs_topup.unsqueeze(1),
                torch.maximum(keep_ordered, topup_mask),
                keep_ordered,
            )

        sim_keep = torch.zeros(B, K, device=u.device, dtype=keep_ordered.dtype)
        sim_keep.scatter_(1, order, keep_ordered)

        if not (self.fg_mode and relevance is not None):
            return sim_keep, None

        u_norm_spatial = u.norm(dim=2)
        weighted = u_norm_spatial * relevance.unsqueeze(1)
        num = weighted.sum(dim=(2, 3))
        den = u_norm_spatial.sum(dim=(2, 3)).clamp(min=1e-8)
        capsule_relevance = num / den

        fg_keep = (capsule_relevance > self.fg_thresh_current).float()
        keep_mask = sim_keep * fg_keep

        n_kept2 = keep_mask.sum(dim=1)
        needs_topup2 = n_kept2 < min_keep
        if needs_topup2.any():
            topup_mask2 = torch.zeros_like(keep_mask)
            topup_mask2.scatter_(1, order[:, -min_keep:], 1.0)
            keep_mask = torch.where(
                needs_topup2.unsqueeze(1),
                torch.maximum(keep_mask, topup_mask2),
                keep_mask,
            )
        return keep_mask, capsule_relevance


# ------------------------------------------------------------
# Simplified attention routing
# ------------------------------------------------------------
class SimplifiedAttentionRouting(nn.Module):
    def __init__(self, dim, alpha=1.5, n_iter=30,
                 fg_mode=False, fg_beta=0.6):
        super().__init__()
        self.dim = dim
        self.alpha = alpha
        self.n_iter = n_iter
        self.orth_W = HouseholderOrthogonal(dim)
        self.norm = nn.LayerNorm(dim)
        self.fg_mode = fg_mode
        self.fg_beta_target = fg_beta
        self.fg_beta_current = 0.0

    def set_beta(self, b):
        self.fg_beta_current = b

    def forward(self, u, keep_mask, relevance=None):
        B, K, D, H, W = u.shape
        W_orth = self.orth_W()

        u_perm = u.permute(0, 3, 4, 1, 2).reshape(B * H * W, K, D)
        u_hat = u_perm @ W_orth.T

        logits = torch.bmm(u_hat, u_perm.transpose(1, 2)) / math.sqrt(D)

        if self.fg_mode and relevance is not None and self.fg_beta_current > 0:
            rel_flat = relevance.reshape(B * H * W, 1, 1)
            scale = 1.0 + self.fg_beta_current * (2.0 * rel_flat - 1.0)
            scale = scale.clamp(min=0.1, max=10.0)
            logits = logits * scale

        keep_expanded = keep_mask.unsqueeze(1).unsqueeze(1) \
                                 .expand(B, H, W, K) \
                                 .reshape(B * H * W, K)
        pair_mask = keep_expanded.unsqueeze(2) * keep_expanded.unsqueeze(1)

        C = entmax_bisect(logits, alpha=self.alpha, dim=-1,
                          n_iter=self.n_iter, mask=pair_mask)
        s = torch.bmm(C, u_hat)
        v = squash(s, axis=-1)
        v = self.norm(v)
        v = v * keep_expanded.unsqueeze(-1)

        v = v.reshape(B, H, W, K, D).permute(0, 3, 4, 1, 2).contiguous()
        return v


# ------------------------------------------------------------
# OrthCaps-S (baseline + optional FG-OSR)
# ------------------------------------------------------------
class OrthCapsS(nn.Module):
    def __init__(self, num_classes=10, K=32, D=16,
                 conv1_channels=256, primary_stride=1,
                 prune_theta=1.0, alpha=1.5, n_iter=30,
                 prune_min_keep_frac=0.5, similarity_mode="flatten",
                 use_fg_routing=False, relevance_mode="learned",
                 relevance_hidden=32, fg_prune_thresh=0.35,
                 fg_temp_beta=0.6):
        super().__init__()
        self.num_classes = num_classes
        self.D = D
        self.K = K
        self.use_fg_routing = use_fg_routing
        self.relevance_mode = relevance_mode

        self.conv1 = nn.Conv2d(3, conv1_channels,
                               kernel_size=9, stride=1, padding=0)
        self.relu = nn.ReLU(inplace=True)
        self.primary = PrimaryCapsules(conv1_channels, K, D,
                                       kernel=9, stride=primary_stride)

        self.pruning = PrunedCapsuleLayer(
            theta=prune_theta,
            fg_mode=use_fg_routing,
            fg_thresh=fg_prune_thresh,
            min_keep_frac=prune_min_keep_frac,
            similarity_mode=similarity_mode,
        )
        self.attention_routing = SimplifiedAttentionRouting(
            D, alpha=alpha, n_iter=n_iter,
            fg_mode=use_fg_routing,
            fg_beta=fg_temp_beta,
        )
        self.flatcaps = nn.Linear(K * D, num_classes * D)

        if use_fg_routing and relevance_mode == "learned":
            self.relevance_head = RelevanceHead(1, hidden=relevance_hidden)
        else:
            self.relevance_head = None

    def forward(self, x, return_diagnostics=False, return_relevance=False):
        feats = self.relu(self.conv1(x))
        prim, h, w = self.primary(feats)
        B = x.size(0)

        relevance = None
        if self.use_fg_routing:
            if self.relevance_mode == "neutral":
                relevance = torch.full((B, h, w), 0.5, device=x.device)
            elif self.relevance_mode == "learned":
                prim_norm = prim.norm(dim=2).mean(dim=1, keepdim=True)
                relevance = self.relevance_head(prim_norm)

        keep_mask, capsule_relevance = self.pruning(prim, relevance=relevance)

        prim_sq = squash(prim.permute(0, 1, 3, 4, 2), axis=-1) \
                       .permute(0, 1, 4, 2, 3)
        prim_sq = prim_sq * keep_mask.view(B, -1, 1, 1, 1)

        routed = self.attention_routing(prim_sq, keep_mask, relevance=relevance)

        routed_pool = routed.mean(dim=(3, 4))
        flat = routed_pool.reshape(B, self.K * self.D)
        class_caps = self.flatcaps(flat).view(B, self.num_classes, self.D)
        lengths = class_caps.norm(dim=-1)

        if return_relevance:
            return lengths, relevance, capsule_relevance

        if return_diagnostics:
            diag = {
                "grid_h": h, "grid_w": w,
                "num_retained_types": keep_mask.sum(dim=1).mean().item(),
                "class_norm_mean": lengths.mean().item(),
                "class_norm_std": lengths.std().item(),
                "beta": self.attention_routing.fg_beta_current,
                "fg_thresh": self.pruning.fg_thresh_current,
            }
            if relevance is not None:
                diag["fg_mean"] = relevance.mean().item()
                diag["fg_std"] = relevance.std().item()
            if capsule_relevance is not None:
                diag["caps_rel_mean"] = capsule_relevance.mean().item()
                diag["caps_rel_std"] = capsule_relevance.std().item()
            return lengths, diag
        return lengths


# ------------------------------------------------------------
# Parameter-count sweep
# ------------------------------------------------------------
def count_params_for_config(K, D, conv1_channels, primary_stride=1,
                            num_classes=10):
    m = OrthCapsS(
        num_classes=num_classes,
        K=K, D=D,
        conv1_channels=conv1_channels,
        primary_stride=primary_stride,
    )
    n = sum(p.numel() for p in m.parameters())
    del m
    return n


def param_count_sweep(paper_target=105_500):
    print("\n" + "=" * 82)
    print(f"PARAMETER-COUNT SWEEP   (paper's OrthCaps-S = {paper_target:,})")
    print("=" * 82)
    print(f"{'conv1_ch':>9} | {'K':>4} | {'D':>4} | {'params':>14} | "
          f"{'vs paper':>10}")
    print("-" * 82)

    rows = []
    for conv1_ch in (8, 16, 32, 64, 128, 256):
        for K in (4, 8, 16, 32):
            for D in (8, 16):
                n = count_params_for_config(K, D, conv1_ch,
                                            primary_stride=1)
                ratio = n / paper_target
                rows.append({
                    "conv1_ch": conv1_ch, "K": K, "D": D,
                    "params": n, "ratio": ratio,
                })
                print(f"{conv1_ch:>9} | {K:>4} | {D:>4} | {n:>14,} | "
                      f"{ratio:>9.2f}x")

    rows.sort(key=lambda r: abs(r["params"] - paper_target))
    print("-" * 82)
    print("TOP 3 CLOSEST TO PAPER TARGET:")
    for i, r in enumerate(rows[:3], 1):
        print(f"  {i}. conv1_ch={r['conv1_ch']:>3}, K={r['K']:>2}, "
              f"D={r['D']:>2}  ->  {r['params']:>10,} params  "
              f"({r['ratio']:.2f}x paper)")
    return rows


# ------------------------------------------------------------
# Theta calibration
# ------------------------------------------------------------
def calibrate_theta(model, sample_x,
                    thetas=(1.0, 0.95, 0.9, 0.85, 0.8, 0.75, 0.7, 0.6, 0.5)):
    model.eval()
    rows = []
    with torch.no_grad():
        feats = model.relu(model.conv1(sample_x))
        prim, h, w = model.primary(feats)
        K = model.K
        for mode in ("flatten", "pooled"):
            probe = PrunedCapsuleLayer(theta=1.0, min_keep_frac=0.0,
                                       similarity_mode=mode).to(sample_x.device)
            for theta in thetas:
                probe.set_theta(theta)
                keep_mask, _ = probe(prim)
                kept = keep_mask.sum(dim=1)
                rows.append({
                    "similarity_mode": mode,
                    "theta": theta,
                    "kept_mean": kept.mean().item(),
                    "kept_min": kept.min().item(),
                    "kept_max": kept.max().item(),
                    "kept_frac_mean": (kept.mean() / K).item(),
                })
    df = pd.DataFrame(rows)
    print(f"\n[calibrate_theta] grid={h}x{w}  K={K}  D={model.D}  "
          f"(at initialization -- NOT a trained network)")
    print(df.to_string(index=False))
    return df


# ------------------------------------------------------------
# Data
# ------------------------------------------------------------
def load_cifar10(data_dir, batch_size):
    train_tf = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
    ])
    test_tf = transforms.Compose([transforms.ToTensor()])

    train_set = datasets.CIFAR10(data_dir, train=True, download=True, transform=train_tf)
    test_set = datasets.CIFAR10(data_dir, train=False, download=True, transform=test_tf)

    n_total = len(train_set)
    train_len = n_total - 5000
    gen = torch.Generator().manual_seed(42)
    train_sub, val_sub = random_split(range(n_total), [train_len, 5000], generator=gen)
    train_ds = Subset(train_set, train_sub.indices)
    val_ds = Subset(train_set, val_sub.indices)

    return (
        DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True),
        DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True),
        DataLoader(test_set, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True),
    )


# ------------------------------------------------------------
# Loss + scheduler
# ------------------------------------------------------------
def margin_loss(y_true, y_pred, m_plus=0.9, m_minus=0.1, lam=0.5):
    pos = torch.clamp(m_plus - y_pred, min=0.0) ** 2
    neg = torch.clamp(y_pred - m_minus, min=0.0) ** 2
    return (y_true * pos + lam * (1.0 - y_true) * neg).sum(dim=1).mean()


def make_scheduler(optimizer, base_lr, warmup_epochs, total_epochs, lr_floor):
    def lr_lambda(epoch):
        if epoch < warmup_epochs:
            return (epoch + 1) / warmup_epochs
        progress = (epoch - warmup_epochs) / max(1, total_epochs - warmup_epochs)
        return (lr_floor / base_lr) + (1 - lr_floor / base_lr) * 0.5 * (1 + math.cos(math.pi * progress))
    return torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)


def evaluate(model, loader, num_classes):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            y_onehot = F.one_hot(y, num_classes).float()
            y_pred = model(x)
            loss = margin_loss(y_onehot, y_pred)
            total_loss += loss.item() * x.size(0)
            correct += (y_pred.argmax(1) == y).sum().item()
            total += x.size(0)
    return total_loss / total, correct / total


def get_theta(epoch, config):
    if not config["pruning_enabled"]:
        return 1.0
    if epoch < config["prune_anneal_start"]:
        return config["prune_theta_start"]
    t = (epoch - config["prune_anneal_start"]) / max(
        1, config["epochs"] - config["prune_anneal_start"])
    return config["prune_theta_start"] + t * (
        config["prune_theta_final"] - config["prune_theta_start"])


def fg_anneal(epoch, warmup, target):
    if warmup <= 0:
        return target
    frac = min(1.0, (epoch + 1) / warmup)
    return target * (0.1 + 0.9 * frac)


# ------------------------------------------------------------
# Train one config
# ------------------------------------------------------------
def train_one_config(seed, config, train_loader, val_loader, test_loader, num_classes):
    set_seed(seed)
    conv1_channels = CAPACITY_CHANNELS[config["capacity_mode"]]
    model = OrthCapsS(
        num_classes=num_classes,
        K=config["K"], D=config["D"],
        conv1_channels=conv1_channels,
        primary_stride=config["primary_stride"],
        prune_theta=config["prune_theta_start"],
        alpha=config["alpha"],
        n_iter=config["entmax_n_iter"],
        prune_min_keep_frac=config["prune_min_keep_frac"],
        similarity_mode=config["similarity_mode"],
        use_fg_routing=USE_FG_ROUTING,
        relevance_mode=config["relevance_mode"],
        relevance_hidden=config["relevance_hidden"],
        fg_prune_thresh=config["fg_prune_thresh"],
        fg_temp_beta=config["fg_temp_beta"],
    ).to(device)

    n_params = sum(p.numel() for p in model.parameters())
    print(f"\nModel parameters: {n_params:,}  "
          f"(capacity_mode={config['capacity_mode']}, "
          f"conv1_channels={conv1_channels}, "
          f"primary_stride={config['primary_stride']})")

    optimizer = torch.optim.AdamW(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    scheduler = make_scheduler(optimizer, config["lr"], config["warmup_epochs"],
                               config["epochs"], config["lr_floor"])

    prune_tag = ("noprune" if not config["pruning_enabled"]
                 else f"prune{config['prune_theta_final']}")
    tag = (f"{ACTIVE_PRESET}_{config['capacity_mode']}_{prune_tag}_"
           f"{config['similarity_mode']}")
    if USE_FG_ROUTING:
        tag = f"fgosr_{config['relevance_mode']}_{tag}"
    model_name = f"{tag}_seed{seed}"

    history = []
    best_val_acc = -1.0
    best_state = None

    for epoch in range(config["epochs"]):
        theta = get_theta(epoch, config)
        model.pruning.set_theta(theta)

        if config["pruning_enabled"] and epoch == config["prune_anneal_start"]:
            print(f"\n[mid-train calibration] epoch {epoch+1}: about to "
                  f"start annealing theta. Checking actual redundancy in the "
                  f"TRAINED model (not at init) before it bites:")
            model.eval()
            with torch.no_grad():
                xb_cal, _ = next(iter(val_loader))
                calibrate_theta(model, xb_cal.to(device))
            model.train()
            print()

        if USE_FG_ROUTING:
            cur_thresh = fg_anneal(epoch, config["fg_prune_warmup"],
                                   config["fg_prune_thresh"])
            cur_beta = fg_anneal(epoch, config["fg_temp_warmup"],
                                 config["fg_temp_beta"])
            model.pruning.set_fg_thresh(cur_thresh)
            model.attention_routing.set_beta(cur_beta)
        else:
            cur_thresh, cur_beta = 0.0, 0.0

        model.train()
        t0 = time.time()
        epoch_loss_sum, total = 0.0, 0
        current_lr = optimizer.param_groups[0]["lr"]

        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            y_onehot = F.one_hot(y, num_classes).float()
            optimizer.zero_grad()

            if USE_FG_ROUTING and config["lam_relevance_sparse"] > 0:
                y_pred, relevance, _ = model(x, return_relevance=True)
                loss = margin_loss(y_onehot, y_pred)
                loss = loss + config["lam_relevance_sparse"] * relevance.mean()
            else:
                y_pred = model(x)
                loss = margin_loss(y_onehot, y_pred)

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(),
                                           max_norm=config["grad_clip"])
            optimizer.step()
            epoch_loss_sum += loss.item() * x.size(0)
            total += x.size(0)

        scheduler.step()
        train_loss = epoch_loss_sum / total
        val_loss, val_acc = evaluate(model, val_loader, num_classes)
        epoch_time = time.time() - t0

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state = {k: v.detach().cpu().clone()
                          for k, v in model.state_dict().items()}

        model.eval()
        with torch.no_grad():
            xb, _ = next(iter(val_loader))
            xb = xb.to(device)
            _, diag = model(xb, return_diagnostics=True)

        row = {
            "model": model_name, "epoch": epoch + 1,
            "train_loss": train_loss, "val_loss": val_loss, "val_acc": val_acc,
            "theta": theta,
            "kept": diag["num_retained_types"],
            "cn_mean": diag["class_norm_mean"],
            "cn_std": diag["class_norm_std"],
            "beta": cur_beta, "fg_thresh": cur_thresh,
            "fg_mean": diag.get("fg_mean", float("nan")),
            "fg_std": diag.get("fg_std", float("nan")),
            "epoch_time_sec": epoch_time, "lr": current_lr, "seed": seed,
        }
        history.append(row)

        print(f"  {model_name} | ep {epoch+1:3d}/{config['epochs']} | "
              f"lr={current_lr:.2e} | theta={theta:.3f} | "
              f"trL={train_loss:.4f} | vL={val_loss:.4f} | vA={val_acc:.4f} | "
              f"best={best_val_acc:.4f} | "
              f"kept={row['kept']:.1f}/{config['K']} | "
              f"cn={row['cn_mean']:.2f}+-{row['cn_std']:.2f} | "
              f"{epoch_time:.0f}s")

    model.load_state_dict(best_state)
    test_loss, test_acc = evaluate(model, test_loader, num_classes)

    summary = {
        "seed": seed,
        "mode": tag,
        "params": n_params,
        "best_val_acc": best_val_acc,
        "best_epoch": int(np.argmax([h["val_acc"] for h in history]) + 1),
        "test_loss": test_loss,
        "test_acc": test_acc,
        "mean_epoch_time_sec": float(np.mean([h["epoch_time_sec"] for h in history])),
    }
    return pd.DataFrame(history), summary


# ------------------------------------------------------------
# Main
# ------------------------------------------------------------
def main():
    print("Loading CIFAR-10...")
    train_loader, val_loader, test_loader = load_cifar10(
        CONFIG["data_dir"], CONFIG["batch_size"])

    # ---- 1. parameter-count sweep (CPU only, ~30 seconds) ----
    param_count_sweep()

    # ---- 2. print active preset ----
    print(f"\n{'=' * 70}")
    print(f"ACTIVE PRESET: {ACTIVE_PRESET}")
    print(f"  {PRESETS[ACTIVE_PRESET]['note']}")
    print(f"{'=' * 70}")

    conv1_channels = CAPACITY_CHANNELS[CONFIG["capacity_mode"]]
    prune_tag = ("noprune" if not CONFIG["pruning_enabled"]
                 else f"prune{CONFIG['prune_theta_final']}")
    tag = (f"{ACTIVE_PRESET}_{CONFIG['capacity_mode']}_{prune_tag}_"
           f"{CONFIG['similarity_mode']}")
    if USE_FG_ROUTING:
        tag = f"fgosr_{CONFIG['relevance_mode']}_{tag}"

    print(f"\nRUN: {tag}")
    print(f"  epochs={CONFIG['epochs']} | lr={CONFIG['lr']} | "
          f"bs={CONFIG['batch_size']}")
    print(f"  D={CONFIG['D']} | K={CONFIG['K']} | "
          f"conv1_channels={conv1_channels} | "
          f"primary_stride={CONFIG['primary_stride']}")
    print(f"  pruning_enabled={CONFIG['pruning_enabled']} | "
          f"theta {CONFIG['prune_theta_start']} -> "
          f"{CONFIG['prune_theta_final']} starting ep "
          f"{CONFIG['prune_anneal_start']} | "
          f"similarity_mode={CONFIG['similarity_mode']} | "
          f"min_keep_frac={CONFIG['prune_min_keep_frac']}")
    print(f"  use_fg_routing={USE_FG_ROUTING}")
    print("=" * 70)

    if CONFIG["run_calibration"]:
        set_seed(CONFIG["seeds"][0])
        probe_model = OrthCapsS(
            K=CONFIG["K"], D=CONFIG["D"],
            conv1_channels=conv1_channels,
            primary_stride=CONFIG["primary_stride"],
        ).to(device)
        xb, _ = next(iter(val_loader))
        calibrate_theta(probe_model, xb.to(device))
        del probe_model

    all_history, all_summaries = [], []
    for seed in CONFIG["seeds"]:
        print(f"\n{'=' * 70}\nTraining [{tag}] | seed={seed}\n{'=' * 70}")
        h_df, summary = train_one_config(
            seed, CONFIG, train_loader, val_loader, test_loader,
            num_classes=10,
        )
        all_history.append(h_df)
        all_summaries.append(summary)

    history_df = pd.concat(all_history, ignore_index=True)
    summary_df = pd.DataFrame(all_summaries)

    history_df.to_csv(f"cifar10_{tag}_history.csv", index=False)
    summary_df.to_csv(f"cifar10_{tag}_summary.csv", index=False)

    print("\n" + "=" * 70)
    print("FINAL SUMMARY")
    print("=" * 70)
    print(summary_df.to_string(index=False))
    if len(summary_df) > 1:
        print(f"\ntest_acc mean={summary_df['test_acc'].mean():.4f}  "
              f"sd={summary_df['test_acc'].std():.4f}  "
              f"(n={len(summary_df)} seeds)")

    fig, axes = plt.subplots(1, 4, figsize=(20, 4.5))
    for model_name, grp in history_df.groupby("model"):
        axes[0].plot(grp["epoch"], grp["train_loss"], label=model_name)
        axes[1].plot(grp["epoch"], grp["val_loss"], label=model_name)
        axes[2].plot(grp["epoch"], grp["val_acc"], label=model_name)
        axes[3].plot(grp["epoch"], grp["kept"], label=model_name)
    axes[0].set_title("Train loss")
    axes[1].set_title("Val loss")
    axes[2].set_title("Val accuracy")
    axes[3].set_title(f"Retained types (theta target="
                      f"{CONFIG['prune_theta_final']}, "
                      f"floor={CONFIG['prune_min_keep_frac'] * CONFIG['K']:.0f})")
    for ax in axes:
        ax.set_xlabel("Epoch"); ax.grid(True, alpha=0.3)
        ax.legend(fontsize=8)
    plt.tight_layout()
    plt.savefig(f"cifar10_{tag}_curves.png", dpi=150)
    plt.close()
    print(f"\nSaved: cifar10_{tag}_history.csv, _summary.csv, _curves.png")


if __name__ == "__main__":
    main()
```
